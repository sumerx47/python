```python
'''1. Calculate each student's average marks.
Do not use a hardcoded number of subjects.'''

'''2. Using a dictionary comprehension, create:
{
    "Ayaan": 82.0,
    "Sara": 92.6,
    ...
}'''

'''3. Using a list comprehension, create a list containing the names of students who:
- are active,
- have attendance of at least 80%, and
- have an average of at least 75.'''

'''4. Write another function:
def get_grade(avg):
that returns:
90+      → A
80–89.99 → B
70–79.99 → C
60–69.99 → D
Below 60 → F'''

'''5. Using a dictionary comprehension and get_grade(), create:
{
    "Ayaan": "B",
    "Sara": "A",
    "Zaid": "F",
    "Maryam": "B"
}'''

'''6. Use a lambda function with sorted() to sort the students from highest to lowest average marks.
The result should contain tuples:
[
    ("Sara", 92.6),
    ("Ayaan", 82.0),
    ...
]'''

'''7. Using a set comprehension, find all unique skills possessed by students who have an average of at least 75.
Expected structure:
{"Python", "SQL", "Excel", "PowerBI", "Java"}
'''

'''8. Find the skills that are common to all qualifying students from 
Question 3. Use appropriate set operations rather than loops where possible.'''


def get_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 80:
        return "B"
    elif avg >= 70:
        return "C"
    elif avg >= 60:
        return "D"
    else :
        return "F" 

def calculate_avg(score):
    return sum(score)/len(score)


def analyze_students(students):
    result = [calculate_avg(student["marks"]) for student in students]

    result1 = {student["name"]:calculate_avg(student["marks"]) for student in students}

    result2 = [student["name"] for student in students if student["active"] == True and student["attendance"] >= 80 and calculate_avg(student["marks"]) >= 75]

    result3 = [get_grade(calculate_avg(student["marks"])) for student in students ]

    result4 = {student["name"]:get_grade(calculate_avg(student["marks"])) for student in students}

    result5 = sorted([(student["name"],calculate_avg(student["marks"])) for student in students], key = lambda student:student[1], reverse = True)

    result6 = [skill for student in students if calculate_avg(student["marks"]) >= 75 for skill in student["skills"]]

    common_skills = [student for student in students if student["active"] == True and student["attendance"] >= 80 and calculate_avg(student["marks"]) >= 75]
    result7 = set.intersection(*(student["skills"] for student in common_skills))

    return {
        "calculate avg": result, 
        "name,avg": result1,
        "active": result2,
        "ABCDEF" : result3,
        "name,grade" : result4,
        "lambda": result5,
        "unique" : result6,
        "Q3,skills":result7
    }


result = analyze_students(students)
print("1. calculate avg of students: ", result["calculate avg"])
print("2. name & avg: ", result["name,avg"])
print("3. students who are active, attendance >= 80% and average >= 75: ", result["active"])
print("4. ABCDEF: ", result["ABCDEF"])
print("5. name & grade: ", result["name,grade"])
print("6. students from highest to lowest average marks: ", result["lambda"])
print("7. unique skills of students with average >= 75: ", set(result["unique"]))
print("8. skills that are common to all qualifying students: ", result["Q3,skills"])



```
