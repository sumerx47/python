```python
m = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
```

```python
n=5
matrix=[]

for row_index in range(n):
    row=[]
    for col_index in range(n):
        if row_index == col_index:
            row.append(1)
        else:
            row.append(0)
    matrix.append(row)
```
