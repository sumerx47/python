#  Q1

```python
text="DevOps Python 2026!"
```

```python
upper=0
lower=0
spaces =0
digits=0
Remaining=0
```

```python
for i in text: 
    if i.isupper():
        upper += 1

    elif i.islower():
        lower += 1

    elif i.isspace():
        spaces+=1

    elif i.isdigit():
        digits+=1
    else:
        Remaining +=1

print("Uppercase:", upper)
print("lowercase:", lower)
print("spaces:", spaces)
print("digits:", digits)
print("Remaning:", Remaining)
```
