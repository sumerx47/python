🔹 Pattern 8 (Grouping)

1. Group employee names by department.


2. Group employee names by city.


3. Group salaries by department.



🔹 Pattern 9 (Searching)

4. Find employee ID 105.


5. Find the first employee from Mumbai.


6. Find the first IT employee whose salary is greater than 70000.


7. Search employee named Rehan (before merging).



🔹 Pattern 10 (Update/Delete/Merge)

8. Increase every IT employee's salary by 5000.


9. Change every Finance employee's city to Hyderabad.


10. Delete employees whose salary is below 60000.


11. Merge new_employees.


12. Search employee Rehan again.


13. Group names by department again after merging.

```python
employees = [
    {"id":101,"name":"Ali","dept":"IT","city":"Hyderabad","salary":65000},
    {"id":102,"name":"Sara","dept":"HR","city":"Mumbai","salary":72000},
    {"id":103,"name":"Ahmed","dept":"IT","city":"Hyderabad","salary":58000},
    {"id":104,"name":"John","dept":"Finance","city":"Delhi","salary":45000},
    {"id":105,"name":"Fatima","dept":"IT","city":"Bangalore","salary":82000},
    {"id":106,"name":"Mary","dept":"HR","city":"Mumbai","salary":51000},
    {"id":107,"name":"Zaid","dept":"Finance","city":"Delhi","salary":69000}
]

new_employees = [
    {"id":108,"name":"Rehan","dept":"IT","city":"Pune","salary":76000},
    {"id":109,"name":"Imran","dept":"HR","city":"Mumbai","salary":68000}
]
```

```python

group_dept = {}
for emp in employees:
    dept = emp["dept"] 
    if dept not in group_dept:
        group_dept[dept] = []
    group_dept[dept].append(emp["name"])
print("Group employee names by department: ",group_dept)

city_group={}
for emp in employees:
    city = emp["city"]
    if city not in city_group:
        city_group[city] = []
    city_group[city].append(emp["name"])
print("Group employee names by city.: ",city_group)

group_salary ={}
for emp in employees:
    dept = emp["dept"]
    if dept not in group_salary:
        group_salary[dept] = []
    group_salary[dept].append(emp["salary"])
print("Group salaries by department: ",group_salary)

print("*"*15)
found=False
for emp in employees: 
    if emp["id"] == 105:
        print("employee ID 105: ",emp)
        found = True
if not found:
    print("NOT FOUND")
print("*"*15)

found=False
for emp in employees:
    if emp["city"] == "Mumbai":
        print("first employee from Mumbai: ",emp)
        found = True
        break
if not found:
    print("NOT FOUND")
print("*"*15)


found = False
for emp in employees:
    if emp["salary"]> 70000:
        print("first IT employee whose salary is greater than 70000: ",emp)
        found=True
        break
if not found:
    print("NOT FOUND")
print("*"*15)

found = False
for emp in employees:
    if emp["name"] == "Rehan":
        print(emp)
        found=True
        break
if not found:
    print("employee named Rehan: NOT FOUND")
print("*"*15)

for emp in employees:
    if emp["dept"] =="IT":
        emp["salary"] += 5000
        print("Increase every IT employee's salary by 5000: ",emp)
print("*"*15)

for emp in employees:
    if emp["dept"] =="Finance":
        emp["city"] = "Hyderabad"
        print("change every Finance employee's city to Hyderabad: ",emp)
print("*"*15)


low_salary =[]
for emp in employees:
    if emp["salary"] >= 60000:
        low_salary.append(emp)
print("Delete employees whose salary is below 60000: ",low_salary)
print("*"*15)

updated_employees = employees + new_employees
print(updated_employees)
print("*"*15)

found= False
for emp in updated_employees:
    if emp["name"] == "Rehan":
        print("Rehan Found: ",emp)
        found=True
        break
if not found:
    print("not found")
print("*"*15)


group_depart={}
for emp in updated_employees:
    dept = emp["dept"]
    if dept not in group_depart:
        group_depart[dept] = []
    group_depart[dept].append(emp["name"])
print(group_depart)

```

🔥 Advanced+ Question 2 – AWS Infrastructure Dashboard



Tasks

🔹 Pattern 8 (Grouping)

1. Group server names by team.


2. Group server IDs by status.


3. Group CPU values by team.



🔹 Pattern 9 (Searching)

4. Find server ID i-104.


5. Find the first Backend server.


6. Find the first Running Frontend server.


7. Search server monitor01 (before merging).



🔹 Pattern 10 (Update/Delete/Merge)

8. Change every Stopped server to Running.


9. Increase Backend server CPU by 10.


10. Delete servers with CPU below 30.


11. Merge new_servers.


12. Search monitor01 again.


13. Group server names by team after merging.


```python
servers = [
    {"id":"i-101","name":"web01","team":"Frontend","status":"Running","cpu":65},
    {"id":"i-102","name":"db01","team":"Database","status":"Stopped","cpu":20},
    {"id":"i-103","name":"app01","team":"Backend","status":"Running","cpu":80},
    {"id":"i-104","name":"cache01","team":"Backend","status":"Running","cpu":35},
    {"id":"i-105","name":"proxy01","team":"Frontend","status":"Running","cpu":75},
    {"id":"i-106","name":"backup01","team":"Database","status":"Stopped","cpu":10}
]

new_servers = [
    {"id":"i-107","name":"monitor01","team":"DevOps","status":"Running","cpu":50},
    {"id":"i-108","name":"jenkins01","team":"DevOps","status":"Running","cpu":70}
]
```

```python
group_team={}
for server in servers:
    group_team.setdefault(server["team"],[]).append(server["name"])
print(group_team)

group_status = {}
for server in servers:
    group_status.setdefault(server["status"],[]).append(server["id"])
print(group_status)

Group_CPU = {}
for server in servers:
    Group_CPU.setdefault(server["team"],[]).append(server["cpu"])
print(Group_CPU)
print("*" * 15)

search = "i-104"
result = next((server for server in servers if server["id"]==search), None)
if result:
    print(result)
else:
    print("Not found")
print("*" * 15)

search = "Backend"
result = next((server for server in servers if server["team"]==search), None)
if result:
    print(result)
else:
    print("Not found")
print("*" * 15)

search = "Frontend"
result = next((server for server in servers if server["team"]==search), None)
if result:
    print(result)
else:
    print("Not found")
print("*" * 15)

search = "monitor01"
result = next((server for server in servers if server["name"]==search),None)
if result:
    print(result)
else:
    print("Not found")
print("*" * 15)


for server in servers:
    if server["status"] == "Stopped": 
        server["status"] = "Running"
        break
print(servers)
print("*" * 15)

for server in servers:
    if server["team"] == "Backend": 
        server["cpu"] += 10
        break
print(servers)
print("*" * 15)

del_list=[]
for server in servers:
    if server["cpu"] >= 30:
        del_list.append(server)
print("Delete servers: ",del_list)
print("*" * 15)

updated_servers= servers + new_servers
print(updated_servers)
print("*" * 15)


search = "monitor01"
result = next((server for server in updated_servers if server["name"]==search), None)
if result:
    print(result)
else:
    print("Not found")
print("*" * 15)

group_teams={}
for server in servers:
    group_teams.setdefault(server["team"],[]).append(server["name"])
print(group_teams)


```
