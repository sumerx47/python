Question 1 – Company Performance Report

Tasks

1. Find the employee with the highest salary.


2. Find the employee with the lowest salary.


3. Count employees in each department.


4. Print IT employees with salary greater than ₹60,000.


5. Print employees whose experience is at least 5 years.


6. Create a new list containing names of employees earning at least ₹60,000.


7. Calculate the total salary of all employees.


8. Calculate the average salary.


9. Sort employees by salary (descending).


10. Sort employees alphabetically by name.


11. Print employees who are:

IT department

Salary ≥ ₹60,000

Experience ≥ 4



12. Create a new list containing names of employees satisfying Task 11.

```python
employees = [
    {"name":"Ali","dept":"IT","salary":65000,"exp":4},
    {"name":"Sara","dept":"HR","salary":72000,"exp":6},
    {"name":"Ahmed","dept":"IT","salary":58000,"exp":3},
    {"name":"John","dept":"Finance","salary":45000,"exp":8},
    {"name":"Fatima","dept":"IT","salary":82000,"exp":7},
    {"name":"Mary","dept":"HR","salary":51000,"exp":2},
    {"name":"Zaid","dept":"Finance","salary":69000,"exp":5}
]
```

```python
highest_salary= employees[0]
lowest_salary = employees[0]

for emp in employees:
    if emp["salary"] > highest_salary["salary"]:
        highest_salary = emp
    if emp["salary"] < lowest_salary["salary"]:
        lowest_salary = emp

print("employee with the highest salary: ",highest_salary)
print("employee with the lowest salary: ",lowest_salary)

emp_count = {}
for emp in employees:
    dept = emp["dept"]
    if dept in emp_count:
        emp_count[dept] += 1
    else:
        emp_count[dept] = 1
    
print("employees in each department: ",emp_count)

print("*"*15)
for emp in employees:
    if emp["dept"] == "IT" and emp["salary"]>60000:
        print("IT employees with salary greater than ₹60,000: ",emp["name"])
print("*"*15)
for emp in employees:
    if emp["exp"] >= 5:
        print("employees whose experience is at least 5 years: ",emp["name"])
print("*"*15)

emp_sal=[]
for emp in employees:
    if emp["salary"] >= 60000:
        emp_sal.append(emp["name"])

print("employees earning at least ₹60,000: ",emp_sal)
print("*"*15)


total_sal=0
for emp in employees:
    total_sal += emp["salary"]
average_salary = total_sal / len(employees)

print("total salary of all employees: ",total_sal)
print("average salary: ",average_salary)
print("*"*15)


print("employees by salary (descending): ",sorted(employees, key = lambda emp: emp["salary"], reverse=True))
print("employees alphabetically by name(ascending): ",sorted(employees, key = lambda emp:emp["name"]))

print("*"*15)
emp_exp=[]
for emp in employees:
    if emp["dept"] == "IT" and emp["salary"] >= 60000 and emp["exp"] >= 4:
        print("IT employees with Salary ≥ ₹60,000 and Experience ≥ 4: ",emp["name"])
        emp_exp.append(emp["name"])

print("*"*15)
print("List of IT employees with Salary ≥ ₹60,000 and Experience ≥ 4: ",emp_exp)
```

🔥 Advanced+ Question 2 – DevOps Infrastructure Report


Tasks

1. Find the server with the highest CPU.


2. Find the server with the lowest CPU.


3. Count Running and Stopped servers.


4. Print Running servers.


5. Print Running servers with CPU greater than 60.


6. Create a list containing names of Running servers.


7. Create a list containing names of Running servers with memory greater than 70.


8. Calculate the total CPU usage.


9. Calculate the average CPU usage.


10. Sort by CPU (highest first).


11. Sort alphabetically by server name.


12. Print servers that are:

Running

CPU ≥ 60

Memory ≥ 70



13. Create a list containing names of servers satisfying Task 12.



```python
servers = [
    {"name":"web01","status":"Running","cpu":65,"memory":70},
    {"name":"db01","status":"Stopped","cpu":0,"memory":40},
    {"name":"app01","status":"Running","cpu":80,"memory":90},
    {"name":"cache01","status":"Running","cpu":35,"memory":30},
    {"name":"monitor01","status":"Running","cpu":55,"memory":60},
    {"name":"backup01","status":"Stopped","cpu":0,"memory":50},
    {"name":"proxy01","status":"Running","cpu":75,"memory":85}
]
```

```python
highest_CPU = max(servers, key = lambda x:x["cpu"])
lowest_CPU = min(servers, key = lambda x:x["cpu"])

print("server with the highest CPU: ",highest_CPU)
print("server with the lowest CPU: ",lowest_CPU)

server_count={}
for server in servers:
    status = server["status"]
    server_count[status] = server_count.get(status,0)+1
print("Count of Running and Stopped servers: ",server_count)

print("*"*15)
for server in servers:
    if server["status"] == "Running":
      print("Running servers are: ",server["name"])

print("*"*15)
for server in servers:
    if server["status"] == "Running" and server["cpu"] >= 60:
      print("Running servers with CPU greater than 60: ",server["name"])
print("*"*15)

Running_servers=[]
Running_servers_memory=[]
for server in servers:
    if server["status"] == "Running":
      Running_servers.append(server["name"])
    if server["status"] == "Running" and server["memory"]>70:
          Running_servers_memory.append(server["name"])

print("names of Running servers: ",Running_servers)
print("Running servers with memory greater than 70: ",Running_servers_memory)
print("*"*15)
total_CPU = sum(server["cpu"] for server in servers)
average_CPU = total_CPU / len(servers)

print("Total CPU usage.: ",total_CPU)
print("Average CPU usage: ",average_CPU)
print("*"*15)


server = servers.copy()
server.sort(key = lambda x:x["cpu"], reverse=True)
print("Sort by CPU (highest first): ",server)
print("Sort alphabetically by server name: ",[sorted(servers, key = lambda x:x["name"])])
print("*"*15)

server_config=[]
for server in servers:
    if server["status"] == "Running" and server["cpu"] >= 60 and server["memory"] >= 70:
        print("servers that are  Running with CPU ≥ 60 and Memory ≥ 70: ", server["name"])
        server_config.append(server["name"])
print("*"*15)
print("server configuration: ",server_config)
print("*"*15)

```
