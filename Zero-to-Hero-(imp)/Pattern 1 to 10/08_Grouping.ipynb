# 8. Multiple Conditions
## Intermediate to Advanced

```python
employees = [
    {"name":"Ali","dept":"IT","salary":65000,"exp":4},
    {"name":"Sara","dept":"HR","salary":72000,"exp":6},
    {"name":"Ahmed","dept":"IT","salary":58000,"exp":3},
    {"name":"John","dept":"Finance","salary":45000,"exp":8},
    {"name":"Fatima","dept":"IT","salary":82000,"exp":7},
    {"name":"Mary","dept":"HR","salary":51000,"exp":2},
    {"name":"Zaid","dept":"Finance","salary":69000,"exp":5}
]
```

### method -1

```python
group={}

for emp in employees:
    dept = emp["dept"]
    if dept not in group:
        group[dept] = []
    group[dept].append(emp["name"])

print(group)

```

### method -2

```python
group = {}
for emp in employees:
    group.setdefault(emp["dept"],[]).append(emp["name"])
print(group)
```

Question 9 – Company Employees


Tasks

1. Group employee names by department.


2. Group salaries by department.

```python
employees = [
    {"name":"Ali","dept":"IT","salary":50000},
    {"name":"Sara","dept":"HR","salary":70000},
    {"name":"Ahmed","dept":"IT","salary":65000},
    {"name":"John","dept":"Finance","salary":45000},
    {"name":"Fatima","dept":"HR","salary":80000},
    {"name":"Mary","dept":"IT","salary":72000}
]
```

```python
group={}
for emp in employees:
    group.setdefault(emp["dept"],[]).append(emp["name"])
print(group)

salary_group = {}
for emp in employees:
    dept=emp["dept"]
    if dept not in salary_group:
        salary_group[dept] = []
    salary_group[dept].append(emp["salary"])
print(salary_group)
```

Question 13 – Group by Two Fields


Tasks

1. Group employee names by department.


2. Group employee names by city.

```python
employees = [
    {"name":"Ali","dept":"IT","city":"Hyderabad"},
    {"name":"Sara","dept":"HR","city":"Mumbai"},
    {"name":"Ahmed","dept":"IT","city":"Hyderabad"},
    {"name":"John","dept":"Finance","city":"Delhi"},
    {"name":"Fatima","dept":"HR","city":"Mumbai"},
    {"name":"Mary","dept":"IT","city":"Bangalore"}
]
```

```python
dept_group={}
for emp in employees:
    dept = emp["dept"]
    if dept not in dept_group:
        dept_group[dept] = []
    dept_group[dept].append(emp["name"])
print(dept_group)

city_group = {}
for emp in employees:
    city_group.setdefault(emp["city"],[]).append(emp["name"])
print(city_group)
```
