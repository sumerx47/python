Question 1 — Employee Analysis

Create functions using list comprehensions to:

1. Get all employee names.


2. Get employees earning ≥ 75000.


3. Get names of employees who know Python.


4. Get all skills from all employees.


5. Get unique skills.


6. Get (name, salary) pairs for employees earning > 70000.


7. Get names sorted by salary using sorted() + lambda.



```python
employees = {
    "E101": {
        "name": "Ali",
        "skills": ["Python", "Linux"],
        "salary": 60000
    },
    "E102": {
        "name": "Sara",
        "skills": ["Python", "Docker"],
        "salary": 85000
    },
    "E103": {
        "name": "Ahmed",
        "skills": ["Linux", "Docker", "Kubernetes"],
        "salary": 95000
    },
    "E104": {
        "name": "Zoya",
        "skills": ["Python", "AWS"],
        "salary": 75000
    }
}
```

```python
# with fubctions 

# 1. Get all employee names.
def all_employee_names():
    return [details["name"] for details in employees.values()]
result = all_employee_names()
print("All employee names: ", result)
print("*" * 15)

# 2. Get employees earning ≥ 85000.
def employees_earning():
    return [details["name"] for details in employees.values() if details["salary"] >= 85000]
result1 = employees_earning()
print("employees earning ≥ 85000: ", result1)
print("*" * 15)

# 3. Get names of employees who know Python.
def employees_Python():
    return [details["name"] for details in employees.values() if "Python" in details["skills"]]
result2 = employees_Python()
print("employees who know Python: ", result2)
print("*" * 15)

# 4. Get all skills from all employees.
def all_skills():
    return [skill for details in employees.values() for skill in details["skills"]]
result3 = all_skills()
print("all skills: ", result3)
print("*" * 15)

# 5. Get unique skills.
def unique_skills():
    return [skill for details in employees.values() for skill in details["skills"]]
result3 = unique_skills()
print("unique skills: ", list(set(result3)))
print("*" * 15)

# 6. Get (name, salary) pairs for employees earning > 70000.
def pairs_employees():
    return [(details["name"],details["salary"]) for details in employees.values() if details["salary"] > 70000]
result4 = pairs_employees()
print("(name, salary) pairs for employees earning > 70000:",result4)
print("*" * 15)


# 7. Get names sorted by salary using sorted() + lambda.
def sorted_salary():
    return [(details["name"],details["salary"]) for details in sorted(employees.values(), key = lambda x:x["salary"])]
result5 = sorted_salary()
print("names sorted by salary using sorted() + lambda: ",result5)
print("*" * 15)
```
