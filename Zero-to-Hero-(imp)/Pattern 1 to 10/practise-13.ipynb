3. Get names of employees who know Python.

5. Get unique skills.


6. Get (name, salary) pairs for employees earning > 70000.

```python
employees = {
    "E101": {
        "name": "Ali",
        "skills": ["Python", "Linux"],
        "salary": 60000
    },
    "E102": {
        "name": "Sara",
        "skills": ["Python", "Docker"],
        "salary": 85000
    },
    "E103": {
        "name": "Ahmed",
        "skills": ["Linux", "Docker", "Kubernetes"],
        "salary": 95000
    },
    "E104": {
        "name": "Zoya",
        "skills": ["Python", "AWS"],
        "salary": 75000
    }
}
```
