Q1. Function to find highest number

```python
def find_highest(numbers):
    highest = numbers[0]
    for num in numbers:
        if num > highest:
            highest = num

    return highest

print(find_highest([10,40,20,80,30]))
```

Q2. Find lowest number

```python
def  find_lowest(numbers):
    lowest = numbers[0]

    for num in numbers:
        if num < lowest:
            lowest = num

    return lowest

print(find_lowest([10,40,20,80,30,2]))
```

Q3. Count even numbers

```python
def find_count(numbers):
    count = 0
    for num in numbers:
        if num % 2 == 0 :
            count += 1
    return count
print(find_count([10,40,20,80,30,2,1,3,5,7,8]))

```

7. Count positive, negative and zero

```python
def count(numbers):
    positive = 0
    negative = 0 
    zero = 0
    for num in numbers:
        if num > 0:
            positive += 1
        elif num < 0:
            negative += 1
        else :
            zero += 1
    return positive, negative, zero
print(count([10,20,-1,-2,22,0,85,-55,0]))
```

Q8. Find the second highest number

```python
def second_largest(numbers):
    unique = list(set(numbers))
    unique.sort()

    return unique[-2]
print(second_largest([10,20,-1,-2,22,0,85,55,0]))
```

```python
def second_largests(numbers):
    first = float("-inf")
    second = float("-inf")

    for num in numbers:
        if num > first:
            second = first
            first = num
        elif first > num > second:
            second = num 
    return second

print(second_largests([10,20,-1,-2,22,0,85,55,0]))
```

Q26. Find the department with the highest average salary





```python
employees = [
    {"name": "Ali", "dept": "IT", "salary": 60000},
    {"name": "Ahmed", "dept": "IT", "salary": 80000},
    {"name": "Sara", "dept": "HR", "salary": 50000},
    {"name": "Zoya", "dept": "HR", "salary": 70000},
    {"name": "Sara", "dept": "IT", "salary": 50000},
    {"name": "Zoya", "dept": "IT", "salary": 70000}
]
```

```python
def highest_average_department(employees):
    departments = {}

    for emp in employees:
        dept = emp["dept"]
        if dept not in departments:
            departments[dept] = []
        departments[dept].append(emp["salary"])

    averages = {}
    for dept, salaries in departments.items():
        averages[dept] = sum(salaries) / len(salaries)

    return max(averages, key=averages.get)

print(highest_average_department(employees))
```


Q27. Find duplicate names


```python
def duplicate_names(employees):
    seen = set()
    duplicates = set()

    for emp in employees:
        name = emp["name"]
        if name in seen:
            duplicates.add(name)
        else:
            seen.add(name)

    return duplicates

print(duplicate_names(employees))

```
