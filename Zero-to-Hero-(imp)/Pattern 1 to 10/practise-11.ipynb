🟢 LEVEL 1 — BASIC

Q1. Function to find highest number

```python
def find_highest(numbers):
    highest = numbers[0]
    for num in numbers:
        if num > highest:
            highest = num

    return highest

print(find_highest([10,40,20,80,30]))
```

Q2. Find lowest number

```python
def  find_lowest(numbers):
    lowest = numbers[0]

    for num in numbers:
        if num < lowest:
            lowest = num

    return lowest

print(find_lowest([10,40,20,80,30,2]))
```

Q3. Count even numbers

```python
def find_count(numbers):
    count = 0
    for num in numbers:
        if num % 2 == 0 :
            count += 1
    return count
print(find_count([10,40,20,80,30,2,1,3,5,7,8]))

```


Q4. Filter numbers greater than 50

```python
def filter_num(numbers):
    result = []
    for num in numbers:
        if num > 50:
            result.append(num)

    return result
print(filter_num([10,40,20,80,30,2,1,3,5,7,8,99,120]))
```

Q5. Create squares

```python
def create_square(numbers):
    result=[]
    for num in numbers:
        result.append(num**2)
    return result
print(create_square([1,2,3,4,5]))
```

Q6. Sum and average


```python
def calculate_sum_avg(numbers):
    total = 0
    for num in numbers:
        total += num

    avg =  total / len(numbers)
    return total, avg
print(calculate_sum_avg([10,20,30,40]))

```

7. Count positive, negative and zero

```python
def count(numbers):
    positive = 0
    negative = 0 
    zero = 0
    for num in numbers:
        if num > 0:
            positive += 1
        elif num < 0:
            negative += 1
        else :
            zero += 1
    return positive, negative, zero
print(count([10,20,-1,-2,22,0,85,-55,0]))
```

Q8. Find the second highest number

```python
def second_largest(numbers):
    unique = list(set(numbers))
    unique.sort()

    return unique[-2]
print(second_largest([10,20,-1,-2,22,0,85,55,0]))
```

```python
def second_largests(numbers):
    first = float("-inf")
    second = float("-inf")

    for num in numbers:
        if num > first:
            second = first
            first = num
        elif first > num > second:
            second = num 
    return second

print(second_largests([10,20,-1,-2,22,0,85,55,0]))
```

Q9. Remove duplicates

```python
def remove_dup(numbers):
    result = []

    for num in numbers:
        if num not in result:
            result.append(num)
    return result

print(remove_dup([10,22,20,20,-1,-2,22,0,85,55,0]))
```

10. Count frequency

```python
def count_freq(numbers):
    result = {}
    for num in numbers:
        result[num] = result.get(num,0)+1

    return result

print(count_freq([10,22,20,20,-1,-2,22,0,85,55,0]))

```

Q11. Find highest-paid employee



```python
employees = [
    {"name": "Ali", "salary": 60000},
    {"name": "Ahmed", "salary": 80000},
    {"name": "Sara", "salary": 50000}
]
```

```python
def highest_paid(employees):
    highest = employees[0]

    for emp in employees:
        if emp["salary"] > highest["salary"]:
            highest = emp
    return highest

print(highest_paid(employees))
```

Q12. Filter employees with salary ≥ 60000

```python
def filter_employees(employees):
    result = []

    for emp in employees:
        if emp["salary"] >= 60000:
            result.append(emp)
    return result
print(filter_employees(employees))
```

Q13. Calculate average employee salary

```python
def calculate(employees):
    total = 0
    for emp in employees:
        total += emp["salary"]
    avg = total / len(employees)
    return total, avg

print(calculate(employees))
```

Q14. Group employees by department

```python
employees = [
    {"name": "Ali","dept": "IT" ,"salary": 60000},
    {"name": "Ahmed","dept": "IT" , "salary": 80000},
    {"name": "Sara","dept": "HR" , "salary": 50000}
]
```

```python
def group_dept(employees):
    group = {}

    for emp in employees:
        dept = emp["dept"]
        if dept not in group:
            group[dept] = []

        group[dept].append(emp["name"])

    return group
print(group_dept(employees))
```

Q15. Search employee by name

```python
def search_employee(employees, name):

    for emp in employees:
        if emp["name"] == name:
            return emp

    return None
print(search_employee(employees, "Sara"))
```
