🟢 Level 1 (Intermediate)

Question 1 – Highest Salary

Find: Highest salary employee

```python
employees = [
    {"name": "Ali", "salary": 45000},
    {"name": "Sara", "salary": 55000},
    {"name": "John", "salary": 65000},
    {"name": "Ahmed", "salary": 50000},
    {"name": "blake", "salary": 65000}
]
```

✅ Method 1: Manual loop (most important for interviews)

```python
highest_salary = employees[0]
```

```python
for employee in employees:
    if employee["salary"] > highest_salary["salary"]:
        highest_salary = employee
highest_salary
```

✅ Method 2: Using max()

```python
highest_salary = max(employees, key = lambda emp : emp["salary"])
highest_salary
```

✅ Method 3: Using sorted()

```python
sorted_employees = sorted(employees, key= lambda emp: emp["salary"], reverse =True)
sorted_employees[0]
```

✅ Method 4: Find all employees with the same highest salary

```python
highest_salary = max(emp["salary"] for emp in employees)

for emp in employees:
    if emp["salary"] == highest_salary:
        print(emp)
```

✅ Method 5: Professional interview approach

```python
highest_salary = max(employees, key=lambda emp:emp["salary"])
highest_salary["name"]
```

🟠 Level 3 (Advanced)

Question 11 – Highest Average

Find:  Student with highest average.

```python
students = [
    {
        "name":"Ali",
        "marks":[80,85,90]
    },
    {
        "name":"Sara",
        "marks":[95,90,98]
    },
    {
        "name":"Ahmed",
        "marks":[70,75,80]
    }
]
```

✅ Method 1: using loops (best for interviews)



```python
highest = students[0]
highest_avg = 0
```

```python
for student in students:
    total = 0
    for mark in student["marks"]:
        total+=mark
        avg = total / len(student["marks"])

    if avg > highest_avg:
        highest_avg =avg
        highest = student
highest["name"] , highest_avg
```

✅ Method 2: Using sum()

```python
highest = students[0]
highest_avg = 0
```

```python
for student in students:
    avg = sum(student["marks"])/len(student["marks"])

    if avg > highest_avg:
        highest_avg = avg
        highest = student
highest, highest_avg
```

✅ Method 3: Using max()

```python
highest = max(students, key = lambda student : sum(student["marks"])/len(student["marks"]))
print(highest)
```

✅ Method 4: every students average

```python
for student in students:
    avg= sum(student["marks"])/len(student["marks"])
    print(student["name"],"=", avg)
```

✅ Method 5: Store average in dictionary

```python
for student in students:
    avg= sum(student["marks"])/len(student["marks"])
    student["avg"] = avg
students
```
