Tasks

1. Find highest salary employee.


2. Find lowest salary employee.


3. Count employees in each department.


4. Count employees in each city.


5. Print IT employees.


6. Print employees with salary ≥ 60000.


7. Print employees with salary ≥ 60000 and experience ≥ 5.


8. Create a list of employee names working in Hyderabad.


9. Create a list of employees with experience ≥ 4.


10. Calculate total salary.


11. Calculate average salary.


12. Sort by salary (descending).


13. Sort by experience.


14. Group employee names by department.


15. Group employee names by city.


16. Group salaries by department.


17. Group experience by city.


```python
employees = [
    {"name":"Ali","dept":"IT","city":"Hyderabad","salary":65000,"exp":4},
    {"name":"Sara","dept":"HR","city":"Mumbai","salary":72000,"exp":6},
    {"name":"Ahmed","dept":"IT","city":"Hyderabad","salary":58000,"exp":3},
    {"name":"John","dept":"Finance","city":"Delhi","salary":45000,"exp":8},
    {"name":"Fatima","dept":"IT","city":"Bangalore","salary":82000,"exp":7},
    {"name":"Mary","dept":"HR","city":"Mumbai","salary":51000,"exp":2},
    {"name":"Zaid","dept":"Finance","city":"Delhi","salary":69000,"exp":5}
]
```

```python
highest_salary = employees[0]
lowest_salary =employees[0]

for emp in employees:
    if emp["salary"] > highest_salary["salary"]:
        highest_salary = emp
    if emp["salary"] < lowest_salary["salary"]:
        lowest_salary = emp

print("highest salary employee: ",highest_salary)
print("lowest salary employee: ",lowest_salary)
print("*"*15)

count_dept = {}
count_city = {}
for emp in employees:
    dept = emp["dept"]
    if dept in count_dept:
        count_dept[dept] +=1
    else:
        count_dept[dept] = 1
    city = emp["city"]
    if city in count_city:
        count_city[city] +=1
    else:
        count_city[city] = 1


print("employees in each department: ",count_dept)
print("employees in each city: ",count_city)

print("*"*15)
for emp in employees:
    if emp["salary"] >= 60000:
        print("employees with salary ≥ 60000: ",emp["name"])
print("*"*15)
for emp in employees:
    if emp["salary"] >= 60000  and emp["exp"] >= 5:
        print("employees with salary ≥ 60000 and experience ≥ 5: ",emp["name"])
print("*"*15)


emp_hyd = []
emp_exp = []
for emp in employees:
    if emp["city"] == "Hyderabad":
        emp_hyd.append(emp["name"])
    if emp["exp"] >= 4 :
        emp_exp.append(emp["name"])

print("employee names working in Hyderabad: ",emp_hyd)
print("employees with experience ≥ 4: ",emp_exp)
print("*"*15)

total_salary = 0
for emp in employees:
    total_salary += emp["salary"]
average_salary = total_salary / len(employees)

print("total salary: ",total_salary)
print("average salary: ",average_salary)
print("*"*15)

print("salary (descending): ",sorted(employees, key =lambda emp :emp["salary"],reverse =True))
print("Sort by experience: ",sorted(employees, key =lambda emp :emp["exp"]))
print("*"*15)


group_dept={}
group_city={}
group_salaries={}
group_experience={}
for emp in employees:
    dept = emp["dept"]
    if dept not in group_dept:
        group_dept[dept] = []
    group_dept[dept].append(emp["name"])

    city = emp["city"]
    if city not in group_city:
        group_city[city] = []
    group_city[city].append(emp["name"])


for emp in employees:
    depart = emp["dept"]
    if depart not in group_salaries:
        group_salaries[depart] = []
    group_salaries[depart].append(emp["salary"])

    city = emp["city"]
    if city not in group_experience:
        group_experience[city] = []
    group_experience[city].append(emp["exp"])

print("employee names by department: ",group_dept)
print("employee names by city: ",group_city)
print("salaries by department: ",group_salaries)
print("experience by city: ",group_experience)
print("*"*15)
```

Tasks

3. Count Running and Stopped servers.


4. Count servers in each team.


5. Print Running servers.


6. Print Running servers with CPU > 60.


7. Print Backend servers.


8. Print Running Backend servers.


9. Create a list of Running server names.


10. Create a list of Frontend server names.


11. Calculate total CPU.

14. Sort by server name.

17. Group CPU values by team.


18. Group memory values by status.



```python
servers = [
    {"name":"web01","status":"Running","team":"Frontend","cpu":65,"memory":70},
    {"name":"db01","status":"Stopped","team":"Database","cpu":0,"memory":40},
    {"name":"app01","status":"Running","team":"Backend","cpu":80,"memory":90},
    {"name":"cache01","status":"Running","team":"Backend","cpu":35,"memory":30},
    {"name":"monitor01","status":"Running","team":"DevOps","cpu":55,"memory":60},
    {"name":"backup01","status":"Stopped","team":"Database","cpu":0,"memory":50},
    {"name":"proxy01","status":"Running","team":"Frontend","cpu":75,"memory":85}
]
```

```python
highest_CPU = max(servers, key = lambda x:x["cpu"])
lowest_CPU = min(servers, key = lambda x:x["cpu"])

print("highest CPU server: ",highest_CPU)
print("lowest CPU server: ",lowest_CPU)
print("*"*15)

Count_Running_Stopped = {}
count_team ={}
for server in servers:
    status = server["status"]
    Count_Running_Stopped[status] = Count_Running_Stopped.get(status,0)+1

    team = server["team"]
    count_team[team] = count_team.get(team,0)+1
    
print("Count Running and Stopped servers.: ",Count_Running_Stopped)
print("Count servers in each team: ",count_team)
print("*"*15)


for server in servers:
    if server["status"] == "Running":
        print("Running servers: ",server["name"])
print("*"*15)

for server in servers:
    if server["status"] == "Running" and server["cpu"]>60:
        print("Running servers with CPU > 60: ",server["name"])
print("*"*15)

for server in servers:
    if server["team"] == "Backend":
        print("Backend servers: ",server["name"])
print("*"*15)

for server in servers:
    if server["status"] == "Running" and server["team"] == "Backend":
        print("Running Backend servers: ",server["name"])
print("*"*15)


Running_server = []
Frontend_server =[]
for server in servers:
    if server["status"] == "Running":
        Running_server.append(server["name"])
    if server["team"] == "Frontend":
        Frontend_server.append(server["name"])

print("Running servers: ",Running_server)
print("Frontend server names: ",Frontend_server)
print("*"*15)


total_CPU = sum(server["cpu"] for server in servers)
average_CPU = total_CPU / len(servers)

print("total CPU: ",total_CPU)
print("average CPU: ",average_CPU)
print("*"*15)


server  = servers.copy()
server.sort(key = lambda x:x["cpu"], reverse = True)
print("Sort by CPU (descending): ",server)

server  = servers.copy()
server.sort(key = lambda x:x["name"])
print("Sort by server name: ",server)
print("*"*15)


group_status = {}
group_team = {}
group_CPU = {}
group_memory = {}

for server in servers:
    group_status.setdefault(server["status"],[]).append(server["name"])
print("Group server names by status: ",group_status)

for server in servers:
    group_team.setdefault(server["team"],[]).append(server["name"])
print("Group server names by team: ",group_team)

for server in servers:
    group_CPU.setdefault(server["team"],[]).append(server["cpu"])
print("Group CPU values by team: ",group_CPU)

for server in servers:
    group_memory.setdefault(server["status"],[]).append(server["memory"])
print("Group memory values by status: ",group_memory)

print("*"*15)
group_cpu ={}
for server in servers:
    group_cpu.setdefault(server["status"],[]).append(server["name"])
print("group cpu by name: ",group_cpu )
```
