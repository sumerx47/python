Question 1 – Employees



Tasks

1. Print IT employees earning more than ₹50,000.


2. Print employees in HR or Finance.


3. Print employees not in IT.


4. Create a list of IT employee names earning more than ₹50,000.


```python
employees = [
    {"name":"Ali","dept":"IT","salary":45000},
    {"name":"Sara","dept":"HR","salary":70000},
    {"name":"Ahmed","dept":"IT","salary":65000},
    {"name":"John","dept":"Finance","salary":40000},
    {"name":"Fatima","dept":"IT","salary":80000}
]
```

```python
for emp in employees:
    if emp["dept"]=="IT" and emp["salary"]>50000:
        print("IT employees earning more than ₹50,000: ",emp)

for emp in employees:
    if  emp["dept"]=='HR' or emp["dept"] == "Finance":
        print(emp)

for emp in employees:
    if  emp["dept"] != "IT":
        print("TEST: ",emp)

emp_IT = {}
for emp in employees:
    if emp["dept"]=="IT" and emp["salary"]>50000:
        print("IT employees earning more than ₹50,000: ",emp["name"])
```

🟠 Level 3 (Advanced)

Question 5 – Company Employees



Tasks

1. Print IT employees with salary > ₹60,000 and experience > 3.


2. Print employees from HR or Finance.


3. Create a list of experienced IT employees.

```python
employees = [
    {"name":"Ali","dept":"IT","salary":45000,"exp":2},
    {"name":"Sara","dept":"HR","salary":70000,"exp":5},
    {"name":"Ahmed","dept":"IT","salary":65000,"exp":4},
    {"name":"John","dept":"Finance","salary":40000,"exp":7},
    {"name":"Fatima","dept":"IT","salary":80000,"exp":6}
]
```

```python
print([emp for emp in employees if emp["dept"] == "IT" and emp["salary"] > 60000 and emp["exp"] > 3])

for emp in employees:
    if emp["dept"] == "Finance" or emp["dept"] == "HR":
        print(emp)

emp_IT = []
for emp in employees:
    if emp["dept"] == "IT" and emp["exp"] > 1 :
        emp_IT.append(emp["name"])
print("IT + EXPERIENCED: ",emp_IT)
```

Question 9 – Mixed Employee Conditions


Tasks

1. Print IT employees with salary ≥ ₹60,000 and experience ≥ 4.


2. Print HR employees with salary > ₹50,000.


3. Print employees in Finance or experience > 5.


4. Create a list of senior IT employee names.

```python
employees = [
    {"name":"Ali","dept":"IT","salary":50000,"exp":2},
    {"name":"Sara","dept":"HR","salary":65000,"exp":6},
    {"name":"Ahmed","dept":"IT","salary":70000,"exp":4},
    {"name":"John","dept":"Finance","salary":45000,"exp":8},
    {"name":"Fatima","dept":"IT","salary":60000,"exp":5},
    {"name":"Aisha","dept":"HR","salary":52000,"exp":2}
]
```

```python
print("IT employees with salary ≥ ₹60,000 and experience ≥ 4: ",[emp for emp in employees if emp["dept"]=="IT" and emp["salary"]>= 60000 and emp["exp"]>=4 ])
print("HR employees with salary > ₹50,000: ", [emp for emp in employees if emp["dept"] == "HR" and emp["salary"]>50000])
print("employees in Finance or experience > 5: ", [emp for emp in employees if emp["dept"] == "Finance" or emp["exp"]>5])
emp_IT = []
for emp in employees:
    if emp["dept"] == "IT" and emp["exp"] >=4:
        emp_IT.append(emp["name"])

print("senior IT employee names: ",emp_IT)
```
