Question 1 — Agent Validation

Given:

Create:

get_average(agent_id)

It should:

Return the agent's average.

Handle an invalid Agent ID.

Handle an empty score list.

Handle invalid score data.


```python
agents = {
    "A101": {
        "name": "Arjun",
        "scores": [85, 92, 78]
    },

    "A102": {
        "name": "Sara",
        "scores": [95, 88, 91]
    }
}
```

```python
def get_average(agent_id):
    try:
        score = agents[agent_id]["scores"]
        average = sum(score)/len(score)
        return average
    except KeyError:
        return "agent key not available"
    except ZeroDivisionError:
        return "no scores available"
    except TypeError:
        return "invalid score data"

print("A102: ",get_average("A102"))
print("A101: ",get_average("A101"))
print("A133: ",get_average("A133"))
```

Question 2 — Safe Data Processing

Given:

Create:

get_salary(employee_id)

It should:

Return the salary.

Handle an employee ID that doesn't exist.

Handle missing "salary" data.

Return a meaningful error message instead of crashing.

```python
employees = {
    "E101": {"name": "Ali", "salary": 60000},
    "E102": {"name": "Sara", "salary": 85000},
    "E103": {"name": "Ahmed", "salary": 95000}
}
```

```python
def get_salary(employee_id):
    try :
        salary = employees[employee_id]["salary"]
        return salary
    
    except KeyError:
        return "employee id or salary doesn't exist"

    except TypeError:
        return "invalid salary data"

print("E102:", get_salary("E102"))
print("E105:", get_salary("E105"))
```

```python
def get_salary(employee_id):
    try :
        employee = employees[employee_id]
    
    except KeyError:
        return "employee id doesn't exist"

    try :
        salary = employee["salary"]
    
    except KeyError:
        return "salary information missing"

    return salary

print("E102:", get_salary("E102"))
print("E105:", get_salary("E105"))
```
