```python
employees = [
    {"id": 101, "name": "Ali", "dept": "IT", "skills": ["python", "linux"], "salary": 60000},
    {"id": 102, "name": "Sara", "dept": "HR", "skills": ["excel", "communication"], "salary": 45000},
    {"id": 103, "name": "Ahmed", "dept": "IT", "skills": ["python", "docker", "linux"], "salary": 80000},
    {"id": 104, "name": "Zoya", "dept": "DevOps", "skills": ["python", "docker", "kubernetes"], "salary": 90000},
    {"id": 105, "name": "Omar", "dept": "DevOps", "skills": ["linux", "docker"], "salary": 75000},
    {"id": 106, "name": "Ali", "dept": "IT", "skills": ["python", "git"], "salary": 70000},
    {"id": 107, "name": "Aliza", "dept": "IT", "skills": ["python", "git"], "salary": 40000},
    {"id": 108, "name": "Maliza", "dept": "DevOps", "skills": ["python", "git"], "salary": 95000}
]
```

```python
def analyze_employees(employees):
    # 1. Highest and Lowest
    highest = employees[0]
    lowest = employees[0]

    #2. Total
    total = 0

    #3. Containers
    Above_average = []
    IT_DevOps_salary = []

    #main processsing
    for emp in employees:
        # 1. Highest-paid employee
        salary = emp["salary"]


        if salary > highest["salary"]:
            highest = emp

        # 2. Lowest-paid employee
        if salary  < lowest["salary"]:
            lowest = emp

        # 3. Average salary
        total += emp["salary"]
        avg = total / len(employees)

        # 4. Employees earning above the average salary
        if salary > 70000:
            Above_average.append(emp["name"])

        # 5. Employees belonging to IT or DevOps whose salary is ≥ 70,000
        if (emp["dept"] == "IT" or emp["dept"] == "DevOps") and emp["salary"] >= 70000:
            IT_DevOps_salary.append(emp["name"])


        # 6. Count of employees in each department


        # 7. Employees grouped by department







            

    return { 
        "Highest" : highest,
        "Lowest" : lowest,
        "Total" : total,
        "Average" : avg,
       "Above_average": Above_average,
       "IT_DevOps_salary": IT_DevOps_salary

    }
result = analyze_employees(employees)
print("Highest:",result["Highest"])
print("Lowest:",result["Lowest"])
print("Total:",result["Total"])
print("Average:",result["Average"])
print("Salary above Average:",result["Above_average"])
print("Employees of IT,DevOps with salary ≥ 70,000:",result["IT_DevOps_salary"])
```
