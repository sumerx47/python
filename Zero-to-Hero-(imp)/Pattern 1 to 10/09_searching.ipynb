Question 1 – Find Employee by Name



Tasks

1. Find employee named Sara.


2. Print "Not Found" if employee does not exist.


```python
employees = [
    {"id":101,"name":"Ali","salary":50000},
    {"id":102,"name":"Sara","salary":70000},
    {"id":103,"name":"Ahmed","salary":65000},
    {"id":104,"name":"John","salary":45000}
]
```

#### method -1

```python
search = "Sara"
found = False
for emp in employees:
    if emp["name"] == search:
        print(emp)
        found = True
        break
if not found:
    print("not found")
```

#### method -2

```python
search = "Sara"
result = next((emp for emp in employees if emp["name"]==search), None)
if result:
    print(result)
else:
    print("Not found")
```
