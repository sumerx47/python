🔴 Question 1 — Employee Management


Tasks

1. Highest salary


2. Lowest salary


3. Count departments


4. Total salary


5. Average salary


6. Print IT employees


7. Print employees with salary > 55,000


8. Create a list of IT employee names


9. Create a list of HR employee names


10. Total experience of all employees


11. Average experience

```python
employees = [
    {"name":"Ali","dept":"IT","salary":50000,"experience":2},
    {"name":"Sara","dept":"HR","salary":65000,"experience":5},
    {"name":"Ahmed","dept":"IT","salary":70000,"experience":4},
    {"name":"John","dept":"Finance","salary":45000,"experience":3},
    {"name":"Fatima","dept":"IT","salary":60000,"experience":6},
    {"name":"Aisha","dept":"HR","salary":52000,"experience":2}
]
```

```python
Highest_salary = max(employees, key = lambda emp : emp["salary"])
Lowest_salary = min(employees , key = lambda emp: emp["salary"])

Count_departments = {}
for emp in employees :
    dept= emp["dept"]
    Count_departments[dept]= Count_departments.get(dept, 0 )+1


Total_salary = sum(emp["salary"] for emp in employees)
Average_salary = Total_salary / len(employees)


print("Highest salary employee: ", Highest_salary)
print("Lowest salary employee: ", Lowest_salary)
print("Count of departments: ", Count_departments)
print("Total salary: ", Total_salary)
print("Average salary: ", Average_salary)



emp_IT =[]
emp_HR = []

print("*" * 15)
for emp in employees:
    if "IT" in emp["dept"]:
        print("IT employees: ",emp["name"])
        emp_IT.append(emp["name"])
    if emp["dept"] == "HR":
        emp_HR.append(emp["name"])
       
print("*" * 15)
for emp in employees:
    if emp["salary"] > 55000:
        print("employees with salary > 55,000: ",emp["name"])
print("*" * 15)

print("IT employees: ", emp_IT)
print("HR employees: ", emp_HR)
print("*" * 15)

Total_experience = sum(emp["experience"] for emp in employees)
Average_experience = Total_experience / len(employees)

print("Total experience of all employees: ", Total_experience)
print("Average experience: ", Average_experience)
```

🔴 Question 2 — Student Records


Tasks

1. Highest marks


2. Lowest marks


3. Count courses


4. Total marks


5. Average marks


6. Print Python students


7. Print students with marks ≥ 80


8. Create a list of Python student names


9. Create a list of students scoring below 75


10. Total marks of all Python students (using filtering + sum)



```python
students = [
    {"name":"Ali","course":"Python","marks":92},
    {"name":"Sara","course":"Java","marks":81},
    {"name":"Ahmed","course":"Python","marks":95},
    {"name":"John","course":"C++","marks":65},
    {"name":"Fatima","course":"Python","marks":88},
    {"name":"Mary","course":"Java","marks":72},
    {"name":"Zaid","course":"Python","marks":79}
]
```

```python
Highest_marks = students[0]
Lowest_marks = students[0]

Count_courses = {}

Total_marks = 0


for student in students:
    if student["marks"] > Highest_marks["marks"]:
        Highest_marks = student

    if student["marks"] < Lowest_marks["marks"]:
        Lowest_marks = student

    course= student["course"]
    if course in Count_courses:
        Count_courses[course] += 1
    else:
        Count_courses[course] = 1

    Total_marks += student["marks"]
    Average_marks = Total_marks / len(students)

print("Highest marks: ", Highest_marks)
print("Lowest marks: ", Lowest_marks)
print("Count of departments: ", Count_courses)
print("Total marks: ", Total_marks)
print("Average marks: ", Average_marks)


print("*" * 15)

student_py = []
students_score = []
total_py =0

for student in students:
    if student["course"] == "Python":
        print("Python students: ", student["name"])
        student_py.append(student["name"])
        total_py += student["marks"]

print("*" * 15)
for student in students:
    if student["marks"] >= 80:
        print("students with marks ≥ 80: ", student["name"])
    if student["marks"] < 75:
        students_score.append(student["name"])
print("*" * 15)

print("Python students: ", student_py)
print("students with score < 75: ", students_score)
print("Total marks of all Python students: ", total_py)

print("*" * 15)
```

🔴 Question 3 — AWS EC2 Monitoring



Tasks

1. Count instance states


2. Highest CPU usage


3. Lowest CPU usage


4. Total CPU usage


5. Average CPU usage


6. Print running instances


7. Create a list of running IDs


8. Total CPU used by running instances only

```python
instances = [
    {"id":"i-101","state":"running","cpu":45},
    {"id":"i-102","state":"stopped","cpu":0},
    {"id":"i-103","state":"running","cpu":65},
    {"id":"i-104","state":"terminated","cpu":0},
    {"id":"i-105","state":"running","cpu":35},
    {"id":"i-106","state":"running","cpu":75}
]
```

```python
Count_instance = {}
for instance in instances:
    state = instance["state"]
    Count_instance[state] = Count_instance.get(state,0)+1

Highest_CPU = max(instances, key = lambda instance : instance["cpu"])
Lowest_CPU = min(instances, key = lambda instance : instance["cpu"])
Total_CPU = sum(instance["cpu"] for instance in instances)
Average_CPU= Total_CPU / len(instances)


print("Count of instance states: ", Count_instance)
print("Highest CPU usage: ", Highest_CPU)
print("Lowest CPU usage: ", Lowest_CPU)
print("Total CPU usage: ", Total_CPU)
print("Average CPU usage: ", Average_CPU)

print("*" * 15)

running_IDs = []
Total_CPU = 0

for instance in instances:
    if instance["state"] == "running":
        print("running instances: ", instance["id"])
        running_IDs.append(instance["id"])
        Total_CPU += instance["cpu"]

print("*" * 15)
print("running IDs: ", running_IDs)
print("Total CPU used by running instances only: ", Total_CPU)
```

🔴 Question 4 — Shopping Cart



Tasks

1. Highest priced item


2. Lowest priced item


3. Total bill


4. Average item price


5. Print items costing more than ₹50


6. Create a list of expensive item names


7. Total quantity purchased

```python
cart = [
    {"item":"Milk","qty":2,"price":30},
    {"item":"Bread","qty":3,"price":40},
    {"item":"Eggs","qty":12,"price":8},
    {"item":"Butter","qty":1,"price":60},
    {"item":"Rice","qty":5,"price":80}
]
```

```python
Highest_priced = cart[0]
Lowest_priced = cart[0]

Total_bill = 0

for items in cart:
    if items["price"] > Highest_priced["price"]:
        Highest_priced = items
    if items["price"] < Lowest_priced["price"]:
            Lowest_priced = items

    Total_bill += items["price"]
    Average_price = Total_bill/len(cart)
    
print("Highest priced item: ", Highest_priced)
print("Lowest priced item: ", Lowest_priced)
print("Total bill: ", Total_bill)
print("Average item price: ", Average_price)
print("*" * 15)

expensive_item =[]

for items in cart:
    if items["price"] > 50:
        print("items costing more than ₹50: ", items["item"])
        expensive_item.append(items["item"])

    Total_quantity_purchased =  items["qty"]* items["price"]

print("*" * 15)

print("expensive item names: ", expensive_item)
print("Total quantity purchased: ", Total_quantity_purchased)


```

🔴 Question 6 — Hospital Patients



Tasks

1. Oldest patient


2. Youngest patient


3. Total hospital bill


4. Average bill


5. Print patients aged 50 or above


6. Create a list of senior patient names


7. Total bill of senior patients only


```python
patients = [
    {"name":"Ali","age":45,"bill":12000},
    {"name":"Sara","age":60,"bill":18000},
    {"name":"Ahmed","age":30,"bill":9000},
    {"name":"John","age":70,"bill":22000},
    {"name":"Fatima","age":50,"bill":15000}
]
```

```python
Oldest_patient = max(patients, key = lambda patient:patient["age"])
Youngest_patient = min(patients, key =lambda patient: patient["age"])
Total_hospital_bill = sum(patient["bill"] for patient in patients)
Average_bill = Total_hospital_bill/len(patients)

print("Oldest_patient: ", Oldest_patient)
print("Youngest_patient: ", Youngest_patient)
print("Total_hospital_bill: ", Total_hospital_bill)
print("Average_bill: ", Average_bill)
print("*" * 15)


senior_patient =[]
senior_patients_bill=0
for patient in patients:
    if patient["age"] >= 50:
        print("patients aged 50 or above: ",patient["name"])
        senior_patient.append(patient["name"])
        senior_patients_bill += patient["bill"]

print("*" * 15)
print("senior patient names: ", senior_patient)
print("Total bill of senior patients only: ", senior_patients_bill)
```
