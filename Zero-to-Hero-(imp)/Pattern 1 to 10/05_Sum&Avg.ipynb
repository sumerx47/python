Question 1 – Employee Salary


Find

1. Total salary


2. Average salary



```python
employees = [
    {"name":"Ali","salary":45000},
    {"name":"Sara","salary":70000},
    {"name":"Ahmed","salary":60000},
    {"name":"John","salary":35000},
    {"name":"Fatima","salary":80000}
]
```

#### method-1

```python
total = 0

for emp in employees:
    total += emp["salary"]


avg_salary = total/len(employees)

print(total)
print(avg_salary)
```

#### method-2

```python
total_salary = sum(emp["salary"] for emp in employees)
avg_salary = total_salary/len(employees)

print(total_salary)
print(avg_salary)
```

Question 2 – Students

```python
total_marks = sum (student["marks"] for student in students)
avg_marks = total_marks/len(students)

print(total_marks)
print(avg_marks)
```

```python
total = 0
for student in students:
    total += student["marks"]
avg_marks = total / len(students)

print(total)
print(avg_marks)
```

Question 4 – Shopping Cart



Find

1. Total bill



(Hint: qty × price)

```python
cart = [
    {"item":"Milk","qty":2,"price":30},
    {"item":"Bread","qty":3,"price":40},
    {"item":"Eggs","qty":12,"price":8}
]
```

```python
total_bill = sum(item["qty"] * item["price"] for item in cart)
total_bill
```

```python
total=0

for items in cart:
    total += items["qty"] * items["price"]
total
```

Question 5 – Product Inventory



Find

1. Total inventory value



(Hint: stock × price)

```python
products = [
    {"name":"Laptop","stock":5,"price":60000},
    {"name":"Mouse","stock":20,"price":800},
    {"name":"Keyboard","stock":10,"price":1500},
    {"name":"Monitor","stock":4,"price":12000}
]
```

```python
total_inventory_value = sum(product["stock"] * product["price"]  for product in products)
total_inventory_value
```

```python
total = 0
for product in products:
    total += product["stock"] * product["price"]
total
```
