Question 1 – Employee Salary


Find

1. Total salary


2. Average salary



```python
employees = [
    {"name":"Ali","salary":45000},
    {"name":"Sara","salary":70000},
    {"name":"Ahmed","salary":60000},
    {"name":"John","salary":35000},
    {"name":"Fatima","salary":80000}
]
```

#### method-1

```python
total = 0

for emp in employees:
    total += emp["salary"]


avg_salary = total/len(employees)

print(total)
print(avg_salary)
```

#### method-2

```python
total_salary = sum(emp["salary"] for emp in employees)
avg_salary = total_salary/len(employees)

print(total_salary)
print(avg_salary)
```
