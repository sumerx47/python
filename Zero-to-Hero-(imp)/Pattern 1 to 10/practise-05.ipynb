Question 1 – Employee Management


Tasks

1. Highest salary


2. Lowest salary


3. Count employees in each department


4. Total salary


5. Average salary


6. Print IT employees


7. Create a list of HR employee names


8. Sort by salary (ascending)


9. Sort by salary (descending)


10. Sort by employee name

```python
employees = [
    {"name":"Ali","dept":"IT","salary":50000,"exp":2},
    {"name":"Sara","dept":"HR","salary":65000,"exp":5},
    {"name":"Ahmed","dept":"IT","salary":70000,"exp":4},
    {"name":"John","dept":"Finance","salary":45000,"exp":3},
    {"name":"Fatima","dept":"IT","salary":60000,"exp":6},
    {"name":"Aisha","dept":"HR","salary":52000,"exp":2}
]
```

```python
Highest_salary = max(employees, key = lambda emp : emp["salary"])
Lowest_salary = min(employees, key =lambda emp :emp["salary"])

print("Highest salary: ", Highest_salary)
print("Lowest salary: ", Lowest_salary)

emp_count = {}
for emp in employees:
    dept = emp["dept"]
    emp_count[dept] = emp_count.get(dept,0)+1

Total_salary = sum(emp["salary"] for emp in employees)
Average_salary = Total_salary / len(employees)


print("employees in each department: ", emp_count)
print("Total salary: ", Total_salary)
print("Average salary: ", Average_salary)
print("*"*15)


employees_HR=[]

for emp in employees:
    if "IT" in emp["dept"]:
        print("IT employees: ", emp["name"])

    if emp["dept"] =="HR":
        employees_HR.append(emp["name"])

print("*"*15)
print("Employees HR: ", employees_HR)
print("*"*15)

emp = employees.copy()
emp.sort(key=lambda x : x["salary"])
print("salary ascending: ",emp)
emp.sort(key=lambda x : x["salary"], reverse = True)
print("salary descending: ",emp)
emp.sort(key=lambda x : x["name"])
print("name ascending: ",emp)
```

🔴 Question 2 – Student Management



Tasks

1. Highest marks


2. Lowest marks


3. Count courses


4. Total marks


5. Average marks


6. Print Python students


7. Create a list of students with marks ≥ 80


8. Sort by marks


9. Sort by marks (descending)


10. Sort by course

```python
students = [
    {"name":"Ali","course":"Python","marks":92},
    {"name":"Sara","course":"Java","marks":81},
    {"name":"Ahmed","course":"Python","marks":95},
    {"name":"John","course":"C++","marks":65},
    {"name":"Fatima","course":"Python","marks":88},
    {"name":"Mary","course":"Java","marks":72},
    {"name":"Zaid","course":"Python","marks":79}
]
```

```python
Highest_marks = students[0]
Lowest_marks =students[0]
course_count = {}


for student in students:
    if student["marks"] > Highest_marks["marks"]:
        Highest_marks = student
    elif student["marks"] < Lowest_marks["marks"]:
        Lowest_marks = student

    course = student["course"]
    if course in course_count:
        course_count[course] += 1
    else:
        course_count[course] = 1


Total_marks = 0
for student in students:
    Total_marks += student["marks"]
Average_marks = Total_marks / len(students)



print("Highest marks: ", Highest_marks)
print("Lowest marks: ", Lowest_marks)
print("course count: ", course_count)

print("Total marks: ", Total_marks)
print("Average marks: ", Average_marks)
print("*"*15)

students_marks=[]
for student in students:
    if "Python" in student["course"]:
        print("Python students: ", student["name"])
    if student["marks"] >= 80:
        students_marks.append(student["name"])

print("*"*15)
print("students with marks ≥ 80: ", students_marks)
print("*"*15)

print("Sort by marks ascending: ",sorted(students, key = lambda x:x["marks"]))
print("Sort by marks (descending): ",sorted(students, key = lambda x:x["marks"], reverse =True))
print("Sort by course ascending: ",sorted(students, key = lambda x:x["course"]))

```
