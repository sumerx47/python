```python
employees = [
    {"name":"Ali","dept":"IT","salary":50000,"exp":2},
    {"name":"Sara","dept":"HR","salary":65000,"exp":5},
    {"name":"Ahmed","dept":"IT","salary":70000,"exp":4},
    {"name":"John","dept":"Finance","salary":45000,"exp":3},
    {"name":"Fatima","dept":"IT","salary":60000,"exp":6},
    {"name":"Aisha","dept":"HR","salary":52000,"exp":2}
]
```

```python
Highest_salary = max(employees, key = lambda emp : emp["salary"])
Lowest_salary = min(employees, key =lambda emp :emp["salary"])

print("Highest salary: ", Highest_salary)
print("Lowest salary: ", Lowest_salary)

emp_count = {}
for emp in employees:
    dept = emp["dept"]
    emp_count[dept] = emp_count.get(dept,0)+1

Total_salary = sum(emp["salary"] for emp in employees)
Average_salary = Total_salary / len(employees)


print("employees in each department: ", emp_count)
print("Total salary: ", Total_salary)
print("Average salary: ", Average_salary)
print("*"*15)


employees_HR=[]

for emp in employees:
    if "IT" in emp["dept"]:
        print("IT employees: ", emp["name"])

    if emp["dept"] =="HR":
        employees_HR.append(emp["name"])

print("*"*15)
print("Employees HR: ", employees_HR)
print("*"*15)

emp = employees.copy()
emp.sort(key=lambda x : x["salary"])
print("salary ascending: ",emp)
emp.sort(key=lambda x : x["salary"], reverse = True)
print("salary descending: ",emp)
emp.sort(key=lambda x : x["name"])
print("name ascending: ",emp)
```
