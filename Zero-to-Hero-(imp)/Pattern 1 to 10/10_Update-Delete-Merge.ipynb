Question 1 – Update Employee Salary


Tasks

1. Update Ali's salary to 60000.

```python
employees = [
    {"id":101,"name":"Ali","salary":50000},
    {"id":102,"name":"Sara","salary":70000},
    {"id":103,"name":"Ahmed","salary":65000}
]
```

#### Method - 1

```python
for emp in employees:
    if emp["name"] == "Sara": 
        emp["salary"] = 90000
        break
print(employees)
```

#### Method - 2

```python
# (optional method)
next(emp for emp in employees if emp["name"] == "Ali")["salary"] = 60000
print(employees)
```

```python
# USE THIS METHOD :
for emp in employees:
    emp.update({"salary": 20000} if emp["name"] == "Ali" else {})
print(employees)
```

Question 13 – Employee Analytics

Tasks

1. Increase every IT employee's salary by 10%.


2. Change every Finance employee's city to Hyderabad.


3. Delete employees whose salary is below 60000.


4. Merge:



new_employees = [
    {"id":108,"name":"Rehan","dept":"IT","city":"Pune","salary":75000},
    {"id":109,"name":"Imran","dept":"HR","city":"Mumbai","salary":68000}
]

5. Print the final updated list.

```python
employees = [
    {"id":101,"name":"Ali","dept":"IT","city":"Hyderabad","salary":65000},
    {"id":102,"name":"Sara","dept":"HR","city":"Mumbai","salary":72000},
    {"id":103,"name":"Ahmed","dept":"IT","city":"Hyderabad","salary":58000},
    {"id":104,"name":"John","dept":"Finance","city":"Delhi","salary":45000},
    {"id":105,"name":"Fatima","dept":"IT","city":"Bangalore","salary":82000},
    {"id":106,"name":"Mary","dept":"HR","city":"Mumbai","salary":51000},
    {"id":107,"name":"Zaid","dept":"Finance","city":"Delhi","salary":69000}
]
```

```python
for emp in employees:
    if emp["dept"] == "IT":
        emp["salary"] = int(emp["salary"]*1.10)
print(employees)

for emp in employees:
    if emp["dept"] == "Finance":
        emp["city"] = "Hyderabad"
print(employees)

less_salary =[]
for emp in employees:
    if emp["salary"] >= 60000:
        less_salary.append(emp)
print(less_salary)

new_employees = [
    {"id":108,"name":"Rehan","dept":"IT","city":"Pune","salary":75000},
    {"id":109,"name":"Imran","dept":"HR","city":"Mumbai","salary":68000}
]

updated_employees = employees + new_employees
print(updated_employees)
```
