Question 1 – Update Employee Salary


Tasks

1. Update Ali's salary to 60000.

```python
employees = [
    {"id":101,"name":"Ali","salary":50000},
    {"id":102,"name":"Sara","salary":70000},
    {"id":103,"name":"Ahmed","salary":65000}
]
```

#### Method - 1

```python
for emp in employees:
    if emp["name"] == "Sara": 
        emp["salary"] = 90000
        break
print(employees)
```

#### Method - 2

```python
next(emp for emp in employees if emp["name"] == "Ali")["salary"] = 60000
print(employees)
```

1. Update Ahmed's marks to 90.

```python
students = [
    {"roll":1,"name":"Ali","marks":85},
    {"roll":2,"name":"Sara","marks":91},
    {"roll":3,"name":"Ahmed","marks":78}
]
```

```python
for student in students:
    if student["name"] == "Ahmed":
        student["marks"] = 90
        break
print(students)
```

```python
next(student for student in students if student["name"] == "Ali")["marks"]=95
print(students)
```
