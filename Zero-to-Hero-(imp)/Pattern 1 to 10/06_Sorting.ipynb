Question 1 – Employee Salary

Tasks

1. Sort by salary (ascending)


2. Sort by salary (descending)


3. Sort by employee name (A–Z)


4. Sort by employee name (Z–A)


```python
employees = [
    {"name":"Ali","salary":45000},
    {"name":"Sara","salary":70000},
    {"name":"Ahmed","salary":60000},
    {"name":"John","salary":35000},
    {"name":"Fatima","salary":80000}
]
```

```python
print(sorted(employees, key = lambda x:x["salary"]))
print(sorted(employees, key = lambda x:x["salary"], reverse = True))
print("*"*15)
print(sorted(employees, key = lambda x:x["name"]))
print(sorted(employees, key = lambda x:x["name"], reverse = True))
```

Question 2 – Students



Tasks

1. Sort by marks (ascending)


2. Sort by marks (descending)


3. Sort by name (A–Z)


4. Sort by name (Z–A)

```python
students = [
    {"name":"Ali","marks":90},
    {"name":"Sara","marks":65},
    {"name":"Ahmed","marks":82},
    {"name":"John","marks":45},
    {"name":"Fatima","marks":96}
]
```

```python
marks_ascending = sorted(students, key = lambda student:student["marks"])
marks_descending =sorted(students, key = lambda student:student["marks"], reverse= True)

name_ascending =sorted(students, key = lambda student:student["name"])
name_descending =sorted(students, key = lambda student:student["name"], reverse =True)

print(marks_ascending)
print(marks_descending)
print(name_ascending)
print(name_descending)
```
