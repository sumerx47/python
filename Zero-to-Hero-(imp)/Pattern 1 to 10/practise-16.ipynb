Question 1 — Student Performance Analysis

Assume students.txt contains:

Ayaan,20,78,85,92,67,88
Sara,21,95,89,93,90,96
Zaid,19,55,62,48,70,59
Maryam,20,88,76,91,84,79
Ali,22,72,81,69,77,85

Create:

analyze_students(filename)

The function should:

1. Read the file line by line.


2. Extract:

Student name

Age

5 marks



3. Calculate each student's average.


4. Assign grades:

A → ≥ 90

B → ≥ 80

C → ≥ 70

D → ≥ 60

F → < 60



5. Find students with average ≥ 75.


6. Find the highest-scoring student.


7. Find the lowest-scoring student.


8. Find the average of all students' averages.


9. Find all marks ≥ 80 across all students.


10. Find the unique marks that appear in the file.


11. Return:

{
    "students": [...],
    "qualified": [...],
    "highest": (...),
    "lowest": (...),
    "overall_average": ...,
    "high_marks": [...],
    "unique_marks": {...}
}

Handle:

FileNotFoundError
ValueError
Malformed lines
Empty lines

Skills tested

TXT
Functions
Loops
Lists
Tuples
Sets
Dictionaries
List comprehension
Set comprehension
Conditions
Exception handling
Sorting
Highest/lowest


---

```python
def calculate_avg(marks):
    return sum(marks)/len(marks)

def get_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 80:
        return "B"
    elif avg >= 70:
        return "C"
    elif avg >= 60:
        return "D"
    else:
        return "F"



def analyze_students(filename):
    # 1. Read the file line by line.
    students = []
    try : 
        with open(filename ,"r") as file:
            for line in file:
                line = line.strip() 

                if not line :
                    continue
                # 2
                try:
                    parts = line.split(",")

                    if len(parts) != 7:
                        continue

                    name = parts[0]
                    age = int[parts[1]]
                    marks = [int(mark) for mark in parts[2:]]

                    # 3
                    average = calculate_avg(marks)
                    grade = get_grade(average)

                    students.append({
                        "name":name,
                        "age":age,
                        "marks": marks,
                        "average":round(average,1),
                        "grade":grade
                    })
                except ValueError:
                    continue

    except FileNotFoundError:
        return "file does not exist"
    except PermissionError:
        return "permission denied"
    if not students:
        return "no valid student data"
```
