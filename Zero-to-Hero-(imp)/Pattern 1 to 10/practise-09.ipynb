4. Count employees in each city.



⭐️ Pattern 3 – Filtering

5. Employees with salary ≥ 70000.


6. IT employees from Hyderabad.



⭐️ Pattern 4 – Create New List

7. Create a list of employee names.


8. Create a list of all salaries.



⭐️ Pattern 5 – Sum / Average

9. Total salary.


10. Average salary.


11. Average salary of IT department.



⭐️ Pattern 6 – Sorting

12. Sort by salary (ascending).


13. Sort by name.


14. Sort by city then salary.

15. IT employees with salary > 70000.


16. HR employees from Mumbai.


17. Finance employees with salary < 60000.

18. Group names by department.


19. Group salaries by city.

20. Find employee ID 106.

23. Increase every IT salary by 5000.


24. Change Finance city to Hyderabad.


25. Delete salary below 60000.


26. Merge new_employees.


27. Search Aisha.


28. Group names by department after merge.

```python
employees = [
    {"id":101,"name":"Ali","dept":"IT","city":"Hyderabad","salary":65000},
    {"id":102,"name":"Sara","dept":"HR","city":"Mumbai","salary":72000},
    {"id":103,"name":"Ahmed","dept":"IT","city":"Hyderabad","salary":58000},
    {"id":104,"name":"John","dept":"Finance","city":"Delhi","salary":45000},
    {"id":105,"name":"Fatima","dept":"IT","city":"Bangalore","salary":82000},
    {"id":106,"name":"Mary","dept":"HR","city":"Mumbai","salary":51000},
    {"id":107,"name":"Zaid","dept":"Finance","city":"Delhi","salary":69000},
    {"id":108,"name":"Rehan","dept":"IT","city":"Pune","salary":76000}
]

new_employees = [
    {"id":109,"name":"Imran","dept":"HR","city":"Mumbai","salary":68000},
    {"id":110,"name":"Aisha","dept":"IT","city":"Hyderabad","salary":87000}
]
```

```python
# ⭐️ Pattern 1 – Highest / Lowest
Highest_salary= employees[0]
Lowest_salary =employees[0]

for emp in employees:
    if emp["salary"] > Highest_salary["salary"]:
        Highest_salary=emp
    if emp["salary"] < Lowest_salary["salary"]:
        Lowest_salary = emp

print("Highest salary employee: ",Highest_salary )
print("Lowest salary employee: ", Lowest_salary)
print("*"*15)

# ⭐️ Pattern 2 – Counting
emp_dept = {}
emp_city = {}
for emp in employees:
    dept = emp["dept"]
    if dept in emp_dept:
        emp_dept[dept] += 1
    else:
        emp_dept[dept] = 1

    city = emp["city"]
    if city in emp_city:
        emp_city[city] += 1
    else:
        emp_city[city] = 1
        
print("employees in each department: ", emp_dept)
print("employees in each city: ", emp_city)
print("*"*15)

# ⭐️ Pattern 3 – Filtering
for emp in employees:
    if emp["salary"]>=70000:
        print("Employees with salary ≥ 70000: ", emp)
print("*"*15)

for emp in employees:
    if emp["dept"]>="IT" and emp["city"] == "Hyderabad":
            print("IT employees from Hyderabad: ", emp)
print("*"*15)

# ⭐️ Pattern 4 – Create New List
employee_names=[]
emp_salaries = []
for empl in employees:
    employee_names.append(empl["name"])

    emp_salaries.append(empl["salary"])

print("list of employee names: ", employee_names)
print("list of all salaries: ", emp_salaries)
print("*"*15)

# ⭐️ Pattern 5 – Sum / Average
Total_salary = 0
for emp in employees:
    Total_salary += emp["salary"]
print("Total salary: ", Total_salary)

avg_sal = Total_salary/ len(employees)
print("avg salary: ", avg_sal)
print("*"*15)

Total_salary_IT = 0
Total_count_IT = 0 
for emp in employees:
    if emp["dept"] == "IT":
        Total_salary_IT += emp["salary"]
        Total_count_IT += 1
print("Total salary of IT employees:",Total_salary_IT)
print("Total count of IT:",Total_count_IT)

avg_salary_IT = Total_salary_IT / Total_count_IT
print("avg salary: ", avg_salary_IT)
print("*"*15)











# ⭐️ Pattern 6 – Sorting
print("Sort by salary (ascending): ",sorted(employees, key= lambda emp : emp["salary"]))
print("Sort by name: ", sorted(employees, key = lambda emp :emp["name"]))
print("Sort by city: ", sorted(employees, key = lambda emp :emp["city"]))
print("Sort by salary: ", sorted(employees, key = lambda emp :emp["salary"]))
print("*"*15)


# ⭐️ Pattern 7 – Multiple Conditions
empl_IT = []
emp_HR = []
emp_FINANCE = []
for emp in employees:
    if emp["dept"] == "IT" and emp["salary"] > 70000:
        empl_IT.append(emp["name"])
    if emp["dept"] == "HR" and emp["city"] == "Mumbai":
       emp_HR.append(emp["name"])
    if emp["dept"] == "Finance" and emp["salary"] < 60000:
        emp_FINANCE.append(emp["name"])
print("IT employees with salary > 70000: ",empl_IT)
print("HR employees from Mumbai: ",emp_HR)
print("Finance employees with salary < 60000: ",emp_FINANCE)
print("*"*15)          


# ⭐️ Pattern 8 – Grouping
group_dept = {}
Group_salaries ={}
for emp in employees:
    dept = emp["dept"]
    if dept not in group_dept:
        group_dept[dept] = []
    group_dept[dept].append(emp["name"])

    city = emp["city"]
    if city not in Group_salaries:
        Group_salaries[city] = []
    Group_salaries[city].append(emp["salary"])

print("Group names by department: ",group_dept)
print("Group salaries by city: ",Group_salaries)
print("*"*15)

# ⭐️ Pattern 9 – Searching
found=False
for emp in employees:
    if emp["id"] == 106:
        print("employee ID 106 FOUND:",emp)
        found = True
        break
if not found:
    print("NOT FOUND")

found=False
for emp in employees:
    if emp["city"] == "Pune":
        print("first employee from Pune:",emp)
        found = True
        break
if not found:
    print("NOT FOUND")
found=False
for emp in employees:
    if emp["dept"] == "HR" and emp["salary"] > 65000:
        print("first HR employee earning more than 65000:",emp)
        found = True
        break
if not found:
    print("NOT FOUND")
print("*"*15)

# ⭐️ Pattern 10 – Update/Delete/Merge
print("Before increase salary : ",employees)
for emp in employees:
    if emp["dept"] == "IT":
        emp["salary"] += 5000
print("Increase every IT salary by 5000: ",employees)

for emp in employees:
    if emp["dept"] == "Finance":
        emp["city"] = "Hyderabad"
print("Change Finance city to Hyderabad: ",employees)

print("*"*15)
new=[]
for emp in employees:
    if emp["salary"] > 60000:
        new.append(emp)
print("Delete salary below 60000: ",new)

print("*"*15)
updated_employees = employees + new_employees
print("Merge employees and new_employees: ",updated_employees)

print("*"*15)
for emp in updated_employees:
    if emp["name"] == "Aisha":
        print("Aisha: ",emp)
        break
print("*"*15)
final_gouup={}
for emp in updated_employees:
    dept = emp["dept"]
    if dept not in final_gouup:
        final_gouup[dept] = []
    final_gouup[dept].append(emp["name"])
print("Group names by department after merge: ",final_gouup)

```

```python
servers = [
    {"id":"i-101","name":"web01","team":"Frontend","status":"Running","cpu":65,"ram":8},
    {"id":"i-102","name":"db01","team":"Database","status":"Stopped","cpu":20,"ram":32},
    {"id":"i-103","name":"app01","team":"Backend","status":"Running","cpu":80,"ram":16},
    {"id":"i-104","name":"cache01","team":"Backend","status":"Running","cpu":35,"ram":8},
    {"id":"i-105","name":"proxy01","team":"Frontend","status":"Running","cpu":75,"ram":4},
    {"id":"i-106","name":"backup01","team":"Database","status":"Stopped","cpu":10,"ram":64},
    {"id":"i-107","name":"monitor01","team":"DevOps","status":"Running","cpu":55,"ram":8}
]

new_servers = [
    {"id":"i-108","name":"jenkins01","team":"DevOps","status":"Running","cpu":70,"ram":16},
    {"id":"i-109","name":"grafana01","team":"DevOps","status":"Running","cpu":40,"ram":8}
]
```

```python
# ⭐️ Pattern 1 – Highest / Lowest
Highest_CPU =  max(servers, key = lambda x:x["cpu"])
Lowest_RAM = min(servers, key = lambda x:x["ram"])

print("Highest CPU server: ",Highest_CPU )
print("Lowest RAM server: ", Lowest_RAM)
print("*"*15)

# ⭐ Pattern 2
Count_server = {}
Count_status = {}
for server in servers:
    team = server["team"]
    Count_server[team] = Count_server.get(team, 0)+1

    status= server["status"]
    Count_status[status] = Count_status.get(status, 0)+1

print("Count servers by team: ", Count_server)
print("Count Running vs Stopped: ", Count_status)
print("*"*15)


# ⭐ Pattern 3
for server in servers:
    if server["status"] == "Running":
        print("Running servers: ", server["id"],",",server["name"] )
print("*"*15)
for server in servers:
    if server["cpu"] > 50:
        print("Backend servers with CPU > 50: ", server["id"], ",",server["name"] )
print("*"*15)


# ⭐ Pattern 4
server_names =[]
CPU_values=[]
for server in servers:
    name = server["name"]
    server_names.append(name)

    cpu = server["cpu"]
    CPU_values.append(cpu)

print("list of server names: ", server_names)
print("list of CPU values: ", CPU_values)
print("*"*15)


# ⭐ Pattern 5
Total_CPU = sum(server["cpu"] for server in servers)
Average_CPU = Total_CPU /len(servers)

Total_RAM = sum(server["ram"] for server in servers)
Average_RAM = Total_RAM / len(servers)

print("Total CPU: ", Total_CPU)
print("Average CPU: ", Average_CPU)
print("Average RAM: ", Average_RAM)
print("*"*15)


# ⭐ Pattern 6
sort_cpu = servers.copy()
sort_cpu.sort(key = lambda x:x["cpu"])
print("Sort by CPU: ", sort_cpu)

sort_ram = servers.copy()
sort_ram.sort(key = lambda x:x["ram"], reverse =True)
print("Sort by RAM: ", sort_ram)

sort_team = servers.copy()
sort_team.sort(key = lambda x:(x["team"],x["cpu"]))
print("Sort by team then CPU: ", sort_team)
print("*"*15)

# ⭐ Pattern 7

for server in servers :
    if server["team"] == "Backend" and server["status"] == "Running" and server["cpu"] > 40:
        print("Running Backend servers with CPU > 40: ",server)

database=[]
for server in servers :
    if server["team"] == "Database" and server["status"] == "Stopped":
        database.append(server["name"])
print("Stopped Database servers: ",database)

front=[]
for server in servers :
    if server["team"] == "Frontend" and server["ram"] >= 8:
        front.append(server["name"])
print("senior IT employee names: ",front)
print("*"*15)
            
# ⭐ Pattern 8
Group_team = {}
for server in servers:
    team = server["team"]
    if team not in Group_team:
        Group_team[team]= []
    Group_team[team].append(server["name"])
print(Group_team)

Group_status = {}
for server in servers:
    status = server["status"]
    if status not in Group_status:
        Group_status[status]= []
    Group_status[status].append(server["cpu"])
print(Group_status)
print("*"*15)


# ⭐ Pattern 9
found =False
for server in servers:
    if server["id"] == "i-105":
        print("server i-105 FOUND: ",server)
        found = True
        break
if not found:
    print("not found")

found =False
for server in servers:
    if server["team"] == "DevOps":
        print("first DevOps server FOUND: ",server)
        found = True
        break
if not found:
    print("not found")

found =False
for server in servers:
    if server["team"] == "Database":
        print("first Running Database server FOUND: ",server)
        found = True
        break
if not found:
    print("not found")
print("*"*15)


# ⭐️ Pattern 10
for server in servers:
    if server["status"] == "Stopped":
        server["status"] = "Running"
print("Change every Stopped server to Running: ",servers)

for server in servers:
    if server["team"] == "Backend":
        server["cpu"] += 10
print("Increase Backend CPU by 10: ",servers)
print("*"*15)

new_cpu = []
for server in servers:
    if server["cpu"] > 30:
        new_cpu.append(server)
print("Delete CPU below 30: ",new_cpu)
print("*"*15)

servers.extend(new_servers)
print("Merge new_servers: ",servers)
print("*"*15)



found= False
for server in servers:
    if server["name"] == "grafana01":
        print("grafana01 FOUND: ", server)
        found=True
        break
if not found:
    print("NOT FOUND")
print("*"*15)

final_group_team = {}
for server in servers:
    team = server["team"]
    if team not in final_group_team:
        final_group_team[team] = []
    final_group_team[team].append(server["name"])
print("Group names by team after merge: ",final_group_team)



'''

27. Search grafana01.


28. Group names by team after merge.

'''


print("*"*15)
```
