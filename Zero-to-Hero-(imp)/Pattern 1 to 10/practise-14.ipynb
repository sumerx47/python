Question 1 — Employee Performance

Create functions using dictionary comprehensions:

1. get_employee_names() → {employee_id: name}


2. get_averages() → {employee_id: average_score}


3. get_high_performers() → employees with average ≥ 85


4. get_python_employees() → employees who have Python


5. get_status() → {employee_id: "Pass"/"Fail"}


6. get_department() → {employee_id: department}


7. get_name_salary_style_result() → create a useful dictionary from the data.

```python
employees = {
    "E101": {
        "name": "Ali",
        "department": "DevOps",
        "skills": ["Python", "Linux"],
        "scores": [80, 85, 90]
    },

    "E102": {
        "name": "Sara",
        "department": "Cloud",
        "skills": ["AWS", "Python"],
        "scores": [92, 95, 90]
    },

    "E103": {
        "name": "Ahmed",
        "department": "DevOps",
        "skills": ["Linux", "Docker"],
        "scores": [70, 75, 72]
    },

    "E104": {
        "name": "Zoya",
        "department": "Cloud",
        "skills": ["AWS", "Docker"],
        "scores": [88, 91, 85]
    }
}
```

```python
# 1. get_employee_names() → {employee_id: name}
def get_employee_names():
    return {
        employee_id: details["name"] for employee_id, details in employees.items()
    }
result = get_employee_names()
print("employee id and names: ", result)
print("*" * 15)

print("CALCULATE AVERAGE FUNCTION")
def calculate_average(score):
    avg = sum(score) / len(score)
    return avg


print("*" * 15)
# 2. get_averages() → {employee_id: average_score}
def get_averages():
    return {
        employee_id: calculate_average(details["scores"]) for employee_id, details in employees.items()
    }
result1 = get_averages()
for employee_id, average in result1.items():
    print(employee_id ,"->", round(average, 2))
print("*" * 15)

# 3. get_high_performers() → employees with average ≥ 85
def get_high_performers():
    return {
        emp_id:(details["name"],calculate_average(details["scores"])) for emp_id, details in employees.items() if calculate_average(details["scores"]) >= 85
    }
result2 = get_high_performers()
print("employees with average ≥ 85: ", result2)
print("*" * 15)

# 4. get_python_employees() → employees who have Python
def get_python_employees():
    return {
        emp_id:details["name"] for emp_id, details in employees.items() if "Python" in details["skills"]
    }
result3 = get_python_employees()
print("employees who have Python: ", result3)
print("*" * 15)


# 5. get_status() → {employee_id: "Pass"/"Fail"}
def get_status():
    return {
       emp_id:"pass" if calculate_average(details["scores"]) >= 80 else "Fail" for emp_id, details in employees.items()
    }
result4 = get_status()
print("employee_id: Pass/Fail: ", result4)
print("*" * 15) 

# 6. get_department() → {employee_id: department}
def get_department():
    return {
        emp_id:details["department"] for emp_id, details in employees.items()
    }
result5 = get_department()
print("get department: ", result5)
print("*" * 15) 

# 7. get_name_salary_style_result() → create a useful dictionary from the data.
def get_name_salary_style_result():
    pass


```

Question 2 — Product Analysis

Create functions using dictionary comprehensions:

1. {product_id: product_name}


2. {product_id: price}


3. Products with price ≥ 10000


4. Products currently in stock


5. {product_id: "Available"/"Out of Stock"}


6. Apply a 10% discount to every product.


7. Create {product_name: price}.


8. Create a dictionary containing only products costing between ₹2,000 and ₹20,000.

```python
products = {
    "P101": {"name": "Laptop", "price": 70000, "stock": 5},
    "P102": {"name": "Mouse", "price": 1200, "stock": 25},
    "P103": {"name": "Keyboard", "price": 2500, "stock": 0},
    "P104": {"name": "Monitor", "price": 18000, "stock": 8}
}
```

```python
# 1. {product_id: product_name}
def get_products_names():
    return {
        products_id: details["name"] for products_id, details in products.items()
    }
result = get_products_names()
print("products id and names: ", result)
print("*" * 15)

# 2. {product_id: price}
def get_price_names():
    return {
        product_id:details["price"] for product_id, details in products.items()
    }
result1 = get_price_names()
print("products id and price: ", result1)
print("*" * 15)

# 3. Products with price ≥ 10000
def get_products_price():
    return {
        product_id:(details["name"],details["price"]) for product_id,details in products.items() if details["price"] >= 10000
    }
result2 = get_products_price()
print("products price: ", result2)
print("*" * 15)

# 4. Products currently in stock
def get_products_instock():
    return {
        product_id:(details["name"],details["stock"]) for product_id , details in products.items() if details["stock"] > 0
    }
result3 = get_products_instock()
print("products instock: ", result3)
print("*" * 15)

# 5. {product_id: "Available"/"Out of Stock"}
def get_products_stock():
    return {
        product_id: "Available" if details["stock"] > 0 else "Out of Stock" for product_id , details in products.items()
    }
result4 = get_products_stock()
print("products Available/Out of Stock: ", result4)
print("*" * 15)

# 6. Apply a 10% discount to every product.
def get_products_discount():
    return {
        product_id:details["price"]*0.90 for product_id , details in products.items()
    }
result5 = get_products_discount()
print("products discount by 10%: ", result5)
print("*" * 15)

# 7. Create {product_name: price}.
def get_products_names_price():
    return {
        product_id:details["price"] for product_id , details in products.items()
    }
result6 = get_products_names_price()
print("products id and names and prices: ", result6)
print("*" * 15)

# 8. Create a dictionary containing only products costing between ₹2,000 and ₹20,000.
def get_products_dictionary():
    return {
        product_id:(details["name"],details["price"]) for product_id , details in products.items() if details["price"] >= 2000 and details["price"] <= 20000
    }
result7 = get_products_dictionary()
print("pproducts costing between ₹2,000 and ₹20,000: ", result7)
print("*" * 15)


```
