Question 5

Create a list of .py files.

```python
files = [
    "main.py",
    "notes.txt",
    "docker.yaml",
    "app.py",
    "config.json"
]
```

```python
py_files = []
yaml_file = []

for file in files:
    extension = file.split(".")[1]
    if extension == "py":
        py_files.append(file)


for file in files:  
    if file.endswith(".yaml"):
       yaml_file.append(file)      

print("py files are: ",py_files)
print("yaml files are: ",yaml_file)
```

Question 11 – Employee Records


Create:

1. List of IT employee names.


2. List of employee names earning above ₹55,000.


3. List of employees knowing Python.


4. List of employees knowing Docker.


5. List of employees knowing Excel.


6. List of salaries below ₹50,000.


7. List of Finance employee names.

```python
employees = [
    {
        "name":"Ali",
        "dept":"IT",
        "salary":50000,
        "skills":["Python","Linux"]
    },
    {
        "name":"Sara",
        "dept":"HR",
        "salary":65000,
        "skills":["Excel"]
    },
    {
        "name":"Ahmed",
        "dept":"IT",
        "salary":70000,
        "skills":["Python","Docker"]
    },
    {
        "name":"John",
        "dept":"Finance",
        "salary":45000,
        "skills":["Excel","PowerBI"]
    },
    {
        "name":"Fatima",
        "dept":"IT",
        "salary":60000,
        "skills":["Linux","Docker"]
    }
]
```

```python
employee_IT = []
emp_sal = []
empl_sal =[]
empl_python =[]
empl_docker=[]
empl_git=[]
employee_finance = []

for emp in employees:
    if "IT" in emp["dept"]:
        employee_IT.append(emp["name"])
    

    if emp["salary"] > 50000:
        emp_sal.append(emp["name"])

    if emp["salary"] < 50000:
            empl_sal.append(emp["name"])

    skill = emp["skills"]
    if "Python" in skill:
        empl_python.append(emp["name"])
    if "Docker" in skill:
            empl_docker.append(emp["name"])
    if "Excel" in skill:
            empl_git.append(emp["name"])

    if emp["dept"] == "Finance" :
          employee_finance.append(emp["name"])

print("employees in IT: ",employee_IT)
print("employee salary above >50000: ",emp_sal)
print("employee salary below <50000: ",empl_sal)
print("employees knowing Python: ",empl_python)
print("employees knowing Docker: ",empl_docker)
print("employees knowing Git: ",empl_git)
print("employees in Finance: ",employee_finance)
```
