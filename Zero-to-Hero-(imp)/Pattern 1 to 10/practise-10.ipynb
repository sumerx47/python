🧩 Question 1 — Employee Analytics System

Given:



Build a program that performs all of the following:

1. Highest / Lowest

Find:

Employee with the highest salary

Employee with the lowest salary

Employee with the highest experience


2. Counting

Count how many employees belong to each department.

3. Filtering

Create a list of employees who:

salary >= 60000
AND
experience >= 5

4. Create New List

Create a list containing:

"name - department"

For example:

["Ali - IT", "Sara - HR", ...]

5. Sum / Average

Calculate:

Total salary of all employees

Average salary

Average salary of IT employees only


6. Sorting

Create:

Employees sorted by salary descending

Employees sorted by experience descending


7. Multiple Conditions

Find employees who satisfy:

IT department
AND salary > 60000
AND experience >= 5

8. Grouping

Create a dictionary grouping employees by department:

{
    "IT": [...],
    "HR": [...],
    "Finance": [...]
}

9. Searching

Search for employee with:

id = 106

Print the complete employee dictionary.

Also search for an employee by name "Ayesha".

10. Update 

Give every IT employee with salary below 60000 a ₹5000 raise.

11. Delete

Remove employees whose:

experience < 3

12. Merge

You are given:

new_employees = [
    {"id": 109, "name": "Bilal", "dept": "IT", "salary": 70000, "exp": 4},
    {"id": 110, "name": "Mariam", "dept": "HR", "salary": 58000, "exp": 3}
]

Merge them into the existing employee collection.

⭐️ Final challenge

After all updates/deletions/merging, print the top 3 highest-paid employees, showing only:

Name | Department | Salary


---

```python
employees = [
    {"id": 101, "name": "Ali",   "dept": "IT",      "salary": 65000, "exp": 3},
    {"id": 102, "name": "Sara",  "dept": "HR",      "salary": 52000, "exp": 5},
    {"id": 103, "name": "Ahmed", "dept": "IT",      "salary": 80000, "exp": 6},
    {"id": 104, "name": "Zoya",  "dept": "Finance", "salary": 72000, "exp": 4},
    {"id": 105, "name": "Omar",  "dept": "IT",      "salary": 55000, "exp": 2},
    {"id": 106, "name": "Hina",  "dept": "HR",      "salary": 60000, "exp": 7},
    {"id": 107, "name": "Usman", "dept": "Finance", "salary": 90000, "exp": 8},
    {"id": 108, "name": "Ayesha","dept": "IT",      "salary": 75000, "exp": 5},
]
```

```python
# 1. Highest / Lowest
highest = employees[0]
lowest = employees[0]
high_exp= employees[0]


for emp in employees:
    if emp["salary"] > highest["salary"]:
      highest = emp
    if emp["salary"] < lowest["salary"]:
          lowest = emp
    if emp["exp"] > high_exp["exp"]:
        high_exp = emp
                             
print("Employee with the highest salary: ", highest["name"])
print("Employee with the lowest salary: ", lowest)
print("Employee with the highest experience: ", high_exp)
print("*"*15)


# 2. Counting
count={}
for emp in employees:
    dept = emp["dept"]
    if dept in count:
        count[dept] += 1
    else:
        count[dept] = 1

print("employees belong to each department: ", count)
print("*"*15)
        
# 3. Filtering
emp_list = []
for emp in employees:
    if emp["salary"] >= 60000 and emp["exp"] >= 5:
        emp_list.append(emp["name"])

print("employees employees whose salary >= 60000 and experience >= 5: ", emp_list)
print("*"*15)

# 4. Create New List
dept_list = []
for emp in employees:
    dept_list.append(f"{emp["name"]} - {emp["dept"]}")
print("list containing name - department: ",dept_list)
print("*"*15)


# 5. Sum / Average
total = 0
for emp in employees:
    total += emp["salary"]
avg = total / len(employees)

print("Total salary of all employees: ",total)
print("avg salary: ",avg)
print("*"*15)


# 6. Sorting
ascending_salary = sorted(employees, key = lambda emp:emp["salary"] , reverse = True)
descending_experience = sorted(employees, key = lambda emp:emp["exp"] , reverse = True)

print("Employees sorted by salary descending: ",ascending_salary)
print("Employees sorted by experience descending: ",descending_experience)
print("*"*15)


# 7. Multiple Conditions
for emp in employees:
    if emp["dept"] == "IT" and emp["salary"] > 60000 and emp["exp"] >=5:
        print("employees in IT department AND salary>60000 AND experience>=5: ", emp["name"])
print("*"*15)


# 8. Grouping
group = {}
for emp in employees:
    dept = emp["dept"]
    if dept not in group:
        group[dept] = []
    group[dept].append(emp["name"])

print("dictionary grouping employees by department: ",group)
print("*"*15)


# 9. searching
found = False
for emp in employees:
    if emp["id"] == 106:
        print("employee with id = 106: ",emp)
        found = True
        break
if not found :
    print("NOT FOUND")
print("*"*15)

found = False
for emp in employees:
    if emp["name"] == "Ayesha":
        print("employee with name = Ayesha: ",emp)
        found = True
        break
if not found :
    print("NOT FOUND")
print("*"*15)


# 10. Update 
for emp in employees:
    if emp["dept"] == "IT" and emp["salary"] < 60000 : 
        emp["salary"] += 5000
print(employees)
print("*"*15)

# 11. Delete
new_list = []
for emp in employees:
    if emp["exp"] > 3:
        new_list.append(emp) 
print("Remove employees whose experience < 3: ",new_list)
print("*"*15)


# 12. Merge

new_employees = [
    {"id": 109, "name": "Bilal", "dept": "IT", "salary": 70000, "exp": 4},
    {"id": 110, "name": "Mariam", "dept": "HR", "salary": 58000, "exp": 3}
]

updated_emp = employees + new_employees

print("Merge them into the existing employee collection.: ",updated_emp)
print("*"*15)



# ⭐️ Final challenge

top_3 = sorted(updated_emp, key =lambda emp:emp["salary"], reverse = True) [:3]
for emp in top_3:
    print(emp["name"], "|",emp["dept"],"|", emp["salary"])
```
