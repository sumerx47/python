Write a program that:

1. Removes duplicate servers based on name.

```python
servers = [
    {"name": "web-01", "ip": "10.0.0.1", "cpu": 85, "memory": 72, "status": "UP"},
    {"name": "web-02", "ip": "10.0.0.2", "cpu": 45, "memory": 55, "status": "UP"},
    {"name": "web-01", "ip": "10.0.0.1", "cpu": 91, "memory": 88, "status": "UP"},
    {"name": "api-01", "ip": "10.0.0.3", "cpu": 95, "memory": 92, "status": "DOWN"},
    {"name": "db-01", "ip": "10.0.0.4", "cpu": 78, "memory": 89, "status": "UP"},
    {"name": "api-01", "ip": "10.0.0.3", "cpu": 82, "memory": 90, "status": "DOWN"},
    {"name": "cache-01", "ip": "10.0.0.5", "cpu": 30, "memory": 40, "status": "UP"},
]
```

```python
def analyze_server(servers):
    '''1. Removes duplicate servers based on name.'''
    unique_servers = {}

    for server in servers:
        try:
            name = server["name"]
            cpu = int(server["cpu"])

            if name not in unique_servers:
                unique_servers[name] = server
            elif cpu > int(unique_servers[name]["cpu"]):
                unique_servers[name] = server
           
        except(KeyError,TypeError,ValueError):
            print("skipping invalid server:" , server)

        servers = list(unique_servers.values())

        if not servers:
            return {"error": "no valid servers found"}

    ''' 2. If a server appears multiple times, keep the record with the highest CPU.
        3. Find the server with the highest CPU.'''
    highest_cpu_server = max(servers, key = lambda server: int(server["cpu"]))


    '''4. Find the server with the lowest CPU.'''
    lowest_cpu_server = min(servers, key = lambda server : int(server["cpu"]))

    '''5. Find average CPU.'''
    #total_cpu = sum(int(server["cpu"]) for server in servers)
    #average_cpu = total_cpu/len(servers)

    '''6. Find all servers where: CPU >= 80 OR memory >= 80.'''
    #overloaded = [server for server in servers if int(server["cpu"]) >= 80 or int(servers["memory"]) >= 80]

    '''7. Find all DOWN servers.'''
    down_servers = [server for server in servers if server["status"].upper() == "DOWN" ]

    '''8. Sort the final servers by CPU descending.'''
    sorted_servers = sorted(servers, key = lambda server : int(server["cpu"]), reverse = True)


    '''9. Produce a final report.'''

    return {
        "total_servers": len(servers),
        "highest_cpu_server":highest_cpu_server,
        "lowest_cpu_server":lowest_cpu_server,
        #"average_cpu": round(average_cpu,2),
        #"overloaded":overloaded,
        "down_servers":down_servers,
        "sorted_servers":sorted_servers
        

    }

report = analyze_server(servers)
print("1. total_servers: ", report["total_servers"])
print("2. highest cpu server: ", report["highest_cpu_server"])
print("3. lowest cpu server: ", report["lowest_cpu_server"])
#print("4. average_cpu: ", report["average_cpu"])
#print("5. overloaded: ", report["overloaded"])
print("6. down_servers: ", report["down_servers"])
print("7. sorted_servers: ", report["sorted_servers"])


```

```python
# ============================================================
# QUESTION 2 ANSWER
# Production Log Analyzer
#
# Concepts:
# Regex + Functions + Dictionaries + Sets +
# Remove Duplicates + Counting + Grouping +
# Filtering + Highest + Sorting
# ============================================================

import re


logs = [
    "2026-09-24 10:01:20 ERROR web-01 CPU=91% message='Database timeout'",
    "2026-09-24 10:02:10 INFO web-02 CPU=45% message='Request completed'",
    "2026-09-24 10:03:15 ERROR web-01 CPU=95% message='Database timeout'",
    "2026-09-24 10:04:30 WARNING api-01 CPU=88% message='High CPU'",
    "2026-09-24 10:05:20 ERROR api-01 CPU=88% message='Connection failed'",
    "2026-09-24 10:06:00 ERROR web-02 CPU=85% message='Database timeout'",
    "2026-09-24 10:07:00 ERROR web-01 CPU=95% message='Database timeout'",
]


def analyze_logs(logs):

    # --------------------------------------------------------
    # Regex pattern
    # --------------------------------------------------------

    pattern = re.compile(
        r"(?P<timestamp>\d{4}-\d{2}-\d{2} "
        r"\d{2}:\d{2}:\d{2}) "
        r"(?P<level>INFO|ERROR|WARNING) "
        r"(?P<server>[\w-]+) "
        r"CPU=(?P<cpu>\d+)% "
        r"message='(?P<message>[^']+)'"
    )

    parsed_errors = []

    # --------------------------------------------------------
    # Parse logs
    # --------------------------------------------------------

    for line in logs:
        match = pattern.search(line)

        if not match:
            continue

        data = match.groupdict()

        # Only ERROR logs
        if data["level"] != "ERROR":
            continue

        data["cpu"] = int(data["cpu"])

        parsed_errors.append(data)

    # --------------------------------------------------------
    # Remove duplicate error events
    #
    # We define duplicate as:
    # server + message + CPU
    # --------------------------------------------------------

    seen = set()
    unique_errors = []

    for error in parsed_errors:

        key = ( 
            error["server"],
            error["message"],
            error["cpu"]
        )

        if key not in seen:
            seen.add(key)
            unique_errors.append(error)

    # --------------------------------------------------------
    # Count errors per server
    # Pattern 2 - Counting
    # --------------------------------------------------------

    error_count = {}

    for error in unique_errors:

        server = error["server"]

        error_count[server] = (
            error_count.get(server, 0) + 1
        )

    # --------------------------------------------------------
    # Group errors by server
    # Pattern 8 - Grouping
    # --------------------------------------------------------

    grouped = {}

    for error in unique_errors:

        server = error["server"]

        grouped.setdefault(server, []).append(error)

    # --------------------------------------------------------
    # Highest CPU error
    # Pattern 1 - Highest
    # --------------------------------------------------------

    highest_cpu_error = max(
        unique_errors,
        key=lambda error: error["cpu"],
        default=None
    )

    # --------------------------------------------------------
    # High CPU errors
    # Pattern 3 - Filtering
    # --------------------------------------------------------

    high_cpu_errors = [
        error
        for error in unique_errors
        if error["cpu"] >= 80
    ]

    # --------------------------------------------------------
    # Sort servers by error count
    # Pattern 6 - Sorting
    # --------------------------------------------------------

    sorted_servers = sorted(
        error_count.items(),
        key=lambda item: item[1],
        reverse=True
    )

    # --------------------------------------------------------
    # Final report
    # --------------------------------------------------------
    return {
        "total_unique_errors": len(unique_errors),
        "errors_per_server": error_count,
        "grouped_errors": grouped,
        "highest_cpu_error": highest_cpu_error,
        "high_cpu_errors": high_cpu_errors,
        "servers_by_error_count": sorted_servers
    }


report = analyze_logs(logs)

#print(report)
print("1. unique_errors: ", report["total_unique_errors"])
print("2. error_count: ", report["errors_per_server"])
print("3. grouped: ", report["grouped_errors"])
print("4. highest_cpu: ", report["highest_cpu_error"])
print("5. highest_cpu_error: ", report["high_cpu_errors"])
print("6. sorted_servers: ", report["servers_by_error_count"])

```
