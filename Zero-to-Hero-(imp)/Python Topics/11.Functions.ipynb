Create a function that receives a number and returns whether it is:

Even
Odd


```python
def check_even_odd(number):
    if number % 2 == 0:
        return "Even"
    else :
        return "Odd"
print(check_even_odd(7))
```

Q5

Create a function that receives a list of numbers and returns the largest number.

Example:

numbers = [10, 50, 20, 80, 30]

Expected:

80

```python
def find_largest(numbers):
    largest = numbers[0]

    for num in numbers:
        if num > largest:
            largest = num

    return largest
    
numbers = [10,20,20,40,90,12,45,77]

print(find_largest(numbers))
```
