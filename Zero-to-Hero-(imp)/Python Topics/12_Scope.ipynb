Q4. Write a function that attempts to increment a global counter. First solve it using global, then rewrite it without global.




```python
count = 0
def increment():
    global count
    count += 1

increment()
increment()
increment()
print(count)
```

Q7. Create a counter using a closure:

counter = create_counter()

counter()  # 1
counter()  # 2
counter()  # 3



```python
def create_counter():
    count = 0
    def counter():
        nonlocal count
        count += 1
        return count
    return counter

counter = create_counter()

print(counter())
print(counter())
print(counter())
```

Q12. Build a closure that remembers an employee's department:

it_employee = create_employee_checker("IT")

it_employee({"name": "Ali", "dept": "IT"})    # True
it_employee({"name": "Sara", "dept": "HR"})   # False

```python
def create_employee_checker(department):
    def check_employee(employee):
        return employee["dept"] == department
    return check_employee 

it_employee = create_employee_checker("IT")

print(it_employee({"name": "Ali", "dept": "IT"})) # True
print(it_employee({"name": "Sara", "dept": "HR"}))   # False
```
