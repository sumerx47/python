Q5

Create a function that receives a list of numbers and returns the largest number.

Example:

numbers = [10, 50, 20, 80, 30]

Expected:

80

```python
def find_largest(numbers):
    largest = numbers[0]

    for num in numbers:
        if num > largest:
            largest = num

    return largest
    
numbers = [10,20,20,40,90,12,45,77]

print(find_largest(numbers))
```

```python
def find_largest(numbers):
    return max(numbers)

numbers = [10,20,20,40,90,12,45,77,555,254]

print(find_largest(numbers))
```

Q11

Create:

find_highest_employee(employees)

using employee dictionaries:

employees = [
    {"name": "Ali", "salary": 60000},
    {"name": "Ahmed", "salary": 80000},
    {"name": "Sara", "salary": 70000}
]

Return the complete employee dictionary with the highest salary.


```python
def find_highest_employee(employees):
    highest = employees[0]

    for emp in employees:
        salary = emp["salary"]
        if salary > highest["salary"]:
            highest = emp
    return highest

employees = [
    {"name": "Ali", "salary": 60000},
    {"name": "Ahmed", "salary": 80000},
    {"name": "Sara", "salary": 70000}
]


print(find_highest_employee(employees))
```

```python
def find_highest_employee(employees):
    return max(employees, key = lambda x:x["salary"])

employees = [
    {"name": "Ali", "salary": 60000},
    {"name": "Ahmed", "salary": 80000},
    {"name": "Sara", "salary": 90000}
]


print(find_highest_employee(employees))
```

Q12

Create:

group_by_department(employees)

Expected structure:

{
    "IT": [...],
    "HR": [...]
}



```python
def group_by_department(employees):
    group = {}

    for emp in employees:
        group.setdefault(emp["dept"],[]).append(emp)

    return group

employees = [
    {"name": "Ali", "dept" : "IT" ,"salary": 60000},
    {"name": "Ahmed", "dept" : "IT" , "salary": 80000},
    {"name": "Sara", "dept" : "HR" , "salary": 90000}
]

print(group_by_department(employees))
```

Return employees whose salary is greater than or equal to the specified salary.




```python
def filter_employees(employees, minimum_salary):
    org = []

    for emp in employees:
        if emp["salary"] >= minimum_salary:
            org.append(emp)

    return org

filter_employees(employees, 70000)
```

Create:

```python
employees = [
    {"name": "Ali", "dept" : "IT" ,"salary": 60000},
    {"name": "Ahmed", "dept" : "IT" , "salary": 80000},
    {"name": "Sara", "dept" : "HR" , "salary": 90000},
    {"name": "mary", "dept" : "IT" , "salary": 80000},
    {"name": "hameed", "dept" : "IT" , "salary": 80000}
]


def count_salary(employees):
    count={}

    for emp in employees:
        salary = emp["salary"]
        count[salary] = count.get(salary , 0 ) + 1 
    return count

print(count_salary(employees))
```


Q18 — Recursion

Write:

factorial(n)

using recursion.

factorial(5) → 120



```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n-1)
print(factorial(5))
```

Q19 — Closure

Create:

create_multiplier(n)

such that:

double = create_multiplier(2)

print(double(10))

outputs:

20

and:

triple = create_multiplier(3)

print(triple(10))

outputs:

30


```python
def create_multiplier(n):
    def multiplier(x):
        return x * n
    return multiplier


double = create_multiplier(2)
print(double(10))


triple = create_multiplier(3)
print(triple(10))
```
