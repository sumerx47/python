Level 5 — Nested Data + Previous Patterns ⭐️⭐️⭐️


```python
servers = [
    {
        "name": "web01",
        "metrics": {
            "cpu": 85,
            "memory": 72
        },
        "services": [
            {"name": "nginx", "status": "running"},
            {"name": "docker", "status": "running"}
        ]
    },
    {
        "name": "web02",
        "metrics": {
            "cpu": 45,
            "memory": 60
        },
        "services": [
            {"name": "nginx", "status": "running"},
            {"name": "docker", "status": "stopped"}
        ]
    }
]
```

```python
frequency = {}

for server in servers:
    for service in server["services"]:
        status = service["status"]

        frequency[status] = frequency.get(status,0)+1
print(frequency)

```

```python
stopped_count = {}

for server in servers:
    count = 0

    for service in server["services"]:
        if service["status"] == "stopped":
            count += 1

    stopped_count[server["name"]] = count

server_name =max(stopped_count, key = stopped_count.get)
print(server_name)
```

Level 8 — Nested Data + Regex ⭐️⭐️⭐️⭐️


```python
import re
logs = """
server=web01 service=nginx status=500
server=web02 service=docker status=200
server=web01 service=nginx status=503
"""

server_frequency = {}

for line in logs.splitlines():
    match = re.search(r"server=(\w+)",line)

    if match:
        server = match.group(1)
        server_frequency[server] = (server_frequency.get(server,0)+1)
print(server_frequency)
```

🔥 Level 10 — Real DevOps Nested Data


```python
servers = [
    {
        "name": "web01",
        "environment": "production",
        "metrics": {
            "cpu": 85,
            "memory": 91
        },
        "services": [
            {"name": "nginx", "status": "running"},
            {"name": "docker", "status": "running"}
        ]
    },
    {
        "name": "web02",
        "environment": "production",
        "metrics": {
            "cpu": 45,
            "memory": 60
        },
        "services": [
            {"name": "nginx", "status": "running"},
            {"name": "docker", "status": "stopped"}
        ]
    },
    {
        "name": "dev01",
        "environment": "development",
        "metrics": {
            "cpu": 92,
            "memory": 88
        },
        "services": [
            {"name": "nginx", "status": "stopped"},
            {"name": "docker", "status": "running"}
        ]
    }
]
```

```python
environment_frequency = {}
high_cpu = []
high_memory = []
highest_cpu = None
stopped_services = 0
servers_with_stopped_services = []

for server in servers:
    name = server["name"]
    environment = server["environment"]
    cpu = server["metrics"]["cpu"]
    memory = server["metrics"]["cpu"]

    # 1
    environment_frequency[environment] = environment_frequency.get(environment,0)+1

    # 2
    if cpu >= 80:
        high_cpu.append(name)

    # 3
    if memory >= 80:
        high_memory.append(name)

    # 4
    if (highest_cpu is None or cpu > highest_cpu["metrics"]["cpu"]):
        highest_cpu = server

    # 5
    has_stopped_service = False

    for service in server["services"]:
        if service["status"] == "stopped":
            stopped_services += 1
            has_stopped_service =True

    if has_stopped_service:
        servers_with_stopped_services.append(name)

report = {
    "environment_frequency":environment_frequency,
    "high_cpu":high_cpu,
    "high_memory":high_memory,
    "highest_cpu":highest_cpu["name"],
    "stopped_services":stopped_services,
    "servers_with_stopped_services":servers_with_stopped_services
}

print(report)

print("environment frequency:", report["environment_frequency"])
print("high cpu:", report["high_cpu"])
print("high memory:", report["high_memory"])
print("highest cpu:", report["highest_cpu"])
print("stopped services:", report["stopped_services"])
print("servers with stopped services:", report["servers_with_stopped_services"])



```

### Q23 — Nested Data + Deduplication ⭐⭐⭐

```python
servers = [
    {"id": 101, "name": "web01", "cpu": 70},
    {"id": 102, "name": "web02", "cpu": 50},
    {"id": 101, "name": "web01", "cpu": 85},
    {"id": 103, "name": "web03", "cpu": 90},
    {"id": 102, "name": "web02", "cpu": 65}
]
```

Question: Remove duplicate server IDs while keeping the record with the highest CPU.

```python
best_servers = {}

for server in servers:
    server_id = server["id"]

    if server_id not in best_servers or server["cpu"] > best_servers[server_id]["cpu"]:
        best_servers[server_id] = server

result = list(best_servers.values())
print(result)
       
```
