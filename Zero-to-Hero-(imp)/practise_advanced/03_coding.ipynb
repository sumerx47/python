```python
servers = [
    {
        "name": "web01",
        "environment": "production",
        "metrics": {
            "cpu": 85,
            "memory": 91
        },
        "services": [
            {"name": "nginx", "status": "running"},
            {"name": "docker", "status": "running"}
        ]
    },
    {
        "name": "web02",
        "environment": "production",
        "metrics": {
            "cpu": 45,
            "memory": 60
        },
        "services": [
            {"name": "nginx", "status": "running"},
            {"name": "docker", "status": "stopped"}
        ]
    },
    {
        "name": "dev01",
        "environment": "development",
        "metrics": {
            "cpu": 92,
            "memory": 88
        },
        "services": [
            {"name": "nginx", "status": "stopped"},
            {"name": "docker", "status": "running"}
        ]
    }
]
```

```python
environment_frequency = {}
high_cpu = []
high_memory = []
highest_cpu = None
stopped_services = 0
servers_with_stopped_services = []

for server in servers:
    name = server["name"]
    environment = server["environment"]
    cpu = server["metrics"]["cpu"]
    memory = server["metrics"]["cpu"]

    # 1
    environment_frequency[environment] = environment_frequency.get(environment,0)+1

    # 2
    if cpu >= 80:
        high_cpu.append(name)

    # 3
    if memory >= 80:
        high_memory.append(name)

    # 4
    if (highest_cpu is None or cpu > highest_cpu["metrics"]["cpu"]):
        highest_cpu = server

    # 5
    has_stopped_service = False

    for service in server["services"]:
        if service["status"] == "stopped":
            stopped_services += 1
            has_stopped_service =True

    if has_stopped_service:
        servers_with_stopped_services.append(name)

report = {
    "environment_frequency":environment_frequency,
    "high_cpu":high_cpu,
    "high_memory":high_memory,
    "highest_cpu":highest_cpu["name"],
    "stopped_services":stopped_services,
    "servers_with_stopped_services":servers_with_stopped_services
}

print(report)

print("environment frequency:", report["environment_frequency"])
print("high cpu:", report["high_cpu"])
print("high memory:", report["high_memory"])
print("highest cpu:", report["highest_cpu"])
print("stopped services:", report["stopped_services"])
print("servers with stopped services:", report["servers_with_stopped_services"])



```

Question: Create a report containing:

```python
total_nodes = len(cluster["nodes"])

total_pods = 0
running_pods = 0
failed_pods = 0
pending_pods = 0
failed_pod_node = None


for node in cluster["nodes"]:
    for pod in node["pods"]:
        total_pods += 1

        status = pod["status"]
        if status == "Running":
            running_pods += 1

        elif status == "Failed":
            failed_pods += 1
            failed_pod_node = node["name"]

        elif status == "Pending":
            pending_pods += 1

report = {
    "total_nodes":total_nodes,
    "total_pods":total_pods,
    "running_pods":running_pods,
    "failed_pods":failed_pods,
    "pending_pods":pending_pods,
    "failed_pod_node":failed_pod_node
}

print(report)
```

```python
docker = {
    "containers": [
        {
            "name": "nginx",
            "image": "nginx:latest",
            "state": {
                "status": "running"
            },
            "resources": {
                "cpu": 20,
                "memory": 512
            }
        },
        {
            "name": "api",
            "image": "myapi:v2",
            "state": {
                "status": "exited"
            },
            "resources": {
                "cpu": 10,
                "memory": 256
            }
        },
        {
            "name": "worker",
            "image": "worker:v1",
            "state": {
                "status": "running"
            },
            "resources": {
                "cpu": 75,
                "memory": 1024
            }
        }
    ]
}
```

Question: Find:

1. Running containers.
2. Containers using CPU ≥ 50.
3. Container using the most memory.
4. Total memory used by all containers.

```python
running_containers = []
high_cpu_containers = []

highest_memory_container = None
total_memory = 0

for container in docker["containers"]:
   status = container["state"]["status"]
   name = container["name"]
   cpu = container["resources"]["cpu"]
   memory = container["resources"]["memory"]

   total_memory += memory

   if status == "running":
      running_containers.append(name)

   if cpu >= 50:
      high_cpu_containers.append(name)

   if (highest_memory_container is None or memory > highest_memory_container["resources"]["memory"]):
      highest_memory_container = container

report = {
   "running_containers":running_containers,
   "high_cpu_containers":high_cpu_containers,
   "highest_memory_container":highest_memory_container["name"],
   "total_memory":total_memory
}

print(report)
      
```

#### Q29 — Nested Data + Regex + Files + Generator 🔥

Question:

Create a generator that reads the file and yields structured dictionaries:

{
    "server": "web01",
    "level": "ERROR",
    "code": 500
}

```python
logs = """
server=web01 level=ERROR code=500
server=web02 level=INFO code=200
server=web01 level=ERROR code=503
server=web03 level=ERROR code=500
server=web02 level=WARNING code=404
server=web01 level=ERROR code=500
"""
```

```python
import re

def parse_logs(file_path):
    pattern = re.compile(r"server=(\w+)\s+level=(\w+)\s+code=(\d+)")

    with open(file_path) as file:
        for line in file:
            match = re.search(line)

        if match:
            yield {
                "server":match.group(1),
                "level":match.group(2),
                "code":match.group(3)
            }

error_count = {}

for log in parse_logs("app.log"):
    if log["level"] == "ERROR":
        server = log["server"]
        error_count[log] = error_count.get(log,0)+1
print(error_count)
```
