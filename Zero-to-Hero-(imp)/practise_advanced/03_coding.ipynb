## pattern 13 - Nested data

Level 5 — Nested Data + Previous Patterns ⭐️⭐️⭐️


```python
servers = [
    {
        "name": "web01",
        "metrics": {
            "cpu": 85,
            "memory": 72
        },
        "services": [
            {"name": "nginx", "status": "running"},
            {"name": "docker", "status": "running"}
        ]
    },
    {
        "name": "web02",
        "metrics": {
            "cpu": 45,
            "memory": 60
        },
        "services": [
            {"name": "nginx", "status": "running"},
            {"name": "docker", "status": "stopped"}
        ]
    }
]
```

```python
frequency = {}

for server in servers:
    for service in server["services"]:
        status = service["status"]

        frequency[status] = frequency.get(status,0)+1
print(frequency)

```

```python
stopped_count = {}

for server in servers:
    count = 0

    for service in server["services"]:
        if service["status"] == "stopped":
            count += 1

    stopped_count[server["name"]] = count

server_name =max(stopped_count, key = stopped_count.get)
print(server_name)
```

Level 6 — Nested Data + Functions ⭐️⭐️⭐️


```python
servers = [
    {
        "name": "web01",
        "metrics": {
            "cpu": 85,
            "memory": 72
        },
        "services": [
            {"name": "nginx", "status": "running"},
            {"name": "docker", "status": "running"}
        ]
    },
    {
        "name": "web02",
        "metrics": {
            "cpu": 45,
            "memory": 60
        },
        "services": [
            {"name": "nginx", "status": "running"},
            {"name": "docker", "status": "stopped"}
        ]
    }
]
```

```python
def high_cpu_servers(servers, threshold):
    result = []

    for server in servers:
        if server["metrics"]["cpu"] >= 80:
            result.append(server["name"])
    return result

print(high_cpu_servers(servers, 80))
```

```python
def count_stopped_servers(servers):
    count = 0

    for server in servers:
        for service in server["services"]:
            if service["status"] == "stopped":
                count += 1
    return count

print(count_stopped_servers(servers))
```
