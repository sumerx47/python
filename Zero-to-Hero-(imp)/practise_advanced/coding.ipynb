```python
servers = [
    {"name": "web01", "cpu": 45},
    {"name": "web02", "cpu": 80},
    {"name": "web01", "cpu": 72},
    {"name": "web03", "cpu": 60},
    {"name": "web02", "cpu": 95},
]
```

```python
result = {}

for server in servers:
    name = server["name"]

    if name not in result:
        result[name] = server

    elif server["cpu"] > result[name]["cpu"]:
        result[name] = server

print(list(result.values()))
```
