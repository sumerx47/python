Q24 — API + Deduplication

The API returns duplicate server records.

Remove duplicates by IP.

```python
def remove_dup_servers(servers):

    seen = set()
    result = []

    for server in servers:
        ip = server["ip"]
        if ip not in seen:
            seen.add(ip)
            result.append(server)
    return result
```

Q25 — API + Filter + Highest

From an API response, find the running production server with the highest CPU.

```python
import requests

def high_cpu_server():
    response = requests.get(
        "https://jsonplaceholder.typicode.com/users",
        timeout=10
        )

    response.raise_for_status()
    servers = response.json()
    production = []

    for server in servers:
        if (server["environmenr"] == "production" and server["status"] == "running"):
            production.append(server)
            
        if not production :
            return None

        return max(production,key = lambda x:x["cpu"])
    
```

Q27 — API + JSON + Save Report

🔥 Q29 — Teacher/Interview-Level API Analyzer

Create:

analyze_servers()

It should:

call the API
handle errors
normalize server names
keep production servers
keep running servers
remove duplicate server names
keep highest CPU record
sort by CPU descending
return the result

```python
import requests

def analyse_servers():
    try :
        response = requests.get(
            "https://api.example.com/servers",
            timeout = 10
        )

        response.raise_for_status()

        servers = response.json()

    except requests.exceptions.RequestException:
        return []

    unique = {}

    for server in servers:
        name = server["name"].strip().lower()
        environment = server["environment"].strip().lower()
        status = server["status"].strip().lower()

        if environment != "production":
            continue
        if status != "running":
            continue

        normalized = {
            "name":name,
            "environment":environment,
            "status":status,
            "cpu" : server["cpu"]
        }

        if (name not in status or normalized["cpu"] > unique[name]["cpu"]):
            unique[name] = normalized

    result = sorted(unique.values(), key = lambda server:server["cpu"], reverse = True)
    return result
```
