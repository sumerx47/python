Q7 — API + Frequency

Count servers by status.

```python
import requests

response = requests.get(
    "https://api.example.com/servers"
)

servers = response.json()

frequency = {}

for server in servers:
    status = server["status"]
    frequency[status] = frequency.get(status,0)+1

print(frequency)
```

```python
import requests

response = requests.get(
    "https://api.example.com/servers"
)


servers = response.json()

highest = max(servers, key=lambda x:x["cpu"], reverse = True)

print(highest)
```

Q12 — API Function with Filtering

Create:

get_running_servers()

```python
import requests

def get_running_servers():

    response = requests.get(
        "https://jsonplaceholder.typicode.com/users"
    )
    servers = response.json()

    result = []

    for server in servers:
        if server["status"]  == "running":
            result.append(server)
    return result
```

Q15 — API + Bearer Token

Never hard-code the production token.

Assume the token is stored in an environment variable:

```python
import requests
import os

token = os.getenv("API_TOKEN")

headers = {
    "Authorization":f"Bearer{token}",
    "Accept":"application/json"
}

response = requests.get(
    "https://jsonplaceholder.typicode.com/users",
    headers = headers
)

print(response.json())
```

Q17 — API + Exception Handling

Handle connection and HTTP errors.

```python
import requests

def get_servers():

    try:
        response = requests.get(
            "https://jsonplaceholder.typicode.com/users",
            timeout = 10
        )
        response.raise_for_status()

        return response.json()
    
    except requests.exceptions.Timeout:
        print("Request timed out")

    except requests.exceptions.HTTPError:
        print("HTTP error")

    except requests.exceptions.ConnectionError:
        print("Connection failed")

    except requests.exceptions.RequestException:
        print("Request failed")

    return None
```

Q18 — API + Function + Error Handling

Return an empty list when the API fails.

```python
import requests

def get_servers():
    try:
        response = requests.get(
            "https://jsonplaceholder.typicode.com/users",
            time = 10
        )

        response.raise_for_status()

        return response.json()

    except requests.exceptions.RequestException:
        return []

servers = get_servers()

for server in servers:
    print(server["name"])
```

Q19 — API + JSON Validation

What if the API responds with invalid JSON?

```python
import requests

def get_servers():
    try:
        response = requests.get(
            "https://jsonplaceholder.typicode.com/users",
            timeout = 10
        )
        response.raise_for_status()
        return response.json()
    except requests.exceptions.JSONDecodeError:
        print("API return invalid JSON")
    except requests.exceptions.RequestException:
        print("API request failed")
    return []
```

Q20 — API + Logging

Create a production-style API function with logging.

```python
import requests
import logging

logging.basicConfig(
    filename = "api.log",
    level=logging.INFO, 
    format = "%(asctime)s %(levelname)s %(message)s"
)

def get_servers():
    try:
        response = requests.get(
            "https://jsonplaceholder.typicode.com/users",
            timeout = 10
        )
        response.raise_for_status()
        data = response.json()
        logging.info(
            "API returned %d servers", len(data)
        )
        return data

    except requests.exceptions.RequestException:
        logging.exception("API request failed")
        return []
```

Q21 — POST Request

Send a new server to an API.

1. GET    → read
2. POST   → create
3. PUT    → replace/update
4. PATCH  → partial update
5. DELETE → remove

```python
import requests


server = {
    "name" : "web05",
    "environment":"production",
    "status":"running"
}

response = requests.post(
    "https://jsonplaceholder.typicode.com/users",
    json = server,
    timeout = 10
)

response.raise_for_status()

print(response.json())
```
