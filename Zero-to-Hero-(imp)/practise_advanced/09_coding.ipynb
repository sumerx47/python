Question 1. Convert CPU values from strings to integers

Input:

servers = [
    {"name": "web01", "cpu": "75"},
    {"name": "web02", "cpu": "90"},
    {"name": "web03", "cpu": "45"}
]

```python
servers = [
    {"name": "web01", "cpu": "75"},
    {"name": "web02", "cpu": "90"},
    {"name": "web03", "cpu": "45"}
]

for server in servers:
    server["cpu"] = int(server["cpu"])

print(servers)
```

Question 8. Add a status field based on CPU usage

Rule: CPU usage of 80 or more is "HIGH"; otherwise, "NORMAL".

```python
servers = [
    {"name": "web01", "cpu": 85},
    {"name": "web02", "cpu": 40},
    {"name": "web03", "cpu": 80}
]

for server in servers:
    if server["cpu"] >= 80:
        server["status"] = "HIGH"
    else :
        server["status"] = "NORMAL"
        
print(servers)
```

Question 12. Calculate remaining disk space

```python
servers = [
    {"name": "web01", "disk_total": 500, "disk_used": 320},
    {"name": "web02", "disk_total": 1000, "disk_used": 650}
]

for server in servers:
    server["disk_free"] = server["disk_total"] - server["disk_used"]
print(servers)
```

Question 13. Calculate disk usage percentage



```python
servers = [
    {"name": "web01", "disk_total": 500, "disk_used": 400},
    {"name": "web02", "disk_total": 1000, "disk_used": 250}
]

for server in servers:
    if server["disk_total"] > 0:
        server["disk_percent"] = round(server["disk_used"]/server["disk_total"] * 100,2)

    else:
        server["disk_percent"] = None

print(servers)
```

Question 18. Transform records and keep only production servers

```python
servers = [
    {"name": "web01", "environment": "Production", "cpu": "85"},
    {"name": "web02", "environment": "dev", "cpu": "40"},
    {"name": "db01", "environment": "PRODUCTION", "cpu": "70"}
]

report = []

for server in servers:
    environment = server["environment"].strip().lower()

    if environment == "production":
        report.append({
            "name":server["name"].strip().lower(),
            "cpu":int(server["cpu"]),
            "environment":environment
        })
print(report)
```

Question 20. Safely transform records with missing fields

```python
servers = [
    {"name": "web01", "cpu": "75"},
    {"name": "web02"},
    {"name": "db01", "cpu": "90"}
]

report = []

for server in servers:
    raw_cpu = server.get("cpu")

    if raw_cpu is None:
        continue

    try:
        cpu = int(raw_cpu)
    except (ValueError,TypeError):
        continue

    report.append({
        "name":server.get("name","unknown"),
        "cpu":cpu
    })

print(report)
```

Question 21. Convert Kubernetes pod data into a summary format

```python
pods = [
    {"metadata": {"name": "api-1", "namespace": "production"},
     "status": {"phase": "Running"}},
    {"metadata": {"name": "api-2", "namespace": "dev"},
     "status": {"phase": "Pending"}},
    {"metadata": {"name": "worker-1", "namespace": "production"},
     "status": {"phase": "Running"}}
]

summary = []

for pod in pods:
    summary.append({
        "pod":pod["metadata"]["name"],
        "namespace":pod["metadata"]["namespace"],
        "status":pod["status"]["phase"].lower()
    })

print(summary)
```

Question 24. Merge server metadata and metrics

```python
metadata = [
    {"name": "web01", "environment": "production"},
    {"name": "web02", "environment": "dev"}
]

metrics = [
    {"name": "web01", "cpu": 85},
    {"name": "web02", "cpu": 40}
]


metrics_by_name = {
    item["name"]:item
    for item in metrics
}

merged = []

for server in metadata:
    metric = metrics_by_name.get(server["name"])

    if metric is not None:
        merged.append({
            **server, "cpu":metric["cpu"]
        })

print(merged)

```
