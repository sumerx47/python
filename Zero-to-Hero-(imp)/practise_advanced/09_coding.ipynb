Question 1. Convert CPU values from strings to integers

Input:

servers = [
    {"name": "web01", "cpu": "75"},
    {"name": "web02", "cpu": "90"},
    {"name": "web03", "cpu": "45"}
]

```python
servers = [
    {"name": "web01", "cpu": "75"},
    {"name": "web02", "cpu": "90"},
    {"name": "web03", "cpu": "45"}
]

for server in servers:
    server["cpu"] = int(server["cpu"])

print(servers)
```

Question 2. Convert server names to lowercase

servers = ["WEB01", "Web02", "DB01"]

```python

servers = ["WEB01", "Web02", "DB01"]

clean_names = []
for name in servers:
    clean_names.append(name.lower())
print(clean_names)
```

Question 3. Remove extra spaces from server names

servers = [" web01 ", "  web02", "db01  "]

```python

servers = [" web01 ", "  web02", "db01  "]

clean_namess = []
for server in servers:
    clean_namess.append(server.strip())
print(clean_namess)
```

Question 4. Rename dictionary keys

Convert name to server_name and cpu to cpu_percent.

```python
server = {"name": "web01", "cpu": 75}

transformed = {
    "server_name" : server["name"],
    "cpu_percent" : server["cpu"]        
    }

print(transformed)
```

Question 5. Convert memory from MB to GB

Use 1 GB=1024 MB.

```python
servers = [
    {"name": "web01", "memory_mb": 4096},
    {"name": "web02", "memory_mb": 8192}
]

for server in servers:
    server["memory_gb"] = server["memory_mb"] / 1024

print(servers)
```
