Project 1. Student marks report

Task: Calculate each student's total, average, and result. A student passes if every subject mark is at least 35.

```python
# PROJECT 1 ANSWER
students = [
    {"name": "Alice", "marks": [85, 78, 92]},
    {"name": "Bob", "marks": [40, 32, 65]},
    {"name": "Charlie", "marks": [90, 88, 95]}
]

report = []

for student in students:
    marks = student["marks"]
    total = sum(marks)
    average = round(total/len(marks),2)
    result = "PASS" if all(mark >= 35 for mark in marks) else "FAIL"

    report.append({
        "name":student["name"],
        "total":total,
        "average":average,
        "result":result
    })

print(report)
```

Project 3. Product inventory manager

Task: Calculate remaining stock and identify products that need restocking. Restock when fewer than 10 items remain.



```python
# PROJECT 3 ANSWER
products = [
    {"name": "Keyboard", "stock": 25, "sold": 18},
    {"name": "Mouse", "stock": 15, "sold": 10},
    {"name": "Monitor", "stock": 30, "sold": 12}
]


for product in products:
    remaining = product["stock"] - product["sold"]
    product["remaining"] = remaining
    product["status"] = "RESTOCK" if remaining < 10 else "OK"

print(products)
```

Project 4. Attendance report

Task: Calculate each employee's attendance percentage and flag attendance below 75%.



```python
# PROJECT 4 ANSWER
employees = [
    {"name": "Asha", "present": 24, "working_days": 26},
    {"name": "Rahul", "present": 18, "working_days": 26},
    {"name": "Imran", "present": 25, "working_days": 26}
]

for employee in employees:
    days = employee["working_days"]
    percentage = (round(employee["present"]/days * 100 , 2) if days > 0 else None)

    employee["attendance_percentage"] = percentage
    employee["status"] = ("LOW" if percentage is not None and percentage < 75 else "OK" if percentage is not None else "INVALID")

print(employees)
```

Project 5. Cricket score analyzer

Task: Calculate each player's total runs and strike rate. Strike rate is runs divided by balls faced, multiplied by 100.



```python
# PROJECT 5 ANSWER
players = [
    {"name": "Player A", "runs": 45, "balls": 30},
    {"name": "Player B", "runs": 60, "balls": 40},
    {"name": "Player C", "runs": 20, "balls": 25}
]


for player in players:
    balls = player["balls"]

    player["strike_rate"] = (round(player["runs"]/balls * 100 , 2) if balls > 0 else None)

players.sort(key = lambda player:player["runs"], reverse = True)

print(players)
```

Project 6. Expense tracker

Task: Calculate total expenses and group them by category.



```python
# PROJECT 6 ANSWER
expenses = [
    {"category": "Food", "amount": 250},
    {"category": "Travel", "amount": 100},
    {"category": "Food", "amount": 300},
    {"category": "Bills", "amount": 500},
    {"category": "Travel", "amount": 150}
]

total = {}

for expense in expenses:
    category = expense["category"]
    total[category] = total.get(category,0)+ expense["amount"]

print("total expenses:", sum(total.values()))
print("by category:", total)
```

Project 8. Movie rating analyzer

Task: Calculate the average rating and identify movies rated 8 or higher.



```python
# PROJECT 8 ANSWER
movies = [
    {"title": "Movie A", "rating": 8.5},
    {"title": "Movie B", "rating": 7.0},
    {"title": "Movie C", "rating": 9.0},
    {"title": "Movie D", "rating": 6.5}
]

average = sum(movie["rating"] for movie in movies) /len(movies)

top_movies = [movie["title"] for movie in movies if movie["rating"] >= 8]

print("Average rating:", round(average,2))
print("Top movies:", top_movies)
```

Project 9. Order processing report

Task: Calculate order values, include only completed orders, and sort by value.




```python
# PROJECT 9 ANSWER
orders = [
    {"id": 101, "status": "completed", "price": 500, "quantity": 2},
    {"id": 102, "status": "pending", "price": 300, "quantity": 1},
    {"id": 103, "status": "completed", "price": 200, "quantity": 5}
]

report = []

for order in orders:
    if order["status"].lower() == "completed":
        report.append({
            "id":order["id"],
            "value": order["price"] * order["quantity"]
        })

report.sort(key = lambda x:x["value"], reverse=True)
print(report)
```
