Project 1. Student marks report

Task: Calculate each student's total, average, and result. A student passes if every subject mark is at least 35.

```python
# PROJECT 1 ANSWER
students = [
    {"name": "Alice", "marks": [85, 78, 92]},
    {"name": "Bob", "marks": [40, 32, 65]},
    {"name": "Charlie", "marks": [90, 88, 95]}
]

report = []

for student in students:
    marks = student["marks"]
    total = sum(marks)
    average = round(total/len(marks),2)
    result = "PASS" if all(mark >= 35 for mark in marks) else "FAIL"

    report.append({
        "name":student["name"],
        "total":total,
        "average":average,
        "result":result
    })

print(report)
```

Project 5. Cricket score analyzer

Task: Calculate each player's total runs and strike rate. Strike rate is runs divided by balls faced, multiplied by 100.



```python
# PROJECT 5 ANSWER
players = [
    {"name": "Player A", "runs": 45, "balls": 30},
    {"name": "Player B", "runs": 60, "balls": 40},
    {"name": "Player C", "runs": 20, "balls": 25}
]


for player in players:
    balls = player["balls"]

    player["strike_rate"] = (round(player["runs"]/balls * 100 , 2) if balls > 0 else None)

players.sort(key = lambda player:player["runs"], reverse = True)

print(players)
```

Project 14. Log severity summary

Task: Count log levels and extract error messages.



```python
# PROJECT 14 ANSWER
logs = [
    "INFO Server started",
    "ERROR Database unavailable",
    "WARNING High CPU",
    "ERROR Connection failed",
    "INFO Health check passed"
]

level_counts = {}
errors = []

for line in logs :
    parts = line.split(maxsplit=1)

    if len(parts) != 2:
        continue

    level,message = parts
    level_counts[level] = level_counts.get(level,0)+1

    if level == "ERROR":
        errors.append(message)

print("counts:", level_counts)
print("ERRORS:", errors)
```

Project 18. Deployment configuration comparison

Task: Identify configuration settings that changed between two deployments.


```python
# PROJECT 18 ANSWER
old_config = {
    "port": 8000,
    "workers": 2,
    "debug": False
}

new_config = {
    "port": 8000,
    "workers": 4,
    "debug": True
}

changes = {}

for key,new_value in new_config.items():
    old_value = old_config.get(key)

    if old_value != new_value:
        changes[key] = {
            "old":old_value,
            "new":new_value
        }

print(changes)
```

Project 20. Directory file inventory

Task: Recursively list .log files in a directory and report their sizes.

```python
from pathlib import Path
root = Path("logs")

report = []

if root.exists() and root.is_dir():
    for file_path in root.rglob("*.log"):
        if file_path.is_file():
            report.append({
                "file":str(file_path),
                "size_bytes":file_path.stat().st_size
            })

print(report)
```

Project 21. Server resource alert generator

Task: Generate alerts for high CPU or memory usage and sort the most critical servers first.



```python
# PROJECT 21 ANSWER
servers = [
    {"name": "web01", "cpu": 85, "memory": 60},
    {"name": "web02", "cpu": 95, "memory": 90},
    {"name": "db01", "cpu": 40, "memory": 88},
    {"name": "cache01", "cpu": 35, "memory": 45}
]

alerts = []

for server in servers:
    if server["cpu"] >= 80 or server["memory"] >= 80:
        alerts.append({
            "name":server["name"],
            "cpu":server["cpu"],
            "memory":server["memory"],
            "alerts":[metric for metric,value in [("CPU",server["cpu"]),("Memory",server["memory"])] if value >= 80]
        })

alerts.sort(key = lambda server:server["cpu"]+server["memory"], reverse=True)
print(alerts)
```
