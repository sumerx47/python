# 12. FREQUENCY ANALYSIS

🟢 LEVEL 1 —  Basic Frequency

```python
numbers = [10, 20, 10, 30, 20, 10, 40, 30]

frequency = {}

for num in numbers:
    if num not in frequency:
        frequency[num] = 1
    else :
        frequency[num] += 1
print(frequency)
```

```python
words = ["apple", "banana", "apple", "orange", "banana", "apple"]

frequency = {}

for word in words:
    if word not in frequency:
        frequency[word] = 1
    else :
        frequency[word] += 1
print(frequency)
```

```python
words = ["apple", "banana", "apple", "orange", "banana", "apple"]

frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) +1
print(frequency)
```

🟢 LEVEL 2 — Frequency Analysis

```python
numbers = [5, 2, 5, 3, 2, 5, 4, 3]

frequency = {}

for num in numbers:
    frequency[num] = frequency.get(num, 0) + 1
print(frequency)

most_frequent_item = max(frequency, key = frequency.get)
print(most_frequent_item, "->" ,frequency[most_frequent_item])
```

```python
servers = [
    "web01",
    "web02",
    "web01",
    "web03",
    "web02",
    "web04"
]

frequency = {}
for server in servers:
    frequency[server] = frequency.get(server, 0) + 1

dup = []
uniq = []

for server,count in frequency.items():
    if count > 1:
        dup.append(server)
    elif count == 1:
        uniq.append(server)

print("duplicates: ",dup)
print("unique: ",uniq)
```
