# 12. FREQUENCY ANALYSIS

```python
numbers = [10, 20, 10, 30, 20, 10, 40, 30]

frequency = {}

for num in numbers:
    if num not in frequency:
        frequency[num] = 1
    else :
        frequency[num] += 1
print(frequency)
```

```python
words = ["apple", "banana", "apple", "orange", "banana", "apple"]

frequency = {}

for word in words:
    if word not in frequency:
        frequency[word] = 1
    else :
        frequency[word] += 1
print(frequency)
```
