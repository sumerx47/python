🟢 LEVEL 1 —  Basic Frequency

```python
numbers = [10, 20, 10, 30, 20, 10, 40, 30]

frequency = {}

for num in numbers:
    if num not in frequency:
        frequency[num] = 1
    else :
        frequency[num] += 1
print(frequency)
```

```python
words = ["apple", "banana", "apple", "orange", "banana", "apple"]

frequency = {}

for word in words:
    if word not in frequency:
        frequency[word] = 1
    else :
        frequency[word] += 1
print(frequency)
```

🟢 LEVEL 2 — Frequency Analysis

```python
numbers = [5, 2, 5, 3, 2, 5, 4, 3]

frequency = {}

for num in numbers:
    frequency[num] = frequency.get(num, 0) + 1
print(frequency)

most_frequent_item = max(frequency, key = frequency.get)
print(most_frequent_item, "->" ,frequency[most_frequent_item])
```

```python
servers = [
    "web01",
    "web02",
    "web01",
    "web03",
    "web02",
    "web04"
]

frequency = {}
for server in servers:
    frequency[server] = frequency.get(server, 0) + 1

dup = []
uniq = []

for server,count in frequency.items():
    if count > 1:
        dup.append(server)
    elif count == 1:
        uniq.append(server)

print("duplicates: ",dup)
print("unique: ",uniq)
```

🟡 LEVEL 3 — Frequency + Sorting

```python
words = [
    "docker",
    "python",
    "docker",
    "linux",
    "python",
    "docker",
    "git",
    "linux"
]
```

```python
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

result = sorted(frequency.items(), key = lambda x:x[1], reverse=True)
print(result)
```

```python
servers = [
    "web01", "web02", "web01",
    "web03", "web02", "web01",
    "web04", "web03", "web02"
]

frequency = {}

for server in servers:
    frequency[server] = frequency.get(server, 0) + 1

result = sorted(frequency.items(), key=lambda x:x[1], reverse=True)[:3]
print(result)
```

🟡 LEVEL 5 — Frequency + Multiple Conditions


```python
servers = [
    {"name": "web01", "cpu": 70},
    {"name": "web02", "cpu": 90},
    {"name": "web03", "cpu": 85},
    {"name": "web04", "cpu": 60},
    {"name": "web05", "cpu": 95}
]

frequency = {}

for server in servers:
    cpu = server["cpu"]
    if cpu >= 80:
        #frequency.append(server)
        frequency[cpu] = frequency.get(cpu, 0)+1

print(frequency)
print(len(frequency))
```

```python
servers = [
    {"name": "web01", "env": "prod", "cpu": 90},
    {"name": "web02", "env": "prod", "cpu": 70},
    {"name": "web03", "env": "prod", "cpu": 85},
    {"name": "test01", "env": "test", "cpu": 95},
    {"name": "dev01", "env": "dev", "cpu": 60},
    {"name": "dev02", "env": "dev", "cpu": 90},
]

overloaded = {}
for server in servers:
    env = server["env"]
    cpu = server["cpu"]
    if cpu >= 80:
        overloaded[env] = overloaded.get(env,0)+1
print(overloaded)
```

🟡 LEVEL 6 — Frequency + JSON

```python
data = {
    "servers": [
        {"id": 101, "status": "running"},
        {"id": 102, "status": "failed"},
        {"id": 103, "status": "running"},
        {"id": 104, "status": "stopped"},
        {"id": 105, "status": "running"},
        {"id": 106, "status": "failed"}
    ]
}

frequency = {}
for server in data["servers"]:
    status = server["status"]
    frequency[status] = frequency.get(status,0)+1
print(frequency)

result = {}
for status, count in frequency.items():
    if count >= 2:
        result[status] = count
print(result)
```

### 🟠 LEVEL 8 — Functions
#### reusable frequency function

```python

def frequency_analysis(data):
    frequency = {}

    for item in data:
        frequency[item] = frequency.get(item,0)+1

    return frequency

servers = [
    "web01",
    "web02",
    "web01",
    "web03",
    "web01"
]

result = frequency_analysis(servers)
print(result)
```

```python
def most_frequent(data):
    frequency = {}

    for item in data:
        frequency[item] = frequency.get(item,0)+1

    return max(frequency, key = frequency.get)

servers = ["web01", "web02", "web01", "web01", "web03"]


print(most_frequent(servers))
```

🟠 LEVEL 9 — Regex + Frequency

```python
import re

logs = """
2026-09-26 INFO Server started
2026-09-26 ERROR Database failed
2026-09-26 INFO Request received
2026-09-26 WARNING CPU high
2026-09-26 ERROR Connection failed
"""
levels = re.findall(r"\b(INFO|ERROR|WARNING)\b",logs)

frequency = {}

for level in levels:
    frequency[level] = frequency.get(level,0)+1
print(frequency)
```

🔴 LEVEL 10 — Regex + File + Frequency


```python
import re

logs = """
2026-09-26 10:00:01 ERROR web01 Database failed
2026-09-26 10:01:02 INFO web02 Request received
2026-09-26 10:02:03 ERROR web01 Connection failed
2026-09-26 10:03:04 WARNING web03 CPU high
2026-09-26 10:04:05 ERROR web02 Timeout
2026-09-26 10:05:06 ERROR web01 Disk failed
"""

errors = re.findall(r"\bERROR\s+(\w+)",logs)

frequency = {}

for server in errors:
    frequency[server] = frequency.get(server,0)+1

total_errors = len(errors)

most_error_server = max(frequency , key=frequency.get)

print("total_errors:",total_errors)
print("frequency:",frequency)
print("most_error_server:",most_error_server)
```
