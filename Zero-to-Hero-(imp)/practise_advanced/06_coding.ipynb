Q1 — Python Dictionary → JSON String

Convert this Python dictionary into a JSON string.

server = {
    "name": "web01",
    "status": "running",
    "cpu": 72
}

```python
server = {
    "name": "web01",
    "status": "running",
    "cpu": 72
}

import json

json_data = json.dumps(server)

print(json_data)
```

Q2 — JSON String → Python Dictionary

Convert:

'{"name": "web01", "status": "running", "cpu": 72}'

into a Python dictionary.

```python
import json

data = '{"name": "web01", "status": "running", "cpu": 72}'

server = json.loads(data)

print(server)
```

Q3 — Access Nested JSON Data
data = '''
{
    "server": {
        "name": "web01",
        "status": "running",
        "resources": {
            "cpu": 75,
            "memory": 64
        }
    }
}
'''

Print the server name and CPU.

```python
data = '''
{
    "server": {
        "name": "web01",
        "status": "running",
        "resources": {
            "cpu": 75,
            "memory": 64
        }
    }
}
'''

import json

server = json.loads(data)

print(server["server"]["name"])
print(server["server"]["resources"]["cpu"])
```

Q4 — JSON List

Given:

[
    {"name": "web01", "status": "running"},
    {"name": "web02", "status": "stopped"},
    {"name": "web03", "status": "running"}
]

Print every server name.

```python
import json

data = '''[
    {"name": "web01", "status": "running"},
    {"name": "web02", "status": "stopped"},
    {"name": "web03", "status": "running"}
]'''

servers = json.loads(data)

for server in servers:
    print(server["name"])
```

Q5 — JSON + Filtering

Print only running servers.

Answer
QUESTION 5 ANSWER

```python
import json

data = '''
[
    {"name": "web01", "status": "running"},
    {"name": "web02", "status": "stopped"},
    {"name": "web03", "status": "running"}
]
'''

servers = json.loads(data)

for server in servers:
    status = server["status"]
    name = server["name"]
    if status == "running":
        print(name)
```
