Count how many servers are running.

Answer

Q7 — JSON + Frequency Analysis

Count servers by status.

Answer
QUESTION 7 



```python
import json

data = '''
[
    {"name": "web01", "status": "running"},
    {"name": "web02", "status": "stopped"},
    {"name": "web03", "status": "running"},
    {"name": "web04", "status": "running"},
    {"name": "web05", "status": "stopped"}
]
'''

servers = json.loads(data)

frequency = {}

for server in servers:
    status  = server["status"]
    frequency[status] = frequency.get(status,0)+1

print(frequency)
```

Q8 — JSON + Highest

Find the server with the highest CPU usage.

servers = [
    {"name": "web01", "cpu": 45},

    {"name": "web03", "cpu": 72}
]

```python
import json

data = '''[
    {"name": "web01", "cpu": 45},
    {"name": "web02", "cpu": 91},
    {"name": "web03", "cpu": 72}
]'''

servers = json.loads(data)

highest = max(servers, key = lambda x:x["cpu"])
print(highest) 
```

Q9 — JSON + Sorting

Sort servers by CPU from highest to lowest.

Answer



```python
import json

data = '''
[
    {"name": "web01", "cpu": 45},
    {"name": "web02", "cpu": 91},
    {"name": "web03", "cpu": 72}
]
'''

servers = json.loads(data)

servers = sorted(servers, key =lambda x:x["cpu"] , reverse = True)

for server in servers:
    name = server["name"]
    cpu = server["cpu"]
    print(name,cpu)
```

Q12 — JSON File + Function

Create:

get_running_servers("servers.json")

```python
import json
from pathlib import Path

def get_running_servers(filename):

    text = Path(filename).read_text()

    servers = json.loads(text)

    result = []

    for server in servers:
        if server["status"] == "running":
            result.append(server["name"])
    return result

print(get_running_servers("servers.json"))
            
```

Q13 — JSON File + Error Handling

Handle a missing JSON file.

```python
import json
from pathlib import Path


def load_json(filename):
    try :
        text = Path(filename).read_text()

        return json.loads(text)

    except FileNotFoundError:
        return "file does not exist"
    except json.JSONDecodeError:
        return "invalid json"
    except PermissionError:
        return "permission denied"

    return None

print(load_json("servers.json"))
```

Q15 — JSON → Modify → Save

Read JSON, change status to "maintenance" for web01, then save it.

```python
import json
from pathlib import Path

filename = "servers.json"

data = json.loads(Path(filename).read_text())

for server in data:
    if server["name"] == "web01":
        server["status"] = "maintenance"

Path(filename).write_text(json.dumps(data,indent = 4))
```

🔴 Level 4 — Advanced JSON
Q16 — JSON + Deduplication

Input:

[
    {"name": "web01", "ip": "10.0.0.1"},
    {"name": "web02", "ip": "10.0.0.2"},
    {"name": "web01", "ip": "10.0.0.1"}
]

Remove duplicate servers based on IP.

```python
servers = [
    {"name": "web01", "ip": "10.0.0.1"},
    {"name": "web02", "ip": "10.0.0.2"},
    {"name": "web01", "ip": "10.0.0.1"}
]

seen = set()
result = []

for server in servers:
    ip = server["ip"]

    if ip not in seen:
        seen.add(ip)
        result.append(server)
print(result)
```
