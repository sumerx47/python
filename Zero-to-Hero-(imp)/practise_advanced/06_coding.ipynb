Q12 — JSON File + Function

Create:

```python
import json
from pathlib import Path

def get_running_servers(filename):

    text = Path(filename).read_text()

    servers = json.loads(text)

    result = []

    for server in servers:
        if server["status"] == "running":
            result.append(server["name"])
    return result

print(get_running_servers("servers.json"))
            
```

🚀 Q28 — JSON + Logging

Create a JSON processor that logs successful and failed processing.

```python
import json
import logging
from pathlib import Path

logging.basicConfig(
    filename = "json_processor.log",
    level = logging.INFO,
    format = "%(asctime)s %(levelname)s %(message)s"
)

def process_json(filename):
    try :
        data = json.loads(Path(filename).read_text())
        logging.info("successfully loaded %s",filename)
        return data
    except FileNotFoundError:
        logging.error("File not found: %s", filename)

    except json.JSONDecodeError:
        logging.error("invalid json: %s",filename)

    except Exception:
        logging.exception("unexpected error processing %s",filename)

    return None

```

🔥 Q29 — Teacher-Level JSON Analyzer

Given:

Create an analyzer that:

1. normalizes server names
2. keeps only production servers
3. keeps only running servers
4. removes duplicate names
5. keeps the highest CPU record
6. sorts by CPU descending

```python
servers = [
    {"name": "web01", "env": "production", "status": "running", "cpu": 85},
    {"name": "web02", "env": "production", "status": "stopped", "cpu": 90},
    {"name": "web03", "env": "development", "status": "running", "cpu": 95},
    {"name": "web04", "env": "production", "status": "running", "cpu": 72},
    {"name": "web01", "env": "production", "status": "running", "cpu": 90}
]

def analyze_servers(servers):
  unique = {}

  for server in servers:
    name = server["name"].strip().lower()
    if server["env"].strip().lower != "production":
        continue
    if server["status"].strip().lower() != "running":
        continue

    normalized = {
        "name":name,
        "env":"production",
        "status":"running",
        "cpu":server["cpu"]
    }

    if (name not in unique or normalized["cpu"] > unique[name]["cpu"]):
        unique[name] = normalized

    result = sorted(
        unique.values(),
        key = lambda x:x["cpu"],
        reverse = True
    )

    return result

```

🏆 Q30 — Final Teacher/Interview-Level JSON Problem

This is the final cumulative question for Pattern 16.

You receive:

analyze_logs(logs)

It must return:

total logs
level frequency
total errors
errors by server
error-code frequency
unique error IPs
unique error messages
most problematic server
servers sorted by error count

```python
# QUESTION 30 ANSWER

def analyze_logs(logs):

    total_logs = len(logs)

    level_frequency = {}
    errors_by_server = {}
    error_code_frequency = {}

    unique_error_ips = set()
    unique_error_messages = set()

    for log in logs:

        level = log["level"].strip().upper()
        server = log["server"].strip().lower()

        # Frequency of every log level
        level_frequency[level] = (
            level_frequency.get(level, 0) + 1
        )

        # Only analyze errors
        if level == "ERROR":

            errors_by_server[server] = (
                errors_by_server.get(server, 0) + 1
            )

            code = log["code"]

            error_code_frequency[code] = (
                error_code_frequency.get(code, 0) + 1
            )

            unique_error_ips.add(
                log["ip"]
            )

            unique_error_messages.add(
                log["message"]
            )

    total_errors = sum(
        errors_by_server.values()
    )

    most_problematic_server = max(
        errors_by_server,
        key=errors_by_server.get
    )

    servers_sorted = sorted(
        errors_by_server.items(),
        key=lambda item: item[1],
        reverse=True
    )

    return {
        "total_logs": total_logs,
        "level_frequency": level_frequency,
        "total_errors": total_errors,
        "errors_by_server": errors_by_server,
        "error_code_frequency": error_code_frequency,
        "unique_error_ips": sorted(unique_error_ips),
        "unique_error_messages": sorted(unique_error_messages),
        "most_problematic_server": most_problematic_server,
        "servers_sorted": servers_sorted
    }

print(analyze_logs("network.log"))
```
