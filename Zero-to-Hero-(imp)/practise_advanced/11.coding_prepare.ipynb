Question 1 — Advanced Employee Performance Analyzer
Difficulty: Advanced
You are given employee records from a company. Build a program that cleans the data, removes duplicates, calculates performance, groups employees by department, and generates a final report.
Input data
employees = [
    {"id": 101, "name": "  ALICE ", "dept": "IT", "salary": 60000,
     "scores": [85, 90, 95], "skills": ["Python", "SQL"]},

    {"id": 102, "name": "bob", "dept": "HR", "salary": 45000,
     "scores": [70, 75, 80], "skills": ["Excel", "Communication"]},

    {"id": 103, "name": "Charlie", "dept": "IT", "salary": 75000,
     "scores": [95, 92, 98], "skills": ["Python", "Docker"]},

    {"id": 104, "name": "alice", "dept": "IT", "salary": 65000,
     "scores": [90, 90, 90], "skills": ["Python", "Linux"]},

    {"id": 105, "name": "Eva", "dept": "Finance", "salary": 50000,
     "scores": [60, 65, 70], "skills": ["Excel", "SQL"]},

    {"id": 102, "name": "BOB", "dept": "HR", "salary": 45000,
     "scores": [70, 75, 80], "skills": ["Excel", "Communication"]},

    {"id": 106, "name": "Frank", "dept": "IT", "salary": 80000,
     "scores": [88, 91, 93], "skills": ["AWS", "Docker"]}
]



Your tasks
1. Clean employee names: remove extra spaces and convert to title case.
2. Remove duplicate employee records using id.
3. Calculate each employee's average score.
4. Assign a performance rating:
   - Average ≥ 90: "Excellent"
   - Average ≥ 80: "Good"
   - Average ≥ 70: "Average"
   - Otherwise: "Needs Improvement"
5. Find the highest-paid employee and the highest-performing employee.
6. Group employees by department.
7. Calculate the average salary for each department.
8. Find employees who know both Python and SQL.
9. Find all unique skills across the company.
10. Sort employees by average score in descending order.
11. Generate a report containing the total number of employees, department summaries, top performers, and skill statistics.
12. Save the report as a JSON file and handle possible file errors.
Concepts tested: data cleaning, strings, lists, dictionaries, sets, loops, functions, filtering, counting, grouping, deduplication, aggregation, sorting, exception handling, JSON, and file handling.

```python
import json
from pathlib import Path


employees = [
    {"id": 101, "name": "  ALICE ", "dept": "IT", "salary": 60000,
     "scores": [85, 90, 95], "skills": ["Python", "SQL"]},

    {"id": 102, "name": "bob", "dept": "HR", "salary": 45000,
     "scores": [70, 75, 80], "skills": ["Excel", "Communication"]},

    {"id": 103, "name": "Charlie", "dept": "IT", "salary": 75000,
     "scores": [95, 92, 98], "skills": ["Python", "Docker"]},

    {"id": 104, "name": "alice", "dept": "IT", "salary": 65000,
     "scores": [90, 90, 90], "skills": ["Python", "Linux"]},

    {"id": 105, "name": "Eva", "dept": "Finance", "salary": 50000,
     "scores": [60, 65, 70], "skills": ["Excel", "SQL"]},

    {"id": 102, "name": "BOB", "dept": "HR", "salary": 45000,
     "scores": [70, 75, 80], "skills": ["Excel", "Communication"]},

    {"id": 106, "name": "Frank", "dept": "IT", "salary": 80000,
     "scores": [88, 91, 93], "skills": ["AWS", "Docker"]}
]


def analyze_employees(records):
    cleaned = []
    seen_ids = set()

    # 1. Clean names and remove duplicate IDs.
    for employee in records:
        emp_id = employee["id"]

        if emp_id in seen_ids:
            continue

        seen_ids.add(emp_id)

        emp = employee.copy()
        emp["name"] = " ".join(emp["name"].split()).title()
        emp["dept"] = emp["dept"].strip().title()

        # 2. Calculate average score.
        scores = emp["scores"]
        average = sum(scores) / len(scores) if scores else 0
        emp["average_score"] = round(average, 2)

        # 3. Assign performance rating.
        if average >= 90:
            rating = "Excellent"
        elif average >= 80:
            rating = "Good"
        elif average >= 70:
            rating = "Average"
        else:
            rating = "Needs Improvement"

        emp["rating"] = rating
        emp["skills"] = sorted({
            skill.strip().title() for skill in emp["skills"]
        })

        cleaned.append(emp)

    # 4. Highest-paid and highest-performing employees.
    highest_paid = max(cleaned, key=lambda e: e["salary"])
    highest_performer = max(cleaned, key=lambda e: e["average_score"])

    # 5. Group employees by department.
    departments = {}

    for emp in cleaned:
        dept = emp["dept"]
        departments.setdefault(dept, []).append(emp)

    # 6. Calculate department average salaries.
    department_summary = {}

    for dept, members in departments.items():
        department_summary[dept] = {
            "employee_count": len(members),
            "average_salary": round(sum(e["salary"] for e in members) / len(members), 2)
        }

    # 7. Find employees who know both Python and SQL.
    python_and_sql = [
        emp["name"] for emp in cleaned
        if {"Python", "Sql"}.issubset(set(emp["skills"]))
    ]

    # 8. Collect unique skills.
    all_skills = sorted({
        skill
        for emp in cleaned
        for skill in emp["skills"]
    })

    # 9. Sort employees by performance.
    ranked = sorted(
        cleaned,
        key=lambda e: e["average_score"],
        reverse=True
    )

    # 10. Create the report.
    return {
        "total_employees": len(cleaned),
        "employees": cleaned,
        "highest_paid": highest_paid["name"],
        "highest_performer": highest_performer["name"],
        "department_summary": department_summary,
        "knows_python_and_sql": python_and_sql,
        "unique_skills": all_skills,
        "ranked_employees": [
            {
                "name": e["name"],
                "average_score": e["average_score"],
                "rating": e["rating"]
            }
            for e in ranked
        ]
    }


report = analyze_employees(employees)

try:
    output_path = Path("employee_report.json")
    output_path.write_text(
        json.dumps(report, indent=4),
        encoding="utf-8"
    )
    print("Report saved successfully.")
    print(json.dumps(report, indent=4))
except OSError as error:
    print(f"File error: {error}")
```

Question 2 — DevOps Server Monitoring and Log Analyzer
Difficulty: Advanced+
You are given server monitoring data and application logs. Your task is to create a monitoring report that identifies unhealthy servers and frequent errors.
Input data
servers = [
    {"host": "web-01", "env": "PROD", "cpu": 92, "memory": 88, "disk": 95},
    {"host": "web-02", "env": "PROD", "cpu": 45, "memory": 60, "disk": 70},
    {"host": "db-01", "env": "PROD", "cpu": 85, "memory": 96, "disk": 91},
    {"host": "test-01", "env": "TEST", "cpu": 97, "memory": 40, "disk": 50},
    {"host": "web-03", "env": "PROD", "cpu": 75, "memory": 82, "disk": 65},
    {"host": "web-02", "env": "PROD", "cpu": 45, "memory": 60, "disk": 70}
]

logs = [
    "ERROR: web-01 connection timeout",
    "INFO: web-02 health check passed",
    "ERROR: db-01 database unavailable",
    "WARNING: web-01 high CPU usage",
    "ERROR: web-01 connection timeout",
    "INFO: web-03 service started",
    "WARNING: db-01 high memory usage",
    "ERROR: db-01 database unavailable"
]



Your tasks
1. Remove duplicate server records using the hostname.
2. Normalize environment names to lowercase.
3. Classify each server:
   - CPU ≥ 90 or memory ≥ 90 or disk ≥ 90: "CRITICAL"
   - Otherwise, CPU ≥ 75 or memory ≥ 80 or disk ≥ 80: "WARNING"
   - Otherwise: "HEALTHY"
4. Find all critical production servers.
5. Calculate average CPU, memory, and disk usage for production servers.
6. Group servers by environment and health status.
7. Count each log severity: ERROR, WARNING, and INFO.
8. Extract hostnames from the logs and count errors per host.
9. Find the host with the most errors.
10. Identify servers that have both a critical health status and at least one error in the logs.
11. Generate an alert report sorted by severity and resource usage.
12. Save the report to JSON and write important alerts to a separate log file.
13. Use functions, validation, exception handling, and Python's logging module.
Concepts tested: almost every major Python data pattern, including nested data, sets, frequency dictionaries, string parsing, conditions, aggregation, sorting, file processing, JSON, logging, exception handling, and data pipelines.

Important detail: A server can have multiple resource problems at once, but it should receive only one final health classification.

```python
import json
import logging
from pathlib import Path


servers = [
    {"host": "web-01", "env": "PROD", "cpu": 92, "memory": 88, "disk": 95},
    {"host": "web-02", "env": "PROD", "cpu": 45, "memory": 60, "disk": 70},
    {"host": "db-01", "env": "PROD", "cpu": 85, "memory": 96, "disk": 91},
    {"host": "test-01", "env": "TEST", "cpu": 97, "memory": 40, "disk": 50},
    {"host": "web-03", "env": "PROD", "cpu": 75, "memory": 82, "disk": 65},
    {"host": "web-02", "env": "PROD", "cpu": 45, "memory": 60, "disk": 70}
]

logs = [
    "ERROR: web-01 connection timeout",
    "INFO: web-02 health check passed",
    "ERROR: db-01 database unavailable",
    "WARNING: web-01 high CPU usage",
    "ERROR: web-01 connection timeout",
    "INFO: web-03 service started",
    "WARNING: db-01 high memory usage",
    "ERROR: db-01 database unavailable"
]


# Configure logging.
logging.basicConfig(
    filename="monitoring.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)


def analyze_servers(records):
    cleaned = []
    seen_hosts = set()

    # 1. Normalize and deduplicate servers.
    for server in records:
        host = server["host"].strip().lower()

        if host in seen_hosts:
            continue

        seen_hosts.add(host)

        item = server.copy()
        item["host"] = host
        item["env"] = server["env"].strip().lower()

        # 2. Classify health.
        if (
            item["cpu"] >= 90
            or item["memory"] >= 90
            or item["disk"] >= 90
        ):
            item["health"] = "CRITICAL"

        elif (
            item["cpu"] >= 75
            or item["memory"] >= 80
            or item["disk"] >= 80
        ):
            item["health"] = "WARNING"

        else:
            item["health"] = "HEALTHY"

        cleaned.append(item)

    # 3. Calculate production averages.
    production = [
        server for server in cleaned
        if server["env"] == "prod"
    ]

    if production:
        production_averages = {
            resource: round(
                sum(server[resource] for server in production)
                / len(production),
                2
            )
            for resource in ("cpu", "memory", "disk")
        }
    else:
        production_averages = {}

    # 4. Group by environment and health.
    grouped = {}

    for server in cleaned:
        env = server["env"]
        health = server["health"]

        grouped.setdefault(env, {})
        grouped[env].setdefault(health, [])
        grouped[env][health].append(server["host"])

    # 5. Identify critical production servers.
    critical_production = [
        server["host"] for server in cleaned
        if server["env"] == "prod"
        and server["health"] == "CRITICAL"
    ]

    return cleaned, production_averages, grouped, critical_production


def analyze_logs(log_lines, known_hosts):
    severity_counts = {
        "ERROR": 0,
        "WARNING": 0,
        "INFO": 0
    }

    error_counts = {}
    error_messages = {}

    for line in log_lines:
        parts = line.split(":", 1)

        if len(parts) != 2:
            logging.warning("Malformed log line: %s", line)
            continue

        severity = parts[0].strip().upper()
        message = parts[1].strip()

        if severity not in severity_counts:
            continue

        severity_counts[severity] += 1

        # Match a hostname at the beginning of the message.
        words = message.split()
        host = words[0].lower() if words else ""

        if host not in known_hosts:
            continue

        if severity == "ERROR":
            error_counts[host] = error_counts.get(host, 0) + 1

            error_messages.setdefault(host, {})
            error_messages[host][message] = (
                error_messages[host].get(message, 0) + 1
            )

    return severity_counts, error_counts, error_messages


def main():
    cleaned, averages, grouped, critical = analyze_servers(servers)

    known_hosts = {server["host"] for server in cleaned}

    severity_counts, error_counts, error_messages = analyze_logs(
        logs, known_hosts
    )

    # 6. Find critical servers that also have errors.
    critical_with_errors = [
        server["host"] for server in cleaned
        if server["health"] == "CRITICAL"
        and error_counts.get(server["host"], 0) > 0
    ]

    # 7. Sort alerts: critical first, then highest resource usage.
    priority = {"CRITICAL": 0, "WARNING": 1, "HEALTHY": 2}

    alerts = sorted(
        cleaned,
        key=lambda server: (
            priority[server["health"]],
            -max(server["cpu"], server["memory"], server["disk"])
        )
    )

    for server in alerts:
        if server["health"] == "CRITICAL":
            logging.error(
                "%s is CRITICAL: CPU=%s, memory=%s, disk=%s",
                server["host"],
                server["cpu"],
                server["memory"],
                server["disk"]
            )

    report = {
        "unique_server_count": len(cleaned),
        "servers": cleaned,
        "production_averages": averages,
        "grouped_servers": grouped,
        "critical_production_servers": critical,
        "log_severity_counts": severity_counts,
        "errors_per_host": error_counts,
        "repeated_error_messages": error_messages,
        "critical_servers_with_errors": critical_with_errors,
        "alerts": alerts
    }

    try:
        Path("server_monitoring_report.json").write_text(
            json.dumps(report, indent=4),
            encoding="utf-8"
        )

        critical_alerts = [
            server for server in alerts
            if server["health"] == "CRITICAL"
        ]

        Path("critical_alerts.json").write_text(
            json.dumps(critical_alerts, indent=4),
            encoding="utf-8"
        )

        print(json.dumps(report, indent=4))
        print("\nMonitoring reports saved successfully.")

    except OSError as error:
        logging.exception("Failed to save monitoring report")
        print(f"File error: {error}")


if __name__ == "__main__":
    main()
```

Question 3 — API Response Validator and Data Transformation Pipeline
Difficulty: Advanced
A monitoring API returns inconsistent records. Build a reusable pipeline that validates records, removes duplicates, transforms valid data, and produces statistics.
Input data
api_response = [
    {"id": 1, "host": " WEB-01 ", "status": "running",
     "cpu": "85", "region": "ap-south-1"},

    {"id": 2, "host": "web-02", "status": "stopped",
     "cpu": "0", "region": "ap-south-1"},

    {"id": 3, "host": "DB-01", "status": "running",
     "cpu": "95", "region": "ap-south-2"},

    {"id": 1, "host": "web-01", "status": "running",
     "cpu": "85", "region": "ap-south-1"},

    {"id": 4, "host": "web-04", "status": "running",
     "cpu": "invalid", "region": "ap-south-1"},

    {"id": 5, "host": "web-05", "status": "running",
     "cpu": "72", "region": "ap-south-2"},

    {"id": 6, "host": "web-06", "status": "unknown",
     "cpu": "50", "region": "ap-south-2"},

    {"id": 7, "host": "web-07", "status": "running",
     "cpu": "105", "region": "ap-south-1"},

    {"id": 8, "host": "web-08", "status": "running",
     "region": "ap-south-1"}
]



Your tasks
1. Validate required fields: id, host, status, cpu, and region.
2. Normalize hostnames to lowercase and remove extra spaces.
3. Convert CPU values to integers, handling invalid or missing values.
4. Reject records with missing required fields, invalid CPU values, or CPU outside the range 0–100. Record the rejection reason for each.
5. Remove duplicate IDs, keeping the first valid occurrence.
6. Normalize status values to lowercase and allow only "running" and "stopped".
7. Categorize valid records:
   - CPU ≥ 90: "HIGH"
   - CPU ≥ 70: "MEDIUM"
   - Otherwise: "LOW"
8. Group valid records by region.
9. Calculate the average CPU usage per region.
10. Find the highest-CPU server in each region.
11. Produce counts for valid, rejected, and duplicate records.
12. Export valid transformed data and rejected records to separate JSON files.
13. Write reusable functions and use try/except, pathlib, json, and logging.

```python
import json
import logging
from pathlib import Path


api_response = [
    {"id": 1, "host": " WEB-01 ", "status": "running",
     "cpu": "85", "region": "ap-south-1"},

    {"id": 2, "host": "web-02", "status": "stopped",
     "cpu": "0", "region": "ap-south-1"},

    {"id": 3, "host": "DB-01", "status": "running",
     "cpu": "95", "region": "ap-south-2"},

    {"id": 1, "host": "web-01", "status": "running",
     "cpu": "85", "region": "ap-south-1"},

    {"id": 4, "host": "web-04", "status": "running",
     "cpu": "invalid", "region": "ap-south-1"},

    {"id": 5, "host": "web-05", "status": "running",
     "cpu": "72", "region": "ap-south-2"},

    {"id": 6, "host": "web-06", "status": "unknown",
     "cpu": "50", "region": "ap-south-2"},

    {"id": 7, "host": "web-07", "status": "running",
     "cpu": "105", "region": "ap-south-1"},

    {"id": 8, "host": "web-08", "status": "running",
     "region": "ap-south-1"}
]


logging.basicConfig(
    filename="api_validation.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)


def transform_records(records):
    required_fields = {
        "id", "host", "status", "cpu", "region"
    }

    seen_ids = set()
    valid_records = []
    rejected_records = []
    duplicate_count = 0

    for record in records:
        reasons = []

        # 1. Check required fields.
        missing = required_fields - record.keys()

        if missing:
            reasons.append(
                f"Missing fields: {', '.join(sorted(missing))}"
            )

        # 2. Validate and normalize the hostname.
        host = record.get("host")

        if not isinstance(host, str) or not host.strip():
            reasons.append("Invalid hostname")
        else:
            host = host.strip().lower()

        # 3. Validate status.
        status = record.get("status")

        if not isinstance(status, str):
            reasons.append("Invalid status")
        else:
            status = status.strip().lower()

            if status not in {"running", "stopped"}:
                reasons.append("Unsupported status")

        # 4. Convert CPU to an integer.
        cpu = None

        try:
            raw_cpu = record.get("cpu")

            if isinstance(raw_cpu, bool):
                raise ValueError("Boolean is not a valid CPU value")

            if isinstance(raw_cpu, float) and not raw_cpu.is_integer():
                raise ValueError("CPU must be a whole number")

            cpu = int(raw_cpu)

            if not 0 <= cpu <= 100:
                reasons.append("CPU must be between 0 and 100")

        except (ValueError, TypeError):
            reasons.append("CPU is missing or invalid")

        # 5. Reject invalid records.
        if reasons:
            rejected_records.append({
                "record": record,
                "reasons": reasons
            })

            logging.warning(
                "Rejected record %s: %s",
                record.get("id"),
                "; ".join(reasons)
            )
            continue

        # 6. Remove duplicate IDs after validation.
        record_id = record["id"]

        if record_id in seen_ids:
            duplicate_count += 1
            logging.warning("Duplicate ID found: %s", record_id)
            continue

        seen_ids.add(record_id)

        # 7. Transform valid data.
        transformed = {
            "id": record_id,
            "host": host,
            "status": status,
            "cpu": cpu,
            "region": record["region"].strip().lower()
        }

        if cpu >= 90:
            transformed["cpu_category"] = "HIGH"
        elif cpu >= 70:
            transformed["cpu_category"] = "MEDIUM"
        else:
            transformed["cpu_category"] = "LOW"

        valid_records.append(transformed)

    return valid_records, rejected_records, duplicate_count


def generate_statistics(valid_records):
    regions = {}

    # 8. Group records by region.
    for record in valid_records:
        region = record["region"]

        regions.setdefault(region, []).append(record)

    summary = {}

    # 9. Calculate average CPU and highest CPU server.
    for region, records in regions.items():
        average_cpu = sum(
            record["cpu"] for record in records
        ) / len(records)

        highest = max(
            records,
            key=lambda record: record["cpu"]
        )

        summary[region] = {
            "server_count": len(records),
            "average_cpu": round(average_cpu, 2),
            "highest_cpu_server": highest["host"],
            "highest_cpu": highest["cpu"]
        }

    return summary


def save_json(path, data):
    Path(path).write_text(
        json.dumps(data, indent=4),
        encoding="utf-8"
    )


def main():
    valid, rejected, duplicates = transform_records(api_response)
    regional_summary = generate_statistics(valid)

    report = {
        "valid_count": len(valid),
        "rejected_count": len(rejected),
        "duplicate_count": duplicates,
        "valid_records": valid,
        "rejected_records": rejected,
        "regional_summary": regional_summary
    }

    try:
        save_json("valid_servers.json", valid)
        save_json("rejected_servers.json", rejected)
        save_json("validation_report.json", report)

        print(json.dumps(report, indent=4))
        print("\nAll reports saved successfully.")

    except OSError:
        logging.exception("Could not save validation reports")
        print("Unable to save reports. Check the file path or permissions.")


if __name__ == "__main__":
    main()
```
