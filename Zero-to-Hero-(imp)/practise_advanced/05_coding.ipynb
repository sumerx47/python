🔥 Q15 — File + Grouping

Suppose servers.txt contains:

web01,prod
web02,dev
web03,prod
web04,test
web05,prod

Group servers by environment.

```python
from pathlib import Path
def group_servers(filename):

    lines = Path(filename).read_text().splitlines()

    groups = {}

    for line in lines:
        name, environment = line.split(",")
        name = name.strip().lower()
        environment = environment.strip().lower()

        if environment not in groups:
            groups[environment] = []
            groups[environment].append(name)
    return groups

print(group_servers("servers.txt"))
```

🔴 Q21 — File + Nested Data

File:

web02,production,stopped,503

🔴 Q23 — File + JSON + Filtering + Sorting

From the same JSON data, find production servers and sort them by CPU usage.

🔴 Q24 — File + Regex + Dedup + Frequency

Log file:

ERROR web01 500 10.0.0.1
ERROR web01 500 10.0.0.1
ERROR web02 503 10.0.0.2
ERROR web01 502 10.0.0.1
INFO web03 200 10.0.0.3

Find unique error IPs and their frequencies.

```python
from pathlib import Path
import re

def error_ip_frequency(filename):
    lines = Path(filename).read_text().splitlines()

    seen = set()
    frequency = {}

    for line in lines:
        if not line.startswith("ERROR"):
            continue
        match = re.search(r"\b(?:\d{1,3}\.){3}\d{1,3}\b", line)

        if not match:
            continue
        ip = match.group()

        if line in seen:
            continue
        seen.add(line)

        frequency[ip] = frequency.get(ip,0)+1

    return frequency

print(error_ip_frequency("app.log"))
```

🚀🔥 Q30 — Teacher/Interview-Level Final Question

This combines almost everything you've learned so far.

Suppose production.log contains:

2026-09-29 ERROR web01 500 10.0.0.1 Database failed
2026-09-29 INFO web02 200 10.0.0.2 Request successful
2026-09-29 ERROR web01 500 10.0.0.1 Database failed
2026-09-29 ERROR web03 503 10.0.0.3 Service unavailable
2026-09-29 WARNING web02 404 10.0.0.2 Page missing
2026-09-29 ERROR web02 502 10.0.0.2 Gateway failed
2026-09-29 ERROR web03 503 10.0.0.3 Service unavailable

Create an analyzer that returns:

1. total logs
2. level frequency
3. total errors
4. errors by server
5. error-code frequency
6. unique error IPs
7. unique error messages
8. most problematic server
9. servers sorted by error count

```python
# QUESTION 30 ANSWER

from pathlib import Path
from collections import Counter

def analyze_production_log(filename):

    lines = Path(filename).read_text().splitlines()

    total_logs = 0
    level_frequency = Counter()
    errors_by_server = Counter()
    error_code_frequency = Counter()
    unique_error_ips = set()
    unique_error_messages = set()

    for line in lines:

        if not line.strip():
            continue

        total_logs += 1

        parts = line.split()

        level = parts[2]
        server = parts[3]
        code = int(parts[4])
        ip = parts[5]

        message = " ".join(parts[6:])

        level_frequency[level] += 1

        if level == "ERROR":

            errors_by_server[server] += 1

            error_code_frequency[code] += 1

            unique_error_ips.add(ip)

            unique_error_messages.add(message)

    total_errors = sum(errors_by_server.values())

    most_problematic_server = max(
        errors_by_server,
        key=errors_by_server.get
    )

    servers_sorted = sorted(
        errors_by_server.items(),
        key=lambda item: item[1],
        reverse=True
    )

    return {
        "total_logs": total_logs,
        "level_frequency": dict(level_frequency),
        "total_errors": total_errors,
        "errors_by_server": dict(errors_by_server),
        "error_code_frequency": dict(error_code_frequency),
        "unique_error_ips": sorted(unique_error_ips),
        "unique_error_messages": sorted(unique_error_messages),
        "most_problematic_server": most_problematic_server,
        "servers_sorted": servers_sorted
    }
```
