🟡 Q7. Create a reusable function

Create:

count_errors(filename)

It should return the number of ERROR lines.

```python
from pathlib import Path

def count_errors(filename):
    lines = Path(filename).read_text().splitlines()

    count = 0

    for line in lines:
        if "ERROR" in line:
            count += 1
    return count

print(count_errors("app.log"))
```

🟡 Level 3 — File + Frequency
Q8. Count log levels

app.log:

Return:
     {
        "INFO": 3,
        "ERROR": 2,
        "WARNING": 1
    }


```python
from pathlib import Path

def level_frequency(filename):
    lines = Path(filename).read_text().splitlines()

    frequency = {}

    for line in lines:
        level = line.split()[0]
        frequency[level] = frequency.get(level,0)+1
    return frequency

print(level_frequency("app.log"))
```

🟡 Q9. Find the longest line

Write:

longest_line(filename)

Return the longest line from the file.

```python
from pathlib import Path

def longest_line(filename):
    lines = Path(filename).read_text().splitlines()

    if not lines:
        return None

    return max(lines, key = len)

print(longest_line("app.log")) 
```

🟡 Q10. Find lines containing a keyword

Create:

find_lines(filename, keyword)

For:

find_lines("app.log", "database")

return every line containing "database" regardless of uppercase/lowercase.

```python
from pathlib import Path

def find_lines(filename, keyword):
    lines = Path(filename).read_text().splitlines()

    result = []

    keyword = keyword.lower()

    for line in lines:
        if keyword in line.lower():
            result.append(line)

    return result

print(find_lines("app.log", "database"))
```

🔥 Level 4 — File + Regex

Q11. Extract all IP addresses

Suppose network.log contains:

Connection from 192.168.1.10
Connection from 10.0.0.5
Connection from 192.168.1.20

Extract all IP addresses.

```python
import re
from pathlib import Path

def extract_ips(filename):
    text = Path(filename).read_text()

    pattern = r"\b(?:\d{1,3}\.){3}\d{1,3}\b"

    return re.findall(pattern,text)

print(extract_ips("network.log"))
```

🔥 Q12 — File + Regex + Frequency

Using the same type of network log, count how many times each IP appears.

```python
from pathlib import Path
import re

def ip_frequency(filename):
    text = Path(filename).read_text()

    pattern =  r"\b(?:\d{1,3}\.){3}\d{1,3}\b"

    ips = re.findall(pattern, text)

    frequency = {}

    for ip in ips:
        frequency[ip] = frequency.get(ip,0)+1

    return frequency

print(ip_frequency("network.log"))
```

🔥 Q13 — File + Deduplication

servers.txt:

web01
web02
web01
web03
web02
web04

Return unique server names while preserving order.

```python
from pathlib import Path

def unique_servers(filename):
    lines = Path(filename).read_text().splitlines()

    seen = set()
    result = []

    for line in lines:
        server = line.strip()
        if server not in seen:
            seen.add(server)
            result.append(server)
    return result

print(unique_servers("servers.txt"))
```

🔥 Q15 — File + Grouping

Suppose servers.txt contains:

web01,prod
web02,dev
web03,prod
web04,test
web05,prod

Group servers by environment.

```python
from pathlib import Path
def group_servers(filename):

    lines = Path(filename).read_text().splitlines()

    groups = {}

    for line in lines:
        name, environment = line.split(",")
        name = name.strip().lower()
        environment = environment.strip().lower()

        if environment not in groups:
            groups[environment] = []
            groups[environment].append(name)
    return groups

print(group_servers("servers.txt"))
```
