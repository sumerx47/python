```python
from pathlib import Path
lines = Path("server.txt").read_text().splitlines()

count = 0

for line in lines:
    if line.strip():
        count += 1
print(count)
```

🟡 Level 2 — File + String Processing
Q5. Count ERROR lines

Suppose app.log contains:

INFO Server started
ERROR Database failed
INFO Request completed
ERROR Connection timeout
WARNING High memory
ERROR API failed

Count the ERROR lines.

```python
from pathlib import Path
lines = Path("app.log").read_text().splitlines()

count = 0

for line in lines:
    if "ERROR" in line:
        count += 1
print(count)
```

🟡 Q6. Extract ERROR lines

Using the same file, return a list containing only ERROR lines.

```python
from pathlib import Path
lines = Path("app.log").read_text().splitlines()

errors = []

for line in lines:
    if "ERROR" in line:
        errors.append(line)
print(errors)
```

🟡 Q7. Create a reusable function

Create:

count_errors(filename)

It should return the number of ERROR lines.

```python
from pathlib import Path

def count_errors(filename):
    lines = Path(filename).read_text().splitlines()

    count = 0

    for line in lines:
        if "ERROR" in line:
            count += 1
    return count

print(count_errors("app.log"))
```

🟡 Level 3 — File + Frequency
Q8. Count log levels

app.log:

Return:
     {
        "INFO": 3,
        "ERROR": 2,
        "WARNING": 1
    }


```python
from pathlib import Path

def level_frequency(filename):
    lines = Path(filename).read_text().splitlines()

    frequency = {}

    for line in lines:
        level = line.split()[0]
        frequency[level] = frequency.get(level,0)+1
    return frequency

print(level_frequency("app.log"))
```
