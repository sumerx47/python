```python
# Q10. Create a reusable word-length function .  Write:    get_long_words(text)

def get_long_words(text):
    words = text.split()

    result = []

    for word in words:
        if len(word) >= 5:
            result.append(word)
    return result

text = "Python is powerful for DevOps automation and it is also Amazing Language"

print(get_long_words(text))
```

```python
# Q12. Count words ignoring uppercase/lowercase

text = "Python python PYTHON DevOps devops"

frequency = {}

for word in text.split():
    word = word.lower()
    frequency[word] = frequency.get(word,0)+1

print(frequency)
```

```python
# Q13. Find the most frequent word
text = "python devops python docker devops python"

frequency = {}

for word in text.split():
    frequency[word] = frequency.get(word,0)+1

most_frequent = max(frequency, key = frequency.get)

print(most_frequent)
```

Q14. Clean and analyze server tags

You receive:

tags = [
    " DEVOPS ",
    "python",
    "PYTHON",
    " Docker ",
    "devops",
    "docker"
]

Create a cleaned frequency dictionary where:

Spaces are removed.
Everything becomes lowercase.
Duplicate names are counted.

```python
tags = [
    " DEVOPS ",
    "python",
    "PYTHON",
    " Docker ",
    "devops",
    "docker"
]

frequency = {}
for tag in tags:
    tag = tag.strip().lower()
    frequency[tag] = frequency.get(tag,0)+1

print(frequency)
```

🔥 Q15 — Advanced Function

Write:

analyze_text(text)

It should return:

total words
unique words
most frequent word
total characters

```python
def analyze_text(text):
    words = text.lower().split()
    frequency = {}

    for word in words:
        frequency[word] = frequency.get(word,0)+1

    total_words = len(words)
    unique_words = len(frequency)
    total_characters = len(text)

    most_frequent_word = max(frequency, key = frequency.get)

    return {
        "total_words":total_words,
        "unique_words":unique_words,
        "total_characters":total_characters,
        "most_frequent_word":most_frequent_word
    }


text = "Python is powerful Python is useful"
print(analyze_text(text))
```
