🔥 Q15 — Advanced Function

Write:

analyze_text(text)

It should return:

total words
unique words
most frequent word
total characters

```python
def analyze_text(text):
    words = text.lower().split()
    frequency = {}

    for word in words:
        frequency[word] = frequency.get(word,0)+1

    total_words = len(words)
    unique_words = len(frequency)
    total_characters = len(text)

    most_frequent_word = max(frequency, key = frequency.get)

    return {
        "total_words":total_words,
        "unique_words":unique_words,
        "total_characters":total_characters,
        "most_frequent_word":most_frequent_word
    }


text = "Python is powerful Python is useful"
print(analyze_text(text))
```

🔥 Q16 — Function + String Cleaning + Frequency

Write a function:

analyze_words(text)

Requirements:

1. Convert everything to lowercase.
2. Remove . , ! ?
3. Count word frequency.
4. Return the frequency dictionary.

```python
def analyze_words(text):
    text = text.lower()
    symbols = ",.!?"

    frequency = {}

    for symbol in symbols:
        text = text.replace(symbol,"")

    for word in text.split():
        frequency[word] = frequency.get(word,0)+1

    return frequency



text = "Python, Python! DevOps is powerful. DevOps is Python?"

print(analyze_words(text))
```

🔥 Q17 — Function + Filtering + Frequency

Write a function that returns only words appearing at least twice.

text = "python docker kubernetes python docker aws python linux"

```python
def repeated_words(text):
    frequency = {}

    for word in text.lower().split():
        frequency[word] = frequency.get(word,0)+1
    
    result = {}

    for word, count in frequency.items():
        if count >= 2:
            result[word] = count

    return result

text = "python docker kubernetes python docker aws python linux"

print(repeated_words(text))

```

🔥 Q18 — Function + String Validation
Rules:

1. Must contain only letters, numbers and _
2. Must be at least 5 characters
3. Must not contain spaces
4. Must start with a letter

```python
def validate_username(username):

    if len(username) < 5:
        return False

    if " " in username:
        return False

    if not username[0].isalpha():
        return False

    if not username.isalnum() and "_" not in username:
        return False

    for char in username:
        if not (char.isalnum() or char == "_"):
            return False
        
    return True

print(validate_username("sumer_55"))
print(validate_username("125sumer_"))
print(validate_username("sam258sumer"))
print(validate_username("sumer@55"))

```

🔥 Q26 — Advanced JSON + Regex

Given:

data = {
    "logs": [
        {"message": "Connection from 192.168.1.10 failed"},
        {"message": "Connection from 10.0.0.5 succeeded"},
        {"message": "Connection from 192.168.1.10 failed"}
    ]
}

Extract all IP addresses and count their frequency.

```python
data = {
    "logs": [
        {"message": "Connection from 192.168.1.10 failed"},
        {"message": "Connection from 10.0.0.5 succeeded"},
        {"message": "Connection from 192.168.1.10 failed"}
    ]
}

import re

def ip_frequency(data):
    frequency = {}

    for log in data["logs"]:
        ips = re.findall(r"\b(?:\d{1,3}\.){3}\d{1,3}\b", log["message"])

        for ip in ips:
            frequency[ip] = frequency.get(ip,0)+1

    return frequency

print(ip_frequency(data))
    

```

🔥 Q28 — Production Log Analyzer

This is much closer to your teacher's test.

Given:

logs = [
    "ERROR web01 Database failed",
    "INFO web02 Request completed",
    "ERROR web01 Connection failed",
    "WARNING web03 High memory",

    "ERROR web01 Database failed"

```python
logs = [
    "ERROR web01 Database failed",
    "INFO web02 Request completed",
    "ERROR web01 Connection failed",
    "WARNING web03 High memory",
    "ERROR web02 Timeout",
    "ERROR web01 Database failed"
]

def analyze_logs(logs):


    level_frequency = {}
    error_by_server = {}

    for log in logs:
        parts = log.split()
        level = parts[0]
        server = parts[1]

        level_frequency[level] = level_frequency.get(level,0)+1

        if level == "ERROR":
            error_by_server[level] = error_by_server.get(level,0)+1

    total_errors = sum(error_by_server.values())

    most_problematic_server = max(error_by_server, key = error_by_server.get)

    return {
    "level_frequency": level_frequency,
    "error_by_server": error_by_server,
    "total_errors": total_errors,
    "most_problematic_server": most_problematic_server
}

print(analyze_logs(logs))
```

🚨 Q29 — Teacher-Test Level

You receive server information:

Requirements:

1. Normalize server names.
2. Normalize environment.
3. Consider only RUNNING servers.
4. Deduplicate by server name.
5. Count total errors for each server.
6. Return the server with the highest error count.

```python
servers = [
    {
        "name": " WEB01 ",
        "environment": " PROD ",
        "status": "RUNNING",
        "logs": [
            "ERROR Database failed",
            "INFO Request completed"
        ]
    },
    {
        "name": "web02",
        "environment": "DEV",
        "status": "STOPPED",
        "logs": [
            "ERROR Timeout"
        ]
    },
    {
        "name": "WEB01",
        "environment": "prod",
        "status": "RUNNING",
        "logs": [
            "ERROR Database failed",
            "ERROR Connection failed"
        ]
    }
]
```

```python
def analyze_servers(servers):

    server_data = {}

    for server in servers:
        name = server["name"].strip().lower()
        environment = server["environment"].strip().lower()
        status = server["status"].strip().lower()

        if status != "running":
            continue

        if name not in server_data:
            server_data[name] = {"environment":environment, "error_count": 0}

        for log in server["logs"]:
            if log.strip().upper().startswith("ERROR"):
                server_data[name]["error_count"] += 1

            if not server_data :
                return {}

        most_problametic_server = max(server_data , key =lambda name:server_data[name]["error_count"])

    return {
        "server_data":server_data,
        "most_problametic_server":most_problametic_server
    }

print(analyze_servers(servers))

```

💀 Q30 — Full Teacher-Test Challenge

This is your Pattern 14 final challenge.

It must return:

1. Total log count
2. Frequency of each level
3. Total errors
4. Errors by server
5. HTTP error-code frequency
6. Most problematic server
7. A list of unique error messages

```python
logs = [
    "2026-09-28 10:00:00 ERROR web01 500 Database failed",
    "2026-09-28 10:01:00 INFO web02 200 Request completed",
    "2026-09-28 10:02:00 ERROR web01 503 Service unavailable",
    "2026-09-28 10:03:00 WARNING web03 404 Page missing",
    "2026-09-28 10:04:00 ERROR web02 500 Database failed",
    "2026-09-28 10:05:00 ERROR web01 500 Database failed",
    "2026-09-28 10:06:00 INFO web03 200 Request completed",
    "2026-09-28 10:07:00 ERROR web02 503 Service unavailable"
]
```

```python
def analyze_production_logs(logs):
    total_logs = len(logs)

    level_frequency = {}
    errors_by_server = {}
    error_code_frequency = {}
    unique_error_messages = set()

    for log in logs:
        parts = log.split()
        level = parts[2]
        server = parts[3]
        code = int(parts[4])
        message = "".join(parts[5:])

        #level frequency
        level_frequency[level] = (level_frequency.get(level,0)+1)

        # Process errors
        if level == "ERROR":

            # Error count by server
            errors_by_server[server] = (errors_by_server.get(server, 0) + 1)

            # Error code frequency
            error_code_frequency[code] = (error_code_frequency.get(code, 0) + 1)

            # Unique error messages
            unique_error_messages.add(message)

    total_errors = sum(errors_by_server.values())

    most_problematic_server = max(errors_by_server,key=errors_by_server.get)

    return {
        "total_logs": total_logs,
        "level_frequency": level_frequency,
        "total_errors": total_errors,
        "errors_by_server": errors_by_server,
        "error_code_frequency": error_code_frequency,
        "most_problematic_server": most_problematic_server,
        "unique_error_messages": sorted(
            unique_error_messages
        )
    }

print(analyze_production_logs(logs))
```
