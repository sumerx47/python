# String interpolation

```python
open_,high,low,close=95,100,90,99
```

```python
'open: '+ str(open_) + ', high: '+ str(high)+', low:'+ str(low)+', close:'+ str(close)
```

#### best approach

```python
'open: {}, high: {}, low: {}, close: {}'.format(open_,high,low,close)
```

```python
bid=1.56760
ask=1.5763
```

```python
'bid: {}, ask: {}, spread: {}'.format(bid,ask, ask-bid)
```

```python
f'bid: {bid}, ask: {ask}, spread: {ask-bid}'
```

```python
f"bid: {bid}, ask: {ask}, spread: {ask-bid}"
```

```python
'bid: {b}, ask: {a}, spread: {s}'.format(b=bid,a=ask, s=ask-bid)
```

```python
'bid: {b:.4f}, ask: {a:.4f}, spread: {s:.4f}'.format(b=bid,a=ask, s=ask-bid)
```
