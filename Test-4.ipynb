```python
'''Count the occurrence of each word in a sentence.
output:
python : 2
is : 2
easy : 1
powerful : 1'''

sentence = "python is easy python is powerful"
```

```python
words= sentence.lower().split()
words
```

```python
freq = {}
```

```python
for word in words:
    if word in freq:
        freq = word + 1  
print(freq)
```
