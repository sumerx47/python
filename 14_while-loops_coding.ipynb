# while-loops coding

```python
price = 100
```

```python
while price > 90:
    print(f"price = {price} - waiting for price to come down...")
    price -=1
print(f'buying at {price}')
```

```python
amount=100 
```

```python
while amount > 90: print(f"{amount} - waiting for price to come down..." ); amount -=1 ; print(f"Buying at: {amount}")
```
