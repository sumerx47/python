```python
l2=l1[:]
```

```python
li=[[1,0,0],[0,1,0],[0,0,1]]
```

```python
li2=li.copy()
```

```python
li2
```

```python
li2 is l1
```

```python
li2.append([2,2,2])
li2
```

```python
li
```
