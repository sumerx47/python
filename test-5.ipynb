```python
sentence = "python is easy python is powerful"
```

```python
count={}
```

```python
words = sentence.split()
print(words)
for word in words:
    if word in count:
        count[word] += 1
    else:
        count[word] = 1

count
```

```python
 for word in words:
     count[word] = count.get(word,0)+1
     
```
