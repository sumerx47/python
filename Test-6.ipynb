```python
'''
QUESTION 4 — Search by Skill
Create a function:
find_agents_by_skill(skill)
The user will enter a skill such as:
Python
The function should return a list containing the names of all agents who have that skill.
'''
def find_agents_by_skill(skill):
    result = []
    for agent_id, details in agents.items():
        if skill in details["skills"]:
            result.append(details["name"])
    return result
skill = find_agents_by_skill("Python")
print("Agents with skill Python:",skill)
print("*" * 15)

''' 
QUESTION 5 — Find Unique Skills
Create a function:
get_all_skills()
Find all the different skills available among all agents.
Duplicate skills must not appear.
Think about which Python data type can automatically remove duplicates.
'''
def get_all_skills():
    all_skills = []
    for agent_id, details in agents.items():
        for skill in details["skills"]:
            all_skills.append(skill)
    return all_skills
skills = get_all_skills()
print("unique skills:",list(set(skills)))
print("*" * 15)

'''
QUESTION 6 — Add a New Agent
Create a function:
add_agent()
Ask the user to enter:
• Agent ID
• Agent Name
• 3 Skills
• 3 Mission Scores
Add the new agent to the agents dictionary.
Important: If the Agent ID already exists, the program should display an error and should not overwrite the existing agent.
'''
def add_agent(agent_id,name,skills,missions):
    if agent_id in agents:
        print("ERROR: Agent id already exists")
        return

    agents[agent_id] = {
        "name": name,
        "skills": skills,
        "missions":missions
    }
    print("Agent added successfully")

add_agent(
        "A105",
        "JOHN",
        ["Python", "Linux", "Java"],
        [82,91,73]
    )
print("New Agent:",agents["A105"])
print("*" * 15)


'''
QUESTION 7 — Mission Status
Create a function:
mission_status(agent_id)
Calculate the selected agent's average mission score and return:
Elite → Average >= 90
Active → Average >= 75 and < 90
Training Required → Average < 75
Also handle the situation where the entered Agent ID does not exist.
'''
def mission_status(agent_id):
    if agent_id not in agents:
        return "Agent ID does not Exist"

    scores = agents[agent_id]["missions"]
    
    total, average = calculate_average(scores)

    if average >= 90:
        return "Elite"
    elif average >= 75:
        return "Active"
    else:
        return "Training Required"
print("A101 Status:", mission_status("A101"))
print("A102 Status:", mission_status("A102"))
print("A103 Status:", mission_status("A103"))
print("A104 Status:", mission_status("A104"))
print("A999 Status:", mission_status("A999"))
print("*" * 15)
```

```python
'''
QUESTION 8 — 🔥 BONUS CHALLENGE
Create a function:
skill_ranking()
Find the average performance of agents belonging to each skill.
For example, suppose the agents having Python skill have average mission scores:
85, 91, 94
Then calculate the overall performance of agents with the Python skill.
Do this for every available skill.
'''

def skill_ranking():
    skill_agents = {}

    # Group agents according to their skills
    for agent_id, details in agents.items():
        total, average = calculate_average(details["missions"])

        for skill in details["skills"]:
            if skill not in skill_agents:
                skill_agents[skill] = []
            skill_agents[skill].append(average)

    # Calculate average performance for each skill
    ranking = {}
    for skill, averages in skill_agents.items():
        total = 0

        for average in averages:
            total += average
        ranking[skill] = total / len(averages)

    return ranking

ranking = skill_ranking()
for skill, average in ranking.items():
    print(
        skill,"→",round(average, 2)
    )
```

```python
'''
Q4 — Search by Department
Create:
find_patients_by_department(department)
For example:
find_patients_by_department("Cardiology")
Return a list containing the names of all patients in that department.
'''

def find_patients_by_department(department):
    result=[]

    for patient_ID, details in patients.items():
        if details["department"] == department:
            result.append(details["name"])
    return result
Cardiology_patients = find_patients_by_department("Cardiology")
print("Cardiology patients:", Cardiology_patients)

        
'''
Q5 — Find Unique Tests
Create:
get_all_tests()
Find all different medical tests performed by the hospital.
Duplicates must not appear.
Hint: Which data type automatically removes duplicates?
'''
def get_all_tests():
    all_tests = set()

    for details in patients.values():
        for test in details["tests"]:
            all_tests.add(test)
    return all_tests
tests = get_all_tests()
print("All tests:" , tests)


'''
Q6 — Add a New Patient
Create:
add_patient(patient_id, name, age, department, tests, scores)
Add the patient to the dictionary.
If the Patient ID already exists
Error: Patient already exists.
Do not overwrite the existing patient.

Test both:
add_patient(
    "P105",
    "Omar",
    40,
    "Neurology",
    ["MRI", "Blood"],
    [80, 85, 90]
)

and an existing ID:

add_patient(
    "P101",
    "Someone",
    50,
    "Cardiology",
    ["ECG"],
    [90, 90, 90]
)

'''

def add_patient(patient_id, name, age, department, tests, scores):

    if patient_id in patients:
        print("Error: patient id already exists")
        return
    patients[patient_id] = {
        "name": name,
        "age": age,
        "department": department,
        "tests": tests,
        "scores": scores
    }
print("patient id successfully added")

add_patient(
    "P105",
    "Omar",
    40,
    "Neurology",
    ["MRI", "Blood"],
    [80, 85, 90]
)

print("new patient: " , patients["P105"])

add_patient(
    "P101",
    "Someone",
    50,
    "Cardiology",
    ["ECG"],
    [90, 90, 90]
)

print("new patient: " , patients["P101"])
```
