7. palindrome number

```python
num= 121
reverse = 0
```

```python
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10

if num == reverse:
    print("it is palindrome")
else: 
    print("not a palindrome")
```

1. count uppercase and lowercase letters

```python
text = "python is easy, PYTHON IS EASY"
```

```python
lowercase=0
uppercase=0
remaining_characters=0
```

```python
for i in text:
    if i.islower():
        lowercase += 1
        
    elif i.isupper():
       uppercase += 1

    elif i == ",":
        remaining_characters += 1
        
print(lowercase, uppercase, remaining_characters)
```
