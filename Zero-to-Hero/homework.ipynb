7. palindrome number

```python
num= 121
reverse = 0
```

```python
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10

if num == reverse:
    print("it is palindrome")
else: 
    print("not a palindrome")
```

4. count vowels, consenents, digits, spaces, special characters

```python
text = "hi!, this is SUMAIR. How are you? my num is 799566"
```

```python
vowels = 0
consonents =0
digits = 0
spaces = 0
special_characters =0
```

```python
for i in text:
    if i.isalpha():
        if i.lower() in "aeiou":
            vowels += 1
        else: 
            consonents += 1
    elif i.isdigit():
        digits += 1
    elif i.isspace():
        spaces += 1
    else:
        special_characters += 1

    
```

```python
print(vowels)
print(consonents)
print(digits)
print(spaces)
print(special_characters)
```
