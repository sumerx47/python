7. palindrome number

```python
num= 121
reverse = 0
```

```python
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10

if num == reverse:
    print("it is palindrome")
else: 
    print("not a palindrome")
```
