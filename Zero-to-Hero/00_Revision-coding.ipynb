## 1.max()  
#### method - 1 

```python
products = [
    {"name": "Laptop", "price": 60000, "stock": 5},
    {"name": "Mouse", "price": 800, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 10},
    {"name": "Monitor", "price": 12000, "stock": 3}
]
```

```python
expensive = max(products, key = lambda x : x ["price"])
print(expensive)
```

#### method - 2

```python
expensive = products[0]
```

```python
for i in products:
    if i["price"] > expensive["price"]:
        expensive = i
print(expensive)
```

## 2.min()  
#### method - 1 

```python
cheapest = min(products, key = lambda x : x["price"])
print(cheapest)
```

#### method - 2

```python
cheap = products[0]
```

```python
for item in products:
    if item["price"] < cheap["price"]:
        cheap = item 
print(cheap)
```
