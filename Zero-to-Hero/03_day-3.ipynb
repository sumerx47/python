### 24. multiplication table

```python
num=5
```

```python
for i in range(11):
    print(f"{num} x {i} =", num*i)
```

```python
for i in range(1,11):
    print(f" TABLE:{i}")
    for j in range(1,11):
        print(f"{i} x {j} =", i*j)
    print("*" * 20)
```
