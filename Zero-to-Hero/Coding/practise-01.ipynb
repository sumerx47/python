Question 1 – Highest Salary + Count Salary

Tasks:

1. Find the employee with the highest salary.


2. Count employees having the same salary.

```python
employees = [
    {"name":"Ali","salary":45000},
    {"name":"luke","salary":70000},
    {"name":"Sara","salary":55000},
    {"name":"Ahmed","salary":45000},
    {"name":"John","salary":70000},
    {"name":"Sara","salary":55000},
    {"name":"Eva","salary":55000}
]
```

```python
# method:1 , max
highest = employees[0]
for emp in employees:
    if emp["salary"] > highest["salary"]:
        highest = emp
highest
```

```python
# method: 2
highestt = max(employees, key = lambda x : x["salary"])
highestt
```

```python
# method: 3
highest_sal = max(emp["salary"] for emp in employees)
for emp in employees:
    if emp["salary"] == highest_sal:
        print(emp)
```

```python
# method 1, count
count_emp = {}
for emp in employees:
    salary = emp["salary"]
    if salary in count_emp:
        count_emp[salary] += 1
    else :
        count_emp[salary] = 1
count_emp
```

```python
# method 2, count
count_empl = {}
for emp in employees:
    salary = emp["salary"]
    count_empl[salary]=count_empl.get(salary,0)+1

count_empl
```

Question 11 – Employee Skills

Tasks:

```python
employees = [
    {"name":"Ali","skills":["Python","Linux"]},
    {"name":"Sara","skills":["Python","Docker"]},
    {"name":"Ahmed","skills":["Linux","Docker"]},
    {"name":"John","skills":["Python","Git"]},
    {"name":"Fatima","skills":["Docker","Git"]}
]
```
