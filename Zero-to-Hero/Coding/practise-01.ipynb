Question 1 – Highest Salary + Count Salary

Tasks:

1. Find the employee with the highest salary.


2. Count employees having the same salary.

```python
employees = [
    {"name":"Ali","salary":45000},
    {"name":"luke","salary":70000},
    {"name":"Sara","salary":55000},
    {"name":"Ahmed","salary":45000},
    {"name":"John","salary":70000},
    {"name":"Sara","salary":55000},
    {"name":"Eva","salary":55000}
]
```

```python
# method:1 , max
highest = employees[0]
for emp in employees:
    if emp["salary"] > highest["salary"]:
        highest = emp
highest
```

```python
# method: 2
highestt = max(employees, key = lambda x : x["salary"])
highestt
```

```python
# method: 3
highest_sal = max(emp["salary"] for emp in employees)
for emp in employees:
    if emp["salary"] == highest_sal:
        print(emp)
```

```python
# method 1, count
count_emp = {}
for emp in employees:
    salary = emp["salary"]
    if salary in count_emp:
        count_emp[salary] += 1
    else :
        count_emp[salary] = 1
count_emp
```

```python
# method 2, count
count_empl = {}
for emp in employees:
    salary = emp["salary"]
    count_empl[salary]=count_empl.get(salary,0)+1

count_empl
```

Question 6 – DevOps Logs

Tasks:

1. Count each log level.


2. Which log level occurs most?

```python
logs = [
    {"level":"INFO"},
    {"level":"ERROR"},
    {"level":"WARNING"},
    {"level":"INFO"},
    {"level":"ERROR"},
    {"level":"INFO"},
    {"level":"CRITICAL"}
]
```

```python
level_count = {}
```

```python
for log in logs:
    level = log["level"]
    if level in level_count:
        level_count[level] += 1
    else:
        level_count[level] = 1

level_count
    
```

```python
most_occured_log = max(level_count, key= level_count.get)
most_occured_log
```

```python
level_counts = {}
for log in logs:
    level = log["level"]
    level_counts[level] = level_counts.get(level,0)+1


most_occured_logs = max(level_counts , key = level_counts.get)
print("Count each log level: ",level_counts)
print("log level that occurs most: ",most_occured_logs)
```

Question 7 – Docker Containers

Tasks:

1. Count each status.


2. Which status occurs most?

```python
containers = [
    {"name":"nginx","status":"running"},
    {"name":"redis","status":"running"},
    {"name":"mysql","status":"stopped"},
    {"name":"jenkins","status":"running"},
    {"name":"mongodb","status":"stopped"}
]
```

```python
status_count = {}
for container in containers:
    status = container["status"]
    if status in status_count:
        status_count[status] +=1
    else:
        status_count[status] = 1

most_occured_status = max(status_count, key = status_count.get)

print("Count each status: ",status_count)
print("status that occurs most: ",most_occured_status)
```

```python
status_counts = {}
for container in containers:
    status = container["status"]
    status_counts[status] = status_counts.get(status,0) + 1

most_occured_status = max(status_counts, key = status_counts.get)

print("Count each status: ",status_counts)
print("status that occurs most: ",most_occured_status)
```
