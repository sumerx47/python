🟢 Level 1 (Intermediate)

📘 Question 1 – Count Same Salary

```python
employees = [
    {"name":"Ali","salary":45000},
    {"name":"Sara","salary":55000},
    {"name":"Ahmed","salary":45000},
    {"name":"John","salary":60000},
    {"name":"Eva","salary":55000}
]
```

1. ✅ Method 1 (if...else) - Best for building logic

```python
salary_count = {}
```

```python
for employee in employees: #entering into list
    salary  = employee["salary"]
    if salary in salary_count: #checking salary in empty dict
        salary_count[salary]+=1
    else:
        salary_count[salary]=1
print(salary_count)
```

2. ✅ Method 2 (dict.get()) - Best for interviews

```python
salary_count = {}
```

```python
for emp in employees:
    salary = emp["salary"]
    salary_count[salary] = salary_count.get(salary,0)+1
print(salary_count)
```

Question 14 – Count File Extensions

Task: Count file extensions.

```python
files = [
    "app.py",
    "main.py",
    "docker.yaml",
    "config.yaml",
    "notes.txt",
    "test.py"
]
```

```python
extension_count = {}
```

```python
for file in files:
    extension = file.split(".")[1]
    if extension in extension_count:
        extension_count[extension] += 1
    else:
        extension_count[extension]=1
extension_count
```

```python
count_extension = {}
for file in files:
    extension = file.split(".")[1]
    count_extension[extension] = count_extension.get(extension,0)+1
count_extension
```
