Question 1 – Lowest Salary

Find: Lowest salary employee

```python
employees = [
    {"name": "Ali", "salary": 45000},
    {"name": "Sara", "salary": 55000},
    {"name": "zaq", "salary": 45000},
    {"name": "John", "salary": 65000},
    {"name": "Ahmed", "salary": 50000},
    {"name": "blake", "salary": 65000}
]
```

✅ Method 1: Manual loop (most important for interviews)



```python
lowest_salary = employees[0]
```

```python
for employee in employees:
    if employee["salary"] < lowest_salary["salary"]:
        lowest_salary = employee
lowest_salary
```

✅ Method 2: Using min()



```python
low_salary = min(employees, key = lambda emp : emp["salary"])
low_salary
```

✅ Method 3: Using sorted()



```python
lowest_salary = sorted(employees, key = lambda emp:emp["salary"])
lowest_salary[0]
```

✅ Method 4: Find all employees with the same lowest salary



```python
lowest_salary = min(emp["salary"] for emp in employees)

for emp in employees:
    if emp["salary"] == lowest_salary:
        print(emp)
```
