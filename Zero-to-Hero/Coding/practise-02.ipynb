Question 1

Tasks

1. Highest salary employee.


2. Lowest salary employee.


3. Count salaries.


4. Count departments.


5. Print employees earning above ₹50,000.


6. Print only IT employees.

```python
employees = [
    {"name":"Ali","salary":45000,"dept":"IT"},
    {"name":"Sara","salary":65000,"dept":"HR"},
    {"name":"Ahmed","salary":55000,"dept":"IT"},
    {"name":"John","salary":70000,"dept":"Finance"},
    {"name":"Fatima","salary":45000,"dept":"HR"}
]
```

```python
highest = employees[0]
lowest =employees[0]
salary_count={}
dept_count={}


for emp in employees:
    if emp["salary"] > highest["salary"]:
        highest = emp
    
    if emp['salary'] < lowest['salary']:
        lowest = emp
        
    salary = emp["salary"]
    if salary in salary_count:
        salary_count[salary] += 1
    else :
        salary_count[salary] =1

    dept =emp["dept"]
    if dept in dept_count:
        dept_count[dept] += 1
    else:
        dept_count[dept] = 1

print("Highest salary employee: ",highest["name"], highest["salary"])
print("Lowest salary employee: ",lowest["name"], lowest["salary"])
print("salaries Count: ",salary_count)
print("departments Count: ",dept_count)
print("*" * 15)

for emp in employees:
    if emp["salary"] > 50000:
        print("employees earning above ₹50,000: ",emp["name"])
print("*" * 15)
for emp in employees:
    if emp["dept"] =="IT":
        print("only IT employees: ",emp["name"])
```

```python
logs = [
    {"level":"INFO"},
    {"level":"ERROR"},
    {"level":"INFO"},
    {"level":"WARNING"},
    {"level":"ERROR"},
    {"level":"CRITICAL"},
    {"level":"INFO"}
]
```

```python
log_level_count={}
for log in logs:
    level = log["level"]
    if level in log_level_count:
        log_level_count[level] += 1
    else:
        log_level_count[level] = 1

most_common_log_level=max(log_level_count, key=log_level_count.get)

print("log level_count: ",log_level_count)
print("most common log level: ",most_common_log_level)
print("*" * 15)

for log in logs:
    if log["level"] == "ERROR":
        print("Error logs: ",log)
print("*" * 15)
        
for log in logs:
    if log["level"] == "INFO":
        print("INFO logs: ",log["level"]) 
```
