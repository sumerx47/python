Question 1

Tasks

1. Highest salary employee.


2. Lowest salary employee.


3. Count salaries.


4. Count departments.


5. Print employees earning above ₹50,000.


6. Print only IT employees.

```python
employees = [
    {"name":"Ali","salary":45000,"dept":"IT"},
    {"name":"Sara","salary":65000,"dept":"HR"},
    {"name":"Ahmed","salary":55000,"dept":"IT"},
    {"name":"John","salary":70000,"dept":"Finance"},
    {"name":"Fatima","salary":45000,"dept":"HR"}
]
```

```python
highest = employees[0]
lowest =employees[0]
salary_count={}
dept_count={}


for emp in employees:
    if emp["salary"] > highest["salary"]:
        highest = emp
    
    if emp['salary'] < lowest['salary']:
        lowest = emp
        
    salary = emp["salary"]
    if salary in salary_count:
        salary_count[salary] += 1
    else :
        salary_count[salary] =1

    dept =emp["dept"]
    if dept in dept_count:
        dept_count[dept] += 1
    else:
        dept_count[dept] = 1

print("Highest salary employee: ",highest["name"], highest["salary"])
print("Lowest salary employee: ",lowest["name"], lowest["salary"])
print("salaries Count: ",salary_count)
print("departments Count: ",dept_count)
print("*" * 15)

for emp in employees:
    if emp["salary"] > 50000:
        print("employees earning above ₹50,000: ",emp["name"])
print("*" * 15)
for emp in employees:
    if emp["dept"] =="IT":
        print("only IT employees: ",emp["name"])
```

```python
servers = [
    {"name":"web01","status":"Running","cpu":55},
    {"name":"db01","status":"Stopped","cpu":92},
    {"name":"app01","status":"Running","cpu":70},
    {"name":"cache01","status":"Stopped","cpu":40}
]
```

```python
highest_server = servers[0]
lowest_server = servers[0]
status_count={}

for server in servers:
    if server["cpu"] > highest_server["cpu"]:
        highest_server = server 
    if server["cpu"] < lowest_server["cpu"]:
        lowest_server = server
    
    status = server["status"]
    if status in status_count:
        status_count[status] += 1
    else:
        status_count[status] = 1

print("Highest server: ",highest_server["name"], highest_server["cpu"])
print("Lowest server: ",lowest_server["name"], lowest_server["cpu"])
print("status Count: ",status_count)
print("*" * 15)

for server in servers:
    if server["status"] == "Running":
        print("server status: ",server["name"],server["status"])
print("*" * 15)
for server in servers:
    if server["cpu"] > 60:
        print("server with cpu above 60: ",server["name"],server["cpu"])
```

```python
products = [
    {"name":"Laptop","category":"Electronics","price":60000},
    {"name":"Mouse","category":"Electronics","price":800},
    {"name":"Chair","category":"Furniture","price":3000},
    {"name":"Monitor","category":"Electronics","price":12000},
    {"name":"Table","category":"Furniture","price":7000}
]
```

```python
categories_count={}

most_expensive= max(products, key = lambda product : product["price"])
cheapest_product = min(products, key = lambda product : product["price"])
for product in products:
    category = product["category"]
    categories_count[category] = categories_count.get(category,0)+1
print("most expensive product: ",most_expensive["name"], most_expensive["price"])
print("cheapest product: ",cheapest_product["name"], cheapest_product["price"])
print("categories count: ",categories_count)
print("*" * 15)

for product in products:
    if product["category"] == "Electronics":
        print("Electronics products: ",product["name"])
print("*" * 15)

for product in products:
    if product["price"] > 10000:
        print("products above 10,000: ",product["name"])


```

Question 5

Tasks

1. Count log levels.


2. Most common log level.


3. Print ERROR logs.


4. Print INFO logs.


```python
logs = [
    {"level":"INFO"},
    {"level":"ERROR"},
    {"level":"INFO"},
    {"level":"WARNING"},
    {"level":"ERROR"},
    {"level":"CRITICAL"},
    {"level":"INFO"}
]
```

```python
log_level_count={}
for log in logs:
    level = log["level"]
    if level in log_level_count:
        log_level_count[level] += 1
    else:
        log_level_count[level] = 1

most_common_log_level=max(log_level_count, key=log_level_count.get)

print("log level_count: ",log_level_count)
print("most common log level: ",most_common_log_level)
print("*" * 15)

for log in logs:
    if log["level"] == "ERROR":
        print("Error logs: ",log)
print("*" * 15)
        
for log in logs:
    if log["level"] == "INFO":
        print("INFO logs: ",log["level"]) 
```

Question 7

Tasks

1. Count skills.

2. Most common skill.

3. Print employees knowing Python.

4. Print employees knowing Docker.

```python
employees = [
    {"name":"Ali","skills":["Python","Linux"]},
    {"name":"Sara","skills":["Python","Docker"]},
    {"name":"Ahmed","skills":["Linux","Docker"]},
    {"name":"John","skills":["Git","Python"]},
    {"name":"Fatima","skills":["Docker","Git"]}
]
```

```python
skill_count={}
for emp in employees:
    for skills in emp["skills"]:
        if skills in skill_count:
            skill_count[skills] += 1
        else:
            skill_count[skills] = 1
Most_common_skill = max(skill_count, key = skill_count.get)

print("skill count: ",skill_count)
print("Most common skill: ",Most_common_skill)
print("*" * 15)

for emp in employees:
    if "Python" in emp["skills"]:
        print("employees knowing Python: ",emp["name"])
print("*" * 15)
for emp in employees:
    if "Docker" in emp["skills"]:
        print("employees knowing Docker: ",emp["name"])
        
```
