Question 1

Tasks

1. Highest salary employee.


2. Lowest salary employee.


3. Count salaries.


4. Count departments.


5. Print employees earning above ₹50,000.


6. Print only IT employees.

```python
employees = [
    {"name":"Ali","salary":45000,"dept":"IT"},
    {"name":"Sara","salary":65000,"dept":"HR"},
    {"name":"Ahmed","salary":55000,"dept":"IT"},
    {"name":"John","salary":70000,"dept":"Finance"},
    {"name":"Fatima","salary":45000,"dept":"HR"}
]
```

```python
highest = employees[0]
lowest =employees[0]
salary_count={}
dept_count={}


for emp in employees:
    if emp["salary"] > highest["salary"]:
        highest = emp
    
    if emp['salary'] < lowest['salary']:
        lowest = emp
        
    salary = emp["salary"]
    if salary in salary_count:
        salary_count[salary] += 1
    else :
        salary_count[salary] =1

    dept =emp["dept"]
    if dept in dept_count:
        dept_count[dept] += 1
    else:
        dept_count[dept] = 1

print("Highest salary employee: ",highest["name"], highest["salary"])
print("Lowest salary employee: ",lowest["name"], lowest["salary"])
print("salaries Count: ",salary_count)
print("departments Count: ",dept_count)
print("*" * 15)

for emp in employees:
    if emp["salary"] > 50000:
        print("employees earning above ₹50,000: ",emp["name"])
print("*" * 15)
for emp in employees:
    if emp["dept"] =="IT":
        print("only IT employees: ",emp["name"])
```

```python
highest_marks = max(students, key = lambda student:student["marks"])
lowest_marks = min(students, key =lambda student:student["marks"])
for student in students:
    marks = student["marks"]
    marks_count[marks] = marks_count.get(marks,0)+1

    course=student["course"]
    course_count[course] = course_count.get(course,0)+1

print("Highest marks: ",highest_marks["name"], highest_marks["marks"])
print("Lowest marks: ",lowest_marks["name"], lowest_marks["marks"])
print("marks Count: ",marks_count)
print("course Count: ",course_count)
print("*" * 15)

for student in students:
    if student["marks"] > 85:
        print("marks above 85: ",student["name"])
print("*" * 15)
for student in students:
    if "Python" in student["course"]:
        print("python students: ",student["name"])
```

Tasks

1. Highest CPU server.


2. Lowest CPU server.


3. Count status.


4. Print running servers.


5. Print servers with CPU above 60.

```python
servers = [
    {"name":"web01","status":"Running","cpu":55},
    {"name":"db01","status":"Stopped","cpu":92},
    {"name":"app01","status":"Running","cpu":70},
    {"name":"cache01","status":"Stopped","cpu":40}
]
```

```python
highest_server = servers[0]
lowest_server = servers[0]
status_count={}

for server in servers:
    if server["cpu"] > highest_server["cpu"]:
        highest_server = server 
    if server["cpu"] < lowest_server["cpu"]:
        lowest_server = server
    
    status = server["status"]
    if status in status_count:
        status_count[status] += 1
    else:
        status_count[status] = 1

print("Highest server: ",highest_server["name"], highest_server["cpu"])
print("Lowest server: ",lowest_server["name"], lowest_server["cpu"])
print("status Count: ",status_count)
print("*" * 15)

for server in servers:
    if server["status"] == "Running":
        print("server status: ",server["name"],server["status"])
print("*" * 15)
for server in servers:
    if server["cpu"] > 60:
        print("server with cpu above 60: ",server["name"],server["cpu"])
```
