Question 1 – Highest Salary + Count Salary

Tasks:

1. Find the employee with the highest salary.


2. Count employees having the same salary.

```python
employees = [
    {"name":"Ali","salary":45000},
    {"name":"luke","salary":70000},
    {"name":"Sara","salary":55000},
    {"name":"Ahmed","salary":45000},
    {"name":"John","salary":70000},
    {"name":"Sara","salary":55000},
    {"name":"Eva","salary":55000}
]
```

```python
# method:1 , max
highest = employees[0]
for emp in employees:
    if emp["salary"] > highest["salary"]:
        highest = emp
highest
```

```python
# method: 2
highestt = max(employees, key = lambda x : x["salary"])
highestt
```

```python
# method: 3
highest_sal = max(emp["salary"] for emp in employees)
for emp in employees:
    if emp["salary"] == highest_sal:
        print(emp)
```

```python
# method 1, count
count_emp = {}
for emp in employees:
    salary = emp["salary"]
    if salary in count_emp:
        count_emp[salary] += 1
    else :
        count_emp[salary] = 1
count_emp
```

```python
# method 2, count
count_empl = {}
for emp in employees:
    salary = emp["salary"]
    count_empl[salary]=count_empl.get(salary,0)+1

count_empl
```

Question 3 – Countries

Tasks:

1. Count each country.


2. Which country appears the most?

```python
visitors = [
    {"country":"India"},
    {"country":"USA"},
    {"country":"India"},
    {"country":"UK"},
    {"country":"India"},
    {"country":"USA"}
]
```

1. count

```python
country_count ={}
```

```python
for visitor in visitors:
    country= visitor["country"]
    if country in country_count:
        country_count[country] += 1
    else: 
        country_count[country] = 1
country_count
```

```python
country_counts={}
for visitor in visitors:
    country= visitor["country"]
    country_counts[country]= country_counts.get(country,0)+1
country_counts
```

2. highest common country

```python
most_common = max(country_counts, key = country_counts.get)
most_common
```

```python
less_common = min(country_counts, key = country_counts.get)
less_common
```

Question 4 – Server CPU

Tasks:
1. Highest CPU server.


2. Lowest CPU server.

```python
servers = [
    {"name":"web01","cpu":65},
    {"name":"db01","cpu":90},
    {"name":"app01","cpu":72},
    {"name":"cache01","cpu":55}
]
```

1. manual

```python
highest = servers[0]
for server in servers:
    if server["cpu"] > highest["cpu"]:
        highest = server
highest
```

extra :  count

```python
highest = {}
for server in servers:
    maxi = server["cpu"]
    if maxi in highest:
        highest[maxi] += 1 
    else:
        highest[maxi] = 1
highest 
```

2. max, min, sorted methods

```python
highest_server = max(servers, key = lambda server : server["cpu"])
highest_server
```

```python
lowest_server = min(servers, key = lambda server : server["cpu"])
lowest_server
```

```python
highest_server = sorted(servers, key = lambda server : server["cpu"], reverse=True)
highest_server[0]
```

```python
lowest_server = sorted(servers, key = lambda server : server["cpu"])
lowest_server[0]
```

Question 5 – Products

Tasks:

1. Most expensive product.

```python
products = [
    {"name":"Laptop","price":60000},
    {"name":"Mouse","price":800},
    {"name":"Keyboard","price":1500},
    {"name":"Monitor","price":12000},
    {"name":"Mouse","price":800}
]
```

```python
most_expensive = max(products, key = lambda product : product["price"])
most_expensive
```

```python
less_expensives = sorted(products, key = lambda product : product["price"])
print("less_expensives: ",less_expensives)
most_expensives = sorted(products, key = lambda product : product["price"], reverse =True)
print("most_expensives: ",most_expensives)
```
