Question 1

Create a list of student names scoring above 80.

```python
students = [
    {"name":"Ali","marks":90},
    {"name":"Sara","marks":75},
    {"name":"Ahmed","marks":88},
    {"name":"John","marks":60}
]
```

```python
score = []
for student in students:
    if student["marks"] > 80:
        score.append(student["name"])
print("student with score above >80: ",score)
```

Question 2

Create a list of employee names earning above ₹50,000.


```python
employees = [
    {"name":"Ali","salary":45000},
    {"name":"Sara","salary":60000},
    {"name":"Ahmed","salary":70000},
    {"name":"John","salary":35000}
]
```

Question 4

Create a list of running server names.

```python
servers = [
    {"name":"web01","status":"Running"},
    {"name":"db01","status":"Stopped"},
    {"name":"app01","status":"Running"}
]
```

```python
running_servers=[]
for server in servers:
    if server["status"] == "Running":
        running_servers.append(server["name"])
print("running servers are: ",running_servers)
```

Question 5

Create a list of .py files.

```python
files = [
    "main.py",
    "notes.txt",
    "docker.yaml",
    "app.py",
    "config.json"
]
```

```python
py_files = []
yaml_file = []

for file in files:
    extension = file.split(".")[1]
    if extension == "py":
        py_files.append(file)


for file in files:  
    if file.endswith(".yaml"):
       yaml_file.append(file)      

print("".py" files are: ",py_files)
print(".yaml files are: ",yaml_file)
```
