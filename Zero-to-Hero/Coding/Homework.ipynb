Question 1

Print students scoring above 80.

```python
students = [
    {"name":"Ali","marks":90},
    {"name":"Sara","marks":75},
    {"name":"Ahmed","marks":88},
    {"name":"John","marks":60}
]
```

#### method -1

```python
for student in students:
    if student["marks"] > 80:
        print(student)
```

#### method -2

```python
for student in students:     
    if student["marks"] > 80:
        print(student["name"])
```

Question 2

Print employees earning more than ₹50,000.

```python
employees = [
    {"name":"Ali","salary":45000},
    {"name":"Sara","salary":60000},
    {"name":"Ahmed","salary":70000},
    {"name":"John","salary":35000}
]
```

```python
# method -1 
for emp in employees:
    if emp["salary"] > 50000:
        print(emp)

print("*"*15)

# method -2 
for emp in employees:
    if emp["salary"] > 50000:
        print(emp["name"],emp["salary"])
```

```python
files = [
    "main.py",
    "docker.yaml",
    "notes.txt",
    "app.py",
    "config.json"
]
```
