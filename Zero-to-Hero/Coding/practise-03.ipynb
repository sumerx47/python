9. Print employees knowing Python


10. Create a list of Python developers


11. Create a list of HR employees


12. Create a list of employees  earning below ₹50,000

```python
employees = [
    {"name":"Ali","dept":"IT","salary":50000,"skills":["Python","Linux"]},
    {"name":"Sara","dept":"HR","salary":65000,"skills":["Excel"]},
    {"name":"Ahmed","dept":"IT","salary":70000,"skills":["Python","Docker"]},
    {"name":"John","dept":"Finance","salary":45000,"skills":["Excel","PowerBI"]},
    {"name":"Fatima","dept":"IT","salary":60000,"skills":["Linux","Docker"]},
    {"name":"Aisha","dept":"HR","salary":52000,"skills":["Excel","Python"]}
]
```

```python
highest = max(employees, key = lambda emp:emp["salary"])
lowest= min(employees, key = lambda emp: emp["salary"])

Count_departments = {}
Count_salaries = {}
Count_skills = {}

employees_IT =[]
employees_SAL =[]
employees_PY=[]

Python_developers = []
employees_HR=[]
emp_less= []

for emp in employees:
    dept = emp["dept"]
    Count_departments[dept] = Count_departments.get(dept,0)+1

    salary = emp["salary"]
    Count_salaries[salary] = Count_salaries.get(salary,0)+1

    for skill in emp["skills"]:
        Count_skills[skill] = Count_skills.get(skill,0)+1

Most_common_skill = max(Count_skills, key = Count_skills.get)

for emp in employees:
    if emp["dept"] == "IT":
        employees_IT.append(emp["name"])

    if emp["salary"] > 50000:
        employees_SAL.append(emp["name"])

    skill = emp["skills"]
    if "Python" in skill:
        employees_PY.append(emp["name"])
    if "Python" in skill:
            Python_developers.append(emp["name"])

    if emp["dept"] == "HR":
         employees_HR.append(emp["name"])

    if emp["salary"] < 55000:
         emp_less.append(emp["name"])

print("Highest salary employee: ", highest)
print("Lowest salary employee: ", lowest)
print("Count of departments: ", Count_departments)
print("Count of salaries: ", Count_salaries)
print("Count of skills: ", Count_skills)
print("Most common skill: ", Most_common_skill)
print("IT employees: ", employees_IT)
print("employees earning above ₹55,000: ", employees_SAL)
print("employees knowing Python: ", employees_PY)
print("Python developers are: ", Python_developers)
print("HR employees are: ", employees_HR)
print("employees  earning below ₹55,000: ", emp_less)

```

🔴 Question 2 – Student Records


Tasks

1. Highest marks


2. Lowest marks


3. Count courses


4. Count marks


5. Print Python students


6. Print students scoring above 85


7. Create a list of Python student names


8. Create a list of students scoring below 75


9. Create a list of Java students


10. Create a list of marks above 80

```python
students = [
    {"name":"Ali","course":"Python","marks":90},
    {"name":"Sara","course":"Java","marks":82},
    {"name":"Ahmed","course":"Python","marks":95},
    {"name":"John","course":"C++","marks":68},
    {"name":"Fatima","course":"Python","marks":88},
    {"name":"Zaid","course":"Java","marks":72},
    {"name":"Mary","course":"Python","marks":79}
]
```

```python
Highest_marks = students[0]
Lowest_marks = students[0]

Count_courses = {}
Count_marks ={}


for student in students:
    if student["marks"] > Highest_marks["marks"]:
        Highest_marks = student

    if student["marks"] < Lowest_marks["marks"]:
        Lowest_marks = student

    course = student["course"]
    if course in Count_courses:
        Count_courses[course] +=1
    else:
        Count_courses[course] =1

    marks = student["marks"]
    if marks in Count_marks:
        Count_marks[marks] += 1
    else:
        Count_marks[marks] = 1

print("Highest marks: ", Highest_marks)
print("Lowest marks: ", Lowest_marks)
print("Count courses: ", Count_courses)
print("Count marks: ", Count_marks)


print("*"*15)
for student in students:
    if student["course"] == "Python":
        print("students with python: ",student["name"])
print("*"*15)
for student in students:
    if student["marks"] > 85:
        print("students with marks above 85: ",student["name"])
print("*"*15)

li_py=[]
li_score=[]
li_java=[]
li_marks=[]

for student in students:
    if "Python" in student["course"]:
        li_py.append(student["name"])

    if student["marks"] > 75:
        li_score.append(student["name"])

    if student["course"] == "Java":
        li_java.append(student["name"])

    if student["marks"] > 80:
        li_marks.append(student["name"])

print("Python student names: ", li_py)
print("students scoring below 75: ", li_score)
print("Java students: ", li_java)
print("marks above 80: ", li_marks)
```
