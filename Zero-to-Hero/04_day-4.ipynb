33. count vowels and consonents

```python
text = "hello guys how are you?,AEIOUA"
```

```python
len(text)
```

```python
vowels=0
consonents=0
```

```python
for i in text.lower():
    if i in "aeiou":
        vowels += 1
    else:
        consonents += 1
print(f'vowels: {vowels}\n consonents: {consonents}')
```
