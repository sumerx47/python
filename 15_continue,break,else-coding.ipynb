# Continue, Break & Else Coding

```python
for i in range(100):
  print(i)
  if i>=5: 
    print(f"Breaking out of Loop at: {i}")
    break
print('done')
```

```python
for i in range(1,11):
    if i % 2 == 1:
        continue
    print(i)
```

```python
for i in range(1,5):
    for j in range(1,5):
        if (i+j)%2 == 1:
            print(f"{i}+{j} =",(i)+(j),"is odd ,skipping number")
        else:
            print(f"{i}+{j} =", (i)+(j))
        
```

```python
for i in range(1,4):
    for j in range(1,4):
        if j>= 3:
            break
        print(i,j)
    print('*'*10)
```
