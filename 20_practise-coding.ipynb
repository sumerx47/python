#   Q5

```python
text="Python @2026DevOps "
```

```python
vowels=0
consonents=0
digits=0
specialcharacters=0
spaces=0
```

```python
for i in text:
    if i.lower() in "a,e,i,o,u":
     vowels+= 1
    elif i.isdigit():
       digits+= 1
    elif i.isspace():
       spaces += 1
    elif i.isalpha():
       consonents += 1
    else:
       specialcharacters += 1
print("vowels:", vowels)
print("consonents:", consonents)
print("digits:", digits)
print("specialcharacters:", specialcharacters)
print("spaces:", spaces)
          
```

```python
data = [
    [10,20],
    [20,30],
    [35,50],
    [45,60]
    ]
```

```python
for index,(openn,closee) in enumerate(data):
    if index==0 :
        data[index].append("None")
    else:
        prevclose = data[index-1][1]
        if openn > prevclose:
            data[index].append("up")
        elif openn < prevclose:
            data[index].append("down")
        else:
            data[index].append("same")
print(data)
```
