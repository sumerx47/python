# Iterating Dictionaries

```python
d={
    'key 1':1,
    'key 2':2,
    3.14:'pt'
}
```

```python
for k in d:
    print(k)
```

```python
for k in d:
    print(f"d[{k}] = {d[k]}")
```

```python
for t in d.items():
    k,v=t
    print(k,v)
```

```python
for k,v in d.items():
    print(k,v)
```

```python
for k,v in d.items():
    print(f"d[{k}]: {v}")
```
